## 输入与几何资格
a=AC=CB，b=BD，均严格正；u/v为等长正交基。坐标是教学模型，有限核查不是一般证明。

```python
from fractions import Fraction as Q
from geometry import construct, verify
c = construct()
r = verify(c)
print("a=3, b=2; A-C-B-D")
print(c["points"])
```

## 三块矩形等积
原论证先由I.36比较AL与CH，再由I.43比较CH与HF。

```python
for name in ("AL", "CH", "HF"):
    print(name, r[name])
```

## 曲尺形与补方
AL加CM组成AM；HF加CM组成曲尺形。再补LG为整平方。

```python
print("AM", r["AM"], "gnomon", r["gnomon"])
print(r["AM"], "+", r["LG"], "=", r["square"])
```

## 精确分数
不用浮点容差，有理输入仍为精确值。

```python
z = verify(construct(Q(2,3), Q(5,7)), Q(2,3), Q(5,7))
print(z["AM"], "+", z["LG"], "=", z["square"])
```

## 拒绝伪证书
只满足终式不够，H还须是指定对角线与竖线的交点。

```python
bad = construct()
bad["points"]["H"] = (Q(6),Q(3))
try:
    verify(bad)
except ValueError as e:
    print("rejected:", str(e))
else:
    raise RuntimeError("tampering accepted")
```
