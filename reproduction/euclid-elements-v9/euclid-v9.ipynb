# Euclid V.9: 用反证见证排除不等

本例以正有理数模拟整倍量比较；有界搜索没有找到反例，不是全称断言。


```python
from equality_probes import Magnitude as M, probe, distinguish, bounded_mismatch
a, b, c = map(M, (102, 101, 100))
print('有限窗口 20：', bounded_mismatch(a,b,c,20))
print('扩大到 51：', bounded_mismatch(a,b,c,51))
```

```python
for left,right in ((a,b), (b,a)):
    cert = distinguish(left,right,c)
    print(cert)
    if len(set(cert.forward)) != 2 or len(set(cert.reverse)) != 2:
        raise RuntimeError('missing mismatch')
```

```python
for m,n in ((1,2),(2,2),(3,2)):
    print((m,n), probe(c,c,c,m,n))
print('相等时无区分证书：', distinguish(c,c,c))
```
