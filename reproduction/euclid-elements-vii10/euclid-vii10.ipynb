## 精确分数独立检查

两组分别走 VII.5 和 VII.6 的汇合分支。

```python
for ab,c,de,f in ((8,12,16,24),(8,12,20,30)):
    p = compare(ab,c,de,f)
    r = Fraction(ab,de)
    if r != Fraction(c,f) or r != Fraction(*p.comparison):
        raise ValueError('unexpected ratio')
    print(p.aggregation, r)
```

## 打破共同份数

只改 F，结论不再成立。

```python
try:
    compare(8,12,20,32)
except ValueError as exc:
    print('rejected:', exc, '; ratios:', Fraction(8,20), Fraction(12,32))
else:
    raise RuntimeError('unexpected acceptance')
```
