## 1. 同一个非夹角和同比两边，可以产生两形

```python
from fractions import Fraction as Q
from ratio_chains import ambiguity, compare, normalized_model, signature, sweep, transform

acute, obtuse = ambiguity()
print("C 向量夹角的余切：", signature(acute)[2], signature(obtuse)[2])
print("角相同、同比、同支、等角、边同比：", compare(acute, obtuse))
print("同一钝角支的有理缩放反射：", compare(obtuse, transform(obtuse, Q(3, 2), True, (2, 3))))
```
