正有理数模型。运行目录需能导入 ratio_witness.py。

```python
from ratio_witness import transport, is_witness, find_witness, exact_witness, compare
first, middle, last = (3, 2), (6, 4), (2, 2)
for probe in [(1, 1), (4, 5)]:
    print(probe, transport(first, middle, last, probe))
```

```python
print('wrong direction:', is_witness(last, middle, (1, 1)))
print('upper endpoint:', is_witness(middle, last, (2, 3)))
```

```python
high, low = (102, 100), (101, 100)
print('cap 20:', find_witness(high, low, 20))
print('cap 60:', find_witness(high, low, 60))
probe = exact_witness(high, low)
print('exact:', probe, compare(high, probe), compare(low, probe))
```
