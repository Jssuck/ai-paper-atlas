# III.19：接点法线是一条过心直线
原创有理坐标教学；标准库 compile/exec 实跑，不是 Jupyter kernel。一般证明见正文，有限测试不是证明。

```python
from geometry import *
a=((0,0),(3,4),5,(-4,3),(-3,-4))
c=certify(*a)
print(c)
print('valid',verify(*a,c))
```

## 反向参数不是失败
把方向改成向外后，圆心在负参数处。因此整线过心，正半射线不过心。

```python
out=certify(*a[:-1],(3,4))
print('parameter',out['t'],'positive ray reaches centre',out['ray_reaches_centre'])
```

## 不预设任意过接点线都相切
非切线输入被代数条件拒绝，该条件的全称意义由正文配方论证承担。

```python
try:
    certify(*a[:3],(1,0),(0,1))
except ValueError as e:
    print('rejected:',e)
```

## 证书可被检查和拒绝

```python
bad=dict(c);bad['t']=F(2)
print('tampered accepted',verify(*a,bad))
```
