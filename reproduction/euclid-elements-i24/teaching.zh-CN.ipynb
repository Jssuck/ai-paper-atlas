## 1 固定两边，增大夹角
t 是正半角参数；不使用浮点角度。

```python
c=certify(case())
print({k:c[k] for k in ('EF_squared','EG_squared','gap')})
```

## 2 三种配置
同一夹角和 b=1，改变 a。分类对应 Heath 指出的原图外、共线、内三型。

```python
for a in (1,2,3):
 c=certify(case(a=a)); print(a,c['configuration'],c['EFG_determinant'])
```

## 3 反射与有限网格
反射不改变大小；有限证书不是一般定理的证明。

```python
cs=[certify(d) for d in positives()]
print('verified',sum(verify(d,c) for d,c in zip(positives(),cs)))
```

## 4 证书不能接受 bool 冒充整数
不只比较 Python 数值相等。

```python
c=certify(case()); c['EFG_orientation']=True
print('tampered accepted',verify(case(),c))
if verify(case(),c): raise RuntimeError('tamper accepted')
```

## 5 面积不是本命题结论

```python
print('heights',unit(Q(2))[1],unit(Q(3))[1])
print('base squared gap',certify(case(s=2,t=3))['gap'])
```
