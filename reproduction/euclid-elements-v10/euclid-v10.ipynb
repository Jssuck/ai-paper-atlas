# Euclid V.10：由较大的比推出量的顺序

用正有理数模拟 V.Def.7 的见证。窗口中没有见证不会给出结论。


```python
from order_witness import Magnitude as M, SHARED_DENOMINATOR as D, SHARED_NUMERATOR as N, infer_order, bounded_search
a,b,c = map(M,(3,2,2))
print('共同后项，有效弱见证：', infer_order(a,b,c,1,1,D))
print('同一对整数不能直接用于另一方向：', infer_order(a,b,c,1,1,N))
print('共同前项，另一有效弱见证：', infer_order(a,b,c,3,2,N))
```

```python
a,b,c = map(M,(7,4,4))
for o,m,n in ((D,2,3),(N,3,2)):
    cert = infer_order(a,b,c,m,n,o)
    if cert is None or cert.conclusion != 'A>B':
        raise RuntimeError('missing conclusion')
    print(o, cert)
```

```python
a,b,c = map(M,(102,101,100))
for o in (D,N):
    print(o, '窗口 20：', bounded_search(a,b,c,20,o))
    print(o, '窗口 52：', bounded_search(a,b,c,52,o))
```
