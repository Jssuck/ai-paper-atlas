# 《几何原本》VII.4：一份与若干份

```python
from part_or_parts import classify,divisor_oracle
for a,b in ((5,8),(4,12),(6,15)):
    print((a,b),classify(a,b,ancient_domain=True))
```

更小的共同量数可给出更细的分组；最大共同量数给出最简的分组。

```python
a,b=12,18
print("所有共同量数",divisor_oracle(a,b))
print("各单位的份数",[(a//d,b//d) for d in divisor_oracle(a,b)])
print("最简表示",classify(a,b))
```

一个单位作为整个输入是现代扩展，不能与互素证明中出现单位混同。

```python
print("现代处理",classify(1,8))
try:
    classify(1,8,ancient_domain=True)
except ValueError as exc:
    print("古域拒绝",exc)
```
