# 《几何原本》IV.7：用切线外接正方形
用分数坐标替代尺规。给定圆心和非零半径向量，在四个四分之一转的接触点作切线。依次联立相邻切线方程得到四个顶点。

```python
from fractions import Fraction as F
from circumscribed_square import circumscribe, tangent_quad, formatted, measurements, verify, run_grid

o, u = (F(1, 2), F(-1, 3)), (2, 1)
vertices = circumscribe(o, u)
print("vertices:", formatted(vertices))
print("side^2, adjacency dot, diagonal^2, signed area:", measurements(vertices))
print("certificate:", verify(o, u, vertices))
```

## 去掉垂直条件
保留四条切线与两组等长、相反的半径向量。预期得到四边等长但相邻边不垂直的菱形。

```python
control = tangent_quad((0, 0), (5, 0), (3, 4))
print("vertices:", formatted(control))
print("side^2, adjacency dot, diagonal^2, signed area:", measurements(control))
print("square certificate:", verify((0, 0), (5, 0), control))
```
