## V.21：错序对应与端点次序

在 Notebook 所在目录依次运行三个代码单元。正有理数结果用于教学，不替代一般证明。

```python
from ratio_chains import chain, verify_v21

left = chain((6, 4, 3), "长度")
right = chain((8, 6, 4), "面积")
trace = verify_v21(left, right)
print(f"左组 A:B, C:B = {trace.left_ratio}, {trace.left_reversed_ratio}")
print(f"右组 E:F, E:D = {trace.right_ratio}, {trace.right_reversed_ratio}")
print(f"共同端点次序：{trace.sign:+d}")
```

### 等于与小于

右组同一前项下，比越大，后项越小；输出始终比较 D 对 F。

```python
for left_values, right_values in (((6, 4, 6), (8, 12, 8)), ((3, 4, 6), (4, 6, 8))):
    trace = verify_v21(chain(left_values, "长度"), chain(right_values, "面积"))
    print(f"{left_values} / {right_values}: {trace.sign:+d}")
```

### 前提拒绝示例

头两例各只保留一对错序比例，且端点次序翻转。最后一例是 V.20 的合法顺配，但不是 V.21 的合法错配，端点次序没有翻转。

```python
for right_values in ((3, 6, 4), (8, 6, 10), (12, 8, 6)):
    try:
        verify_v21(left, chain(right_values, "面积"))
    except ValueError as exc:
        print(f"{right_values}: {exc}")
    else:
        raise RuntimeError("expected perturbed group mismatch")
```
