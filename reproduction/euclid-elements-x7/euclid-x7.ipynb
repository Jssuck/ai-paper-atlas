以平方差的符号作精确比较，另列正对照。

```python
from fractions import Fraction as F
from ratio_witness import compare_ratio
for m,n in ((1,1),(3,2),(99,70)):
    print((m,n),compare_ratio(2,1,m,n))
```

```python
print('Exact positive control:',compare_ratio(F(9,4),1,3,2))
print('Scaled approximation:',compare_ratio(8,2,99,70))
```

```python
from itertools import product
values=[compare_ratio(2,1,m,n).sign for m,n in product(range(1,129),repeat=2)]
print('Bounded comparisons and equalities:',len(values),values.count(0))
print('A bounded search is not an irrationality proof.')
```
