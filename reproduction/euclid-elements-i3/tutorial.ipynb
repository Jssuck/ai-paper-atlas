# 《几何原本》I.3：精确坐标证书，不是尺规引擎

目标：从较长的 AB 中截取 AE，使 AE 等于较短的给定长度。
这里 **C 是目标长度的平方**，要求 `0 < C < |AB|²`。
先在本目录运行；仅用 Python 标准库。我们计算一个解析见证，再独立检查几何关系。
`D` 是外部给定的半径端点；检查 `|AD|²=C` 不等于重演或证明 I.2。


```python
from fractions import Fraction as F
from euclid_i3 import Point, Certificate, analytic_witness, verify, domain_status
from experiments import mutations, run_families

a, b, C = Point(0, 0), Point(5, 0), F(4)
cert = analytic_witness(a, b, C)
print("E =", (str(cert.e.x), str(cert.e.y)))
print("全部条件通过:", verify(a, b, C, cert).valid)
print("注意：D 是解析提供的点，不是 I.2 的尺规构造记录。")
```

## 为什么“距离相等”不够？

令 `v=B-A`、`w=E-A`。必须同时检查：
- `cross(v,w)=0`：点在线上；
- `0 < dot(v,w) < dot(v,v)`：结合共线性，点在开线段中；
- `|AE|²=|AD|²=C`：半径及目标长度正确。

反向交点会通过长度与共线性检查；同距离的离线点甚至可以通过严格点积界限。
等长、过长、零长和退化 AB 都不是本模型的严格 I.3 输入。


```python
for label, mutant in mutations(a, b, cert).items():
    result = verify(a, b, C, mutant)
    print(label, "拒绝:", ", ".join(result.failures))
for target in (F(0), F(25), F(36)):
    print("C =", target, "→", domain_status(a, b, target))
print("A=B →", domain_status(a, a, C))
# 验证器比解析见证生成器更一般：两条长度可为无理数。
p = Point(F(1, 2), F(1, 2))
print("有理坐标、无理长度证书:", verify(a, Point(1, 1), F(1, 2), Certificate(p, p)).valid)
```
