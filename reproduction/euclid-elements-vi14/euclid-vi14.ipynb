# 欧几里得 VI.14：共用参照图形
适合知道平行四边形面积与分数的读者。目标：将古代比例证明映射到能复核的坐标模型。先看共用面积，再去掉等角假设，最后检验一次线性变换。

```python
from fractions import Fraction as Q
from ratio_chains import reconstruct
r = reconstruct(6, 4, 2, 3)
print("AB, BC, FE:", *(str(v) for v in (r.area_ab, r.area_bc, r.area_fe)))
print("AB:FE, BC:FE:", str(r.area_ab / r.area_fe), str(r.area_bc / r.area_fe))
print("Equal areas:", r.equal_areas, "; reciprocal sides:", r.reciprocal_sides)
```

只替换第二个平行四边形的夹角，保持其两边长度。AB 仍使用直角，BC 使用下格的倾斜方向。问：四个边长因子不变时，面积会不会改变？

```python
tilt = reconstruct(6, 4, 2, 3, v=(Q(3, 5), Q(4, 5)))
print("Mixed angles:", str(r.area_ab), str(tilt.area_bc))
print("Ratio terms for mixed angles:", str(r.ratio_db_be), str(tilt.ratio_gb_bf))
```

练习：若三个图形一起使用 u=(2,1)、v=(1,-2) 的新方向，面积因子和两组比各怎样变？答案：面积乘 5，比不变。注意此时系数不再等于边长，但同一方向内的边长比不变。

```python
oblique = reconstruct(6, 4, 2, 3, (2, 1), (1, -2))
print("Area scaling:", str(oblique.area_ab / r.area_ab),
      str(oblique.area_bc / r.area_bc), str(oblique.area_fe / r.area_fe))
print("Ratios unchanged:", (r.ratio_db_be, r.ratio_gb_bf) ==
      (oblique.ratio_db_be, oblique.ratio_gb_bf))
```
