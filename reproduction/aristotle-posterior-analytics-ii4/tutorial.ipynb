任取中项成员，先落入A，再由A=C落入C，所以中项与端点相等。程序只检查有限实例，任意集合结论靠这个元素论证。

## 3. 相同集合，不同声明
full_definition标签在本模型中至少要求外延相等，但相等不会生成这个标签；essential_part表示组成条件而非完整定义。

```python
ext = {'C': {0}, 'B': {0}, 'A': {0}}
for role in ('attribute', 'essential_part', 'full_definition'):
    minor = Premise('C', 'B', role, ('minor',))
    major = Premise('B', 'A', role, ('major',))
    result = compose(minor, major, ext)
    print(role, result['declared_result_role'], result['minor_already_declares_subject_definition'], result['actual_essence_certified'])
```

## 4. 显式支持来源
有目标自身标签即报告显式依赖；没有该标签不能认证认识上独立，也不能发现未登记的同义改写。这不是一般乞题判定器。

```python
minor = Premise('C', 'B', 'full_definition', ('target:C-definition',))
major = Premise('B', 'A', 'full_definition', ('major',))
print(compose(minor, major, ext))
```

## 练习与答案
把第三单元的小前提改成attribute、大前提保留full_definition，结果是什么？答案：attribute。要获取完整定义标签，不能仅加强大前提。

把所有集合换为空集，是否证明灵魂或人的本质？答案：否；这里只留下空外延的形式真值，未作存在或解释认证。
