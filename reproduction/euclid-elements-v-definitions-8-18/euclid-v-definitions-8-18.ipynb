## 1. 连续比例

按相邻项的相同比值检验前提，再读取首末项之比。比较平方和普通倍数。

```python
from ratio_ops import Ratio, terms, power_ratio, transform, endpoint_ratios, grid
a = terms(27, 18, 12, 8)
u = Ratio(*a[:2]).value
print("基比、二重比、三重比、普通双倍:", *(str(x) for x in (u, power_ratio(a[:3]).value, power_ratio(a).value, 2*u)))
```

## 2. 并比、分比、倒比和反成比

对同一输入逐一计算。后一变换不是倒比；涉及差量时必须为正。

```python
r = Ratio(*terms(3, 2))
for op in ("composition", "separation", "inverse", "conversion"):
    print(op, transform(r, op).value)
```

## 3. 依次与换序连接

先用局部比值决定配对，不从首末恰好相等倒推合法性。结束后运行小规模精确对照。

```python
left = terms(12, 6, 2)
for right, swapped in ((terms(18, 9, 3, kind="area"), False), (terms(18, 6, 3, kind="area"), True)):
    print("换序:", swapped, "| 端点比:", tuple(str(r.value) for r in endpoint_ratios(left, right, perturbed=swapped)))
    try:
        endpoint_ratios(left, right, perturbed=not swapped)
    except ValueError as exc:
        print("错误配对:", exc)
    else:
        raise RuntimeError("错误配对应拒绝")
print("有效连接和错配拒绝:", grid())
```
