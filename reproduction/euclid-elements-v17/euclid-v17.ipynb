目的：用精确有理数检查比较移位和消去。先让各比内部同类，组与组可不同类。

## 追踪等号探针

目标探针取 p=3、q=2；已知合比使用 (3,5)。每组的共同量各自消去，绝非跨组减同一个绝对量。

```python
from ratio_separation import replay

left = tuple(map(Magnitude, (2, 3)))
right = tuple(Magnitude(v, "area") for v in (4, 6))
r = replay(*left, *right, 3, 2)
print("whole signs:", r.whole_signs)
print("canceled terms:", tuple(str(x.value) + " " + x.kind for x in r.canceled))
print("remainder signs:", r.remainder_signs)
```

## 负例：两个整比不是同比

只有一组输出比较相等，不足以推出另一组也相等。下面刻意破坏整比前提。

```python
try:
    replay(*left, Magnitude(5, "area"), right[1], 3, 2)
except ValueError as exc:
    print("rejected:", exc)
else:
    raise RuntimeError("expected hypothesis rejection")
```
