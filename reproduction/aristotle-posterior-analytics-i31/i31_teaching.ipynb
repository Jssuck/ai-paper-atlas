# I.31：看过若干次，为什么还不等于证明？

原创有限教学模型，不是亚里士多德原文形式化，也不进行玻璃、光线、月食物理实验。读者需要 Python 基础。目标是区分观察约束、有限全覆盖与因果解释。所有输出由随附纯 Python 顺序执行器真实生成，不宣称使用 Jupyter 内核。

```python
from model import report, compatible_models, mechanism
import json
print(json.dumps(report(4, [(0, True), (1, True)]), ensure_ascii=False))
```

## 1. 未见位置留下反模型
4 个位置只看过 2 个，未看位置可独立真假选择，因此有 4 个相容赋值。全真只是其中之一。这个结果依赖选择了全部 Boolean 函数作为假设类。

```python
once = compatible_models(4, [(0, True), (1, True)])
repeated = compatible_models(4, [(0, True), (1, True)] * 10)
print("重复不增加约束:", once == repeated)
print("全覆盖:", json.dumps(report(4, [(i, True) for i in range(4)]), ensure_ascii=False))
```

## 2. 有限全覆盖不是原因
这里只认证预先封闭的 4 个位置，没有推出其他位置、未来时间或自然律。无限域的外推还需论证。下一步展示两个人为规定的机制，在观察限制 x=z 下完全相同。干预保持 z=False 而令 x=True，是教学设定，并非从观察识别出来的干预规律。

```python
for v in (False, True):
    print("观察 x=z=", v, "机制X:", mechanism("X", v, v), "机制Z:", mechanism("Z", v, v))
print("规定的干预 (x=True,z=False):", mechanism("X", True, False), mechanism("Z", True, False))
```

## 3. 矛盾不是证实
如果同一位置同时记录真和假，在本确定性无噪声模型下相容集合为空，程序报告不一致，不以空集上的真值当作经验确认。现实噪声模型需要另加假设。

```python
print(json.dumps(report(3, [(0, True), (0, False)]), ensure_ascii=False))
```
