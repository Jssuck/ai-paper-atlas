## 正反传递：公度 3 的量度次数

```python
counts = (to_sum(6,9,3),from_sum(6,9,3),from_sum(9,6,3))
if counts != (5,3,2):
    raise RuntimeError('wrong counts')
print(counts)
```

## 仅量度一个加数还不够

```python
try:
    from_sum(8,15,2)
except ValueError as exc:
    print('rejected:',exc)
else:
    raise RuntimeError('unexpected acceptance')
```
