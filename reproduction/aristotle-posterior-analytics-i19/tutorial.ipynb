# 《后分析篇》I.19：有限链与未关闭的问题

这是现代外延教学重构，不是古代自然谓述的唯一语义。运行本Notebook不能证明无限链终止，也不能认证科学直接性。所有对象集合非空。

```python
from chains import middles, strict_chain, equivalence_classes, dyadic_prefix, experiment
terms={"C":frozenset({0}),"A":frozenset({0,1,2})}
print("原词表中项:",middles(terms,"C","A"))
terms["B"]=frozenset({0,1})
print("扩充词表中项:",middles(terms,"C","A"))
print("链:",strict_chain(terms,["C","B","A"]))
```

## 同外延名字不是两个严格层级

合并同外延类只是集合模型中的操作，不能判断人和白物之间自然/偶然谓述的方向。

```python
terms["B_alias"]=terms["B"]
print("名字:",middles(terms,"C","A"))
print("外延商类:",equivalence_classes(terms))
```

## 两端固定，仍可继续插入

精确有理中点总在两端之间。这里仅执行8次：一般可再插入是由不等式证明，而不是由8次输出证明。有限子集包含模型与有理数序模型是不同模型。

```python
points=dyadic_prefix(8)
print("点:",[str(x) for x in points])
print("剩余上端间隙:",str(points[-1]-points[-2]))
```
