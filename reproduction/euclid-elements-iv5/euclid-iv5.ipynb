适合了解点积、垂直平分线和 Python 分数的读者。请在本目录启动 Notebook。
1. 比较锐角、直角、钝角三角形的外心。
2. 检查直角换标记后，外心仍在有限斜边中点。
3. 比较近共线与完全共线，并重放有限网格。

```python
from fractions import Fraction as Q
from circumcircle import circumcircle, grid

def show(triangle):
    result = circumcircle(*triangle)
    print(result.angle_type, result.location,
          tuple(map(str, result.center)), str(result.radius_squared))
    return result

examples = (
    ((0, 0), (4, 0), (1, 3)),
    ((0, 0), (4, 0), (0, 3)),
    ((0, 0), (4, 0), (1, 1)),
)
for triangle in examples:
    _ = show(triangle)
```

练习：把直角顶点标记成 B，再标记成 C。外心的位置是否跟着改变？先写下答案，再运行下一格。注意斜边是直角顶点对面的边，不总叫 BC。

```python
right = (0, 0)
for triangle in (((1, 0), right, (0, 1)), ((1, 0), (0, 1), right)):
    result = show(triangle)
    print("right vertex:", "ABC"[result.right_vertex],
          "| weights:", tuple(map(str, result.barycentric)))
    if result.center != (Q(1, 2), Q(1, 2)) or result.radius_squared != Q(1, 2):
        raise AssertionError("relabeled circle mismatch")
```

直角例的半径是无理数，但半径平方是精确分数。带噪测量可能难以分辨近共线与真共线；这个练习只讨论所给坐标值，不施加经验阈值。可进一步缩小第三点的高度，观察外心越移越远。

```python
_ = show(((0, 0), (2, 0), (1, Q(1, 100))))
try:
    circumcircle((0, 0), (2, 0), (1, 0))
except ValueError as error:
    print("rejected:", error)
else:
    raise AssertionError("collinear input not rejected")

counts, right_labels = grid()
print(counts)
print("right-label counts (A, B, C):", right_labels)
```
