# 四问不是一个问题

《后分析篇》II.1的现代教学账本。仅验证输入与问题类型的管理，不判断证据真伪、原因、本质或古代认识论。标准库顺序真实执行，不是Jupyter内核。

```python
from model import InquiryKey, Ledger, Status, exhaustive_check
ledger = Ledger()
eclipse = InquiryKey('attribute', 'sun', 'illustrative-event', 'eclipsed')
print('初始：', ledger.next_question(eclipse))
ledger.record(eclipse, Status.YES, 'synthetic observation')
print('事实肯定后：', ledger.next_question(eclipse))
print('已认证原因：', ledger.certified_answer(eclipse))
```

## 一份事实可搭配多个候选
候选字符串只表示有提案，不保证它是真正原因。

```python
ledger.propose(eclipse, 'candidate A')
ledger.propose(eclipse, 'candidate B')
print('候选：', ledger.candidates(eclipse))
print('已认证原因：', ledger.certified_answer(eclipse))
other = InquiryKey('attribute', 'sun', 'different-event', 'eclipsed')
print('换事件：', ledger.next_question(other))
```

## 理解名称并未证明实存
人马名称与假想研究不被禁止，但本账本不会凭描述声称它在现实中存在。

```python
centaur = InquiryKey('existence', 'centaur', 'actual-world')
ledger.propose(centaur, 'nominal description')
print('仅描述：', ledger.next_question(centaur))
ledger.record(centaur, Status.NO, 'hypothetical negative verdict, not empirical proof')
print('显式否定：', ledger.next_question(centaur))
ledger.withdraw(centaur)
print('撤回：', ledger.next_question(centaur))
```
