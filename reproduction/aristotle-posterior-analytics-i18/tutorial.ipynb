# 观察信息足够吗？

后分析篇 I.18 的现代有限教学对照。学习目标：区分观察保留信息、额外背景和有限样本。不是亚里士多德归纳的算法，也不是人的感官或学习能力实验。请从本目录运行。

```python
from observability import decoder, completions, run
x, y = (0, 0, 1, 1), (0, 1, 0, 1)
print("只见 x，求 y:", decoder(x, y))
print("替代通道直接给 y:", decoder(y, y))
```

`None` 表示同一观察类内目标冲突，不表示计算慢。添加替代通道改变了可用信息；这不是由旧通道凭空推得。

```python
print("额外假设 x=y 后:", decoder((0, 1), (0, 1)))
print("保留形状，忽略材料:", decoder(x, x))
```

背景收缩与抽象分别改变可能世界集合和目标。它们都需要明确说明；本例不会证明现实中所有相关背景假设为真。

```python
print("只检查第0项为1后的两个候选:", completions((1, 1), (0,)))
print("空域:", decoder((), ()))
```
