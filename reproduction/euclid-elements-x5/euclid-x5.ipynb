# X.5：公度量给出整数比
用精确计数核对三种等倍比较。

```python
from fractions import Fraction as F
from ratio_witness import ratio_witness
a,b,d=F(21,10),F(9,10),F(3,10)
r=ratio_witness(a,b,d)
print('Witness:',r)
```

```python
for u,v in ((1,3),(3,7),(1,2)):
    print((u,v),'signs:',r.compare(u,v))
```

```python
print('Subdivided:',ratio_witness(a,b,d/4))
print('Scaled:',ratio_witness(13*a,13*b,13*d))
try:
    ratio_witness(a,b,F(1,5))
except ValueError as exc:
    print('Rejected incompatible measure:',exc)
else:
    raise RuntimeError('Expected rejection')
```
