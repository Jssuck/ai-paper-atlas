# X.3：求最大公度量
从正向余量到逆向整数计数。

```python
from fractions import Fraction as F
from greatest_common_measure import greatest_common_measure,common_measure_at,is_common_measure
r=greatest_common_measure(F(21,10),F(9,10))
print('Delayed divisions:',r.divisions)
print('Backward counts:',r.backward_counts)
```

```python
print('Direct:',greatest_common_measure(12,3))
print('Equal:',greatest_common_measure(3,3))
```

```python
print('Common measures:',[str(common_measure_at(r,i)) for i in range(1,5)])
print('1/5 is common?',is_common_measure(r,F(1,5)))
```
