# Euclid I.8：三边数据如何产生可核验的等距证书

这是现代有理坐标教学重构，不是原文叠合证明，也不把有限测试当作一般定理。原文结论是相等边所夹角相等。请从本 Notebook 所在目录顺序执行；只需 Python 标准库。

```python
from fractions import Fraction as F
from geometry import *
S=((0,0),(3,0),(1,2))
R=((5,-2),(5,1),(3,-1))
print("AB², AC², BC²:", sides_squared(S))
```

## 1. 极化恒等式
设 u=B−A，v=C−A。由 ‖u−v‖²=‖u‖²+‖v‖²−2u·v 得到内积。固定对应不可省略。

```python
print("Gram:", gram(S))
print("从三边恢复:", gram_from_sides(sides_squared(S)))
print("顶角 (余弦符号, 余弦平方):", angle_signature(S))
```

## 2. 显式映射而不是 SSS 黑箱
非零面积保证 X=[u v] 可逆。计算 M=YX⁻¹，平移 t=A′−MA。Gram 相等在一般代数层面推出 MᵀM=I；下面只核验一个具体证书。

```python
m,t=certificate(S,R)
print("M =",m,"t =",t)
print("正交并逐点映射:",verify_certificate(S,R,m,t))
print("det(M) =",determinant(m))
```

## 3. 反射也保距
反射行列式为 −1；只允许平移与旋转会错误排除镜像。错配对应与退化则属于不同问题。

```python
H=((0,0),(3,0),(1,-2))
print("镜像 det:",determinant(certificate(S,H)[0]))
for target in [(S[0],S[2],S[1]), ((0,0),(1,0),(2,0))]:
    try: certificate(S,target)
    except ValueError as error: print(type(error).__name__,str(error))
```
