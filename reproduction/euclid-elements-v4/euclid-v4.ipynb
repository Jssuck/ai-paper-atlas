# 《几何原本》V.4：同比保持与探针换标

目标：区分保持两组同比、保持比值、保持同一编号的符号。

## 1. 匹配换标

原同比为 3:2 与 6:4，前项取 2 倍、后项取 1 倍。变换后比较 (p,q) 对应原来的 (2p,q)。

```python
from scaled_ratios import ratio, equal_oracle, mapped_probe, mismatch, regression_grid
left, right = ratio(3,2), ratio(6,4,"area")
print("原同比、新同比:", equal_oracle(left,right), equal_oracle(left.scale(2,1),right.scale(2,1)))
print("固定编号 (1,2) 的变前、变后符号:", left.sign(1,2), left.scale(2,1).sign(1,2))
for p,q in ((1,2),(1,3),(1,4)):
    print((p,q), "变换后与换标后:", mapped_probe(left,2,1,p,q))
```

## 2. 给对应前项使用不同的倍数

只放大左前项，不保证同比。四元组依次是两探针编号、左右符号。

```python
print("不同倍数的见证:", mismatch(left.scale(2,1),right,4))
```

## 3. 有限区间内无反例，不等于同比

同一有理数域中的交叉乘积用作独立于枚举的精确对照。

```python
a,b=ratio(101,100),ratio(102,101)
print("精确同比:", equal_oracle(a,b))
print("上界 100 与 101:", mismatch(a,b,100),mismatch(a,b,101))
print("网格（组合、四符号核验次数）:", regression_grid())
```

## 局限
