## 1 等底与同平行线

```python
from geometry import make_certificate, verify_certificate
from cases import EXAMPLE, build, positives, negatives
c = make_certificate(EXAMPLE)
print(c['area'])
```

## 3 支持线上的位移
重叠、相接、分离与重合均允许。

```python
for t in [0, '1/2', 1, 2, -2]:
    q = make_certificate(build(t))
    print(t, q['bridge']['lower_case'], q['area'])
```

## 4 几何反例
更改上支持线不能继续用定理。

```python
bad = dict(EXAMPLE)
bad['E'] = [8,4]
try:
    make_certificate(bad)
except ValueError as e:
    print('直接拒绝:', str(e))
else:
    raise RuntimeError('negative accepted')
```

## 5 有限精确输入族

```python
n = 0
for d in positives():
    if not verify_certificate(d, make_certificate(d)):
        raise RuntimeError('failed')
    n += 1
m = 0
for d in negatives():
    try:
        make_certificate(d)
    except ValueError:
        m += 1
    else:
        raise RuntimeError('negative accepted')
print('正例/直接拒绝负例:', n, m)
```

## 6 证书安全与边界

```python
import copy
bad = copy.deepcopy(c)
bad['area']['ABCD'] = True
print('布尔攻击通过:', verify_certificate(EXAMPLE,bad))
shift = {k:[x+1 for x in v] for k,v in EXAMPLE.items()}
print('移绑通过:', verify_certificate(shift,c))
print('有限实跑不证明一般定理；compile/exec非kernel。')
```
