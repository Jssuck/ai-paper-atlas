# 《几何原本》VI.22：长度比例与面积比例相互推出

```python
from fractions import Fraction as Q
from proportional_figures import area, compare, fourth, isosceles, right_345, square, third

values = tuple(map(area,(square(2),square(3),isosceles(4),isosceles(6))))
print("面积", values, "两对面积比", Q(values[0],values[1]), Q(values[2],values[3]))
print("两条第三比例线", third(2,3), third(4,6))
```

有了面积比例，尝试通过前三条底作第四条，并验证误选底长不可能。

```python
d = fourth(2,3,4)
print("应作第四线", d, "作出的同类图形面积", area(isosceles(d)))
for proposed in (5,d,7):
    print("第四底", proposed, "是否给出所需面积比", compare(4,9,8,area(isosceles(proposed))))
```

错配边或移除对内相似条件，都会使面积和底的比例不再等价。

```python
r,s = right_345(4),right_345(6)
print("两对面积成比例", compare(4,9,area(r),area(s)))
print("把第二个三角形斜边 10 错当对应底", compare(2,3,4,10))
thin=((0,0),(3,0),(3,1),(0,1))
print("底成比例、对内不再相似，则面积成比例？", compare(4,area(thin),8,18))
skew, rectangle=((0,0),(4,0),(5,3),(1,3)),((0,0),(9,0),(9,3),(0,3))
print("对内不相似时可以面积成比例", compare(4,9,area(skew),area(rectangle)))
print("但底不成比例", compare(2,3,4,9))
```
