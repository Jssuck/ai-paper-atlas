```python
from fractions import Fraction
xs = (8,12,18,27)
t = construct(xs,3)
if any(Fraction(y,x) != Fraction(3,2)**3 for x,y in zip(t.extracted,t.extracted[1:])):
    raise RuntimeError('wrong extracted ratio')
print(t.extracted)
```

```python
try:
    construct((4,7,9),2)
except ValueError as exc:
    print('rejected:',exc)
else:
    raise RuntimeError('unexpected acceptance')
```
