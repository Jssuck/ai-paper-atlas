## 1. 给出两个候选顶点
有理端点之间的向量旋转90度，并乘√3/2，产生两侧顶点。生成候选尚未代替验证。

```python
a, b = (0,0), (1,0)
vertices = construct(a,b)
print([[v.record() for v in c] for c in vertices])
```

## 2. 独立检查三边平方与方向
证书检查器只读取坐标，不调用构造公式。非零方向排除共线退化。

```python
for c in vertices:
    print(certificate(a,b,c))
```
