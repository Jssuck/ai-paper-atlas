# Euclid V.11：三组整倍比较共用同一个探针

正有理数精确门槛与有限试探分开；有限试探不证明同比。


```python
from ratio_chains import Ratio, probe_chain, exact_chain, bounded_mismatch
equal = Ratio(2,3),Ratio(4,6,'area'),Ratio(6,9,'mass')
print('有理数精确检查：', exact_chain(*equal))
for m,n in ((1,1),(3,2),(2,1)):
    signs = probe_chain(*equal,m,n)
    print((m,n), signs)
    if len(set(signs)) != 1:
        raise RuntimeError('mismatch')
```

```python
near = Ratio(101,100),Ratio(102,100),Ratio(103,100)
print('窗口 20：', bounded_mismatch(*near,20))
print('窗口 40：', bounded_mismatch(*near,40))
print('精确检查：', exact_chain(*near))
```

```python
a,b,c = Ratio(1,2),Ratio(1,3),Ratio(1,4)
print('各自挑选探针得到三个零：', (a.sign(2,1),b.sign(3,1),c.sign(4,1)))
print('真正的同一探针：', probe_chain(a,b,c,2,1))
```
