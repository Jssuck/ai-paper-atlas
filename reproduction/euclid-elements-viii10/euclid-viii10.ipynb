# 欧几里得 VIII.10：从两条单位链拼出混合比例链

```python
from join_unit_chains import join
x,y = (1,4,16,64),(1,6,36,216)
t = join(x,y)
print(t)
```

## 精确分数逐步复核

```python
from fractions import Fraction
check = tuple(64*Fraction(3,2)**i for i in range(4))
if check != t.mixed:
    raise RuntimeError('independent step check failed')
print(check)
```

## 不同长度不能直接配对

```python
try:
    join(x,(1,6,36))
except ValueError as exc:
    print('rejected:',exc)
else:
    raise RuntimeError('unexpected acceptance')
```
