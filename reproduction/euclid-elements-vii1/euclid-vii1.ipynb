# 《几何原本》VII.1：更相减法与互素判定

```python
from mutual_primes import trace, divisor_oracle
state,rows=trace(35,22)
for row in rows:
    print(row)
print("结束情形",state,"共同正因子",divisor_oracle(35,22))
```

到达新单位与被较小数整除是不同分支。

```python
for a,b in ((42,18),(21,7),(9,9),(9,1),(1,1)):
    state,rows=trace(a,b)
    print((a,b),state,"段数",len(rows),"共同正因子",divisor_oracle(a,b))
```

普通相减与带余除法成批计算的轨迹一致，操作计数不同。

```python
state,rows=trace(10001,2)
print("重复相减步数",sum(row.quotient for row in rows))
print("成批段数",len(rows),"轨迹一致",trace(10001,2,True)==(state,rows))
```
