# 欧几里得 IV.4：三角形的内切圆
用精确分数检查角平分线、内切半径与切点。请在本目录启动 Notebook。

```python
from incircle import FIXTURES, bisector_center, construct, grid

def pretty(P):
    return tuple(map(str, P))

P = FIXTURES["right"]
circle = construct(*P)
print("D =", pretty(circle.center), "r =", circle.radius)
for label in ("E", "F", "G"):
    print(label, "=", pretty(getattr(circle, label)))
```

## 三角形是钝角时，圆也必须在内部
比较权重构造与两条内角平分射线求交。另试一个合法三角形，但斜边不在本程序的有理数域内。

```python
P = FIXTURES["obtuse"]
circle = construct(*P)
other = bisector_center(*P)
if circle.center != other:
    raise ArithmeticError("centers disagree")
print("obtuse D =", pretty(other), "r =", circle.radius)
try:
    construct((0, 0), (1, 0), (0, 1))
except ValueError as error:
    print("exact domain limit:", error)
else:
    raise ArithmeticError("expected irrational-side rejection")
```
