用同一排数配对汇总，观察约分后的份数怎样保留。

## 另一条核对路线

直接从输入构造精确分数，与配对模型比较。

```python
for a, bc, d, ef in ((4,12,12,36), (6,18,10,30)):
    p = paired_rows(a, bc, d, ef)
    local, pooled = Fraction(a, d), Fraction(bc, ef)
    if not local == pooled == Fraction(p.m, p.k):
        raise RuntimeError('ratio mismatch')
    print(p.kind, local, pooled)
```

## 两排复制次数不相同

这里只改右边的整条，比较不同结果。

```python
try:
    paired_rows(6, 18, 10, 40)
except ValueError as exc:
    print('rejected:', exc, '; ratios:', Fraction(6,10), Fraction(18,40))
else:
    raise RuntimeError('unexpected acceptance')
```
