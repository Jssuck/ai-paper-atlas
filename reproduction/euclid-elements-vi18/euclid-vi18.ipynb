# VI.18：在指定边上复制相似图形

用精确有理坐标比较全局矩阵与局部角射线。

```python
from ratio_chains import direct_similarity, reconstruct
from fractions import Fraction as Q
C,D,E,F = (0,0),(4,0),(3,2),(1,3)
A,B = (1,1),(1,7)
T = direct_similarity(C,D,A,B)
print("G, H via matrix:", T(F), T(E))
```

## 验证两次角交点

先复制 CDF 得 G，再沿 DF 对应的新边 BG 复制 DFE 得 H。

```python
G,H = reconstruct(C,D,F,E,A,B)
if (G,H) != (T(F),T(E)):
    raise ValueError("construction mismatch")
print("G, H via rays:", G, H)
```

## 镜像反例

目标边是竖直线 x=1。镜像保持距离，却不是保向相似的同一个输出。

```python
from ratio_chains import det, sub
mirror = (2-G[0], G[1])
print("oriented determinants:", det(sub(B,A),sub(G,A)), det(sub(B,A),sub(mirror,A)))
```
