# VI.15：用共同三角形连接两次比例

目标：从坐标重建 ABC、ADE、BAD，检查等角假设及去掉它后的反例。

## 1. 构造共同的比较对象

取斜坐标方向 u=(1,0)、v=(1,1)，三块面积应为 6、6、4。

```python
from fractions import Fraction as Q
from ratio_chains import reconstruct

r = reconstruct(6, 4, 2, 3, v=(1, 1))
print("ABC, ADE, BAD:", *(str(a) for a in (r.area_abc, r.area_ade, r.area_bad)))
```

## 2. 检查两次换底

ABC 与 BAD 共用顶点 B；ADE 与 BAD 共用顶点 D。

```python
p, q, s = r.area_abc, r.area_ade, r.area_bad
print("area ratios:", p/s, q/s)
print("crossed side ratios:", r.ratio_ca_ad, r.ratio_ea_ab)
if (p/s, q/s) != (r.ratio_ca_ad, r.ratio_ea_ab):
    raise ValueError("the two base comparisons disagree")
```

## 3. 改变一个角

用独立构型分别取一个直角三角形 ABC 和一个斜角三角形 ADE；不再假定存在原命题的共同图。

```python
right = reconstruct(6, 4, 2, 3)
tilted = reconstruct(6, 4, 2, 3, v=(Q(3, 5), Q(4, 5)))
adjusted = reconstruct(6, 4, 2, Q(15, 4), v=(Q(3, 5), Q(4, 5)))
print("equal side ratios, areas:", right.area_abc, tilted.area_ade)
print("equal areas, side ratios:", right.ratio_ca_ad, adjusted.ratio_ea_ab)
```
