## 1 构图
h为半段基系数，d为中点右侧偏移，0<d<h；非单位正交等长基需按基长换算实际长度。

```python
from geometry import construct, verify, normalized_segments
from fractions import Fraction as Q
c = construct(5,2)
print({k:tuple(map(str,v)) for k,v in c["points"].items()})
```

## 2 四次I.47
先验证六个直角、等长条件，再比较ACE、EGF、AEF和ADF；不是把目标结论当成输入。

```python
r = verify(c,5,2)
for name,(left,right) in r["routes"].items():
 print(name,str(left),"=",str(right))
print("two routes",str(r["lhs"]),str(r["rhs"]))
```

## 3 左半分点
交换端点身份把负偏移规范化，原始AD与DB交换，平方和不变。中点与端点仍拒绝。

```python
n = normalized_segments(5,-2)
print({k:str(v) for k,v in n.items()})
print("sum",verify(construct(n["h"],n["d"]),n["h"],n["d"])["lhs"])
```

## 4 非单位正交基
两基长均5，平方量乘25，而非乘5。

```python
scaled=verify(construct(5,2,u=(3,4),v=(-4,3)),5,2,u=(3,4),v=(-4,3))
print("scale squared",scaled["basis_norm_squared"],"sum",scaled["lhs"])
```
