# 《几何原本》III.34：在已给圆中切弓形

从固定切点复制角，求第二交点，再用两侧见证核验指定的弧。请从本目录运行。


## 1. 作弦并选交替弧

切线朝右；复制的角用一个朝圆内的有理向量表示。


```python
from cut_segment import construct, inspect, finite_sweep

for ray in [(3,4), (0,4), (-3,4)]:
    c,p,q = construct(5,ray)
    checks = inspect(5,ray,c,p,q)
    print(ray, '弦端：', c, '目标弧点：', p, '全部通过：', all(checks.values()))
```

## 2. 故意换错弧

错弧一般给补角，直角时值相等，但侧别仍然不通过。


```python
for ray in [(3,4), (0,4), (-3,4)]:
    c,p,q = construct(5,ray)
    swapped = inspect(5,ray,c,q,p)
    print(ray, '交替侧：', swapped['alternate_side'], '角值相同：', swapped['matched_angle'])
```
