# VIII.18：相似平面数之间的等比中项

```python
from similar_plane import construct
print(construct(4,6,6,9))
print(construct(6,9,4,6))
```

```python
from math import isqrt
t = construct(4,6,6,9)
a,g,b = t.chain
if isqrt(a*b) != g or g != t.alternate_middle:
    raise RuntimeError('wrong middle term')
print('mean:',g,'; endpoint ratio:',t.growth**2)
```

```python
print('integer mean:',isqrt((2*8)*(4*9)),'; invalid mixed product:',8*4)
try:
    construct(2,8,4,9)
except ValueError as exc:
    print('rejected:',exc)
else:
    raise RuntimeError('unexpected acceptance')
```
