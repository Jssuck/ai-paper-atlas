```python
from fractions import Fraction as Q
from circumscribed_triangle import as_text, construct, dot, grid, sub
source, contact, output, _ = construct(Q(1, 2), Q(1, 3), width=5)
for label, values in zip(("D E F", "A B C", "L M N"), (source, contact, output)):
    print(label, as_text(values))
```

## 钝角也要使用内角射线

令底角 E 为钝角；它对应输出 M。点积应严格为负。然后尝试非法的两个直角输入。


```python
_, _, (l, m, n), _ = construct(Q(3, 2), Q(1, 4))
value = dot(sub(l, m), sub(n, m))
if value >= 0:
    raise ArithmeticError("角 M 应是钝角")
print("M 内角两射线向量的点积：", value)
try:
    construct(1, 1)
except ValueError as e:
    print("退化输入拒绝：", e)
else:
    raise ArithmeticError("不应接受退化输入")
```
