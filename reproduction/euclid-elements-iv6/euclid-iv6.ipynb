# 《几何原本》IV.6：作圆的内接正方形
这是现代有理坐标教学重构。目标是分别检验共圆、等边和直角，并观察拿掉垂直条件后哪些性质仍成立。
在本目录启动 Notebook，保证能导入同目录的 Python 源文件。

```python
from fractions import Fraction as F
from inscribed_square import inscribe, verify, measurements, rectangle_control, grid_sweep

o, a = (F(1, 2), F(-1, 3)), (F(5, 2), F(2, 3))
points = inscribe(o, a)
m = verify(o, a, points)
print("A B C D:", [tuple(map(str, p)) for p in points])
for key in ("radii_squared", "sides_squared", "adjacent_dot", "diagonals_squared"):
    print(key, tuple(map(str, m[key])))
print("area:", m["signed_area"])
```

## 去掉垂直条件
仍用两个等长的直径，对径端点交错连接。先猜：四角、等边、两条对角线等长，这三个条件哪些会保留？

```python
p = rectangle_control()
m = measurements((0, 0), p)
print("side^2:", tuple(map(str, m["sides_squared"])))
print("adjacent dot:", tuple(map(str, m["adjacent_dot"])))
print("diagonal^2:", tuple(map(str, m["diagonals_squared"])))
try:
    verify((0, 0), p[0], p)
except ValueError as exc:
    print("nonsquare rejected:", exc)
else:
    raise RuntimeError("control should have been rejected")
```
