## 同高的三个比较分支

三列符号依次属于底边、三角形和平行四边形。降低右边的高度可破坏这一同步。

```python
for m, n in ((1, 1), (3, 2), (4, 2)):
    print(m, n, probe(2, 3, 4, 4, m, n)[2])
print("different heights", probe(2, 3, 4, 2, 1, 1)[2])
```

## 不可公度底边的精确探针

高度取 2，底边为 1 和 √2。比较正数时可比较其平方，不需要浮点数。有限探针不代替一般证明。

```python
for m, n in ((1, 1), (2, 1), (99, 70), (239, 169)):
    print(m, n, sqrt2_probe(m, n))
```
