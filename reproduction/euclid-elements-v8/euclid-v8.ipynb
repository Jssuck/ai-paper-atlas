# 《几何原本》V.8：构造夹在两个倍量之间的第三个倍量
教学目标：区分 V.Def.7 的存在性证据、原文证明构造与有限试探。

```python
from witness import Magnitude as M, construct, sample_cases, witness_signs, bounded_witness, certifies
for branch, (a,b,c) in sample_cases():
    p = construct(a,b,c)
    print(branch, p.m, p.n, p.small_multiple, p.shared_multiple, p.large_multiple)
    assert witness_signs(a,c,b,c,p.m,p.n) == (1,-1)
    assert witness_signs(c,b,c,a,p.n,p.m) == (1,-1)
```

## 边界见证
正向第二项可以相等；同一对倍数在互换位置后不一定是逆次序的有效见证。

```python
a,b,c = map(M, (3,2,2))
for quadruple in ((a,c,b,c), (c,b,c,a)):
    print(witness_signs(*quadruple,1,1), certifies(*quadruple,1,1))
```

## 有限搜索会漏掉见证
只扫到 20 不表示两个比相等或没有大小关系。对照更大窗口和构造见证。

```python
a,b,c = map(M,(102,101,100))
for bound in (20,52):
    print(bound, bounded_witness(a,c,b,c,bound))
p = construct(a,b,c)
print('construction',p.m,p.n)
```

有限网格只是对实现的检验；一般论证依赖倍量的存在性与严格夹逼。
