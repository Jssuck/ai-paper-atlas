# I.45：两块等积平行四边形的拼接
原创有限精确模型；不是尺规模拟或一般定理证明。标准库compile/exec，非Jupyter kernel。

```python
from geometry import construct,verify,area
from cases import BASE,valid_cases,INVALID
c=construct(BASE)
print(list(map(str,c["areas"])))
```

严格凸四边形由DB分成ABD与DBC，输入范围不包括一般凹多边形。

```python
print({k:list(map(str,v)) for k,v in c["points"].items()})
print("verified",verify(BASE,2,1,2,c))
```

共边HG的两侧各一块，拼合输出K-F-L-M，不是任意两等积图形都能拼合。

```python
P=c["points"]
print("seed",area([P[k] for k in "KFGH"]))
print("added",area([P[k] for k in "HGLM"]))
print("whole",area([P[k] for k in "KFLM"]))
```

锐角、直角、钝角都可在q>0规范中表达。

```python
for p in (-2,0,1):
    z=construct(BASE,2,p,2)
    print(p,verify(BASE,2,p,2,z),str(z["areas"][2]))
```

最后检查有限网格和非法输入；通过不证明无限域一般命题。

```python
n=sum(verify(*a,construct(*a)) for a in valid_cases())
r=0
for name,*a in INVALID:
    try: construct(*a)
    except ValueError: r+=1
print("valid",n,"invalid rejected",r)
```
