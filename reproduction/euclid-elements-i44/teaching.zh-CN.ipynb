# I.44：把面积施于给定底边
五格标准库compile/exec真实执行，非Jupyter kernel。规范有理坐标模型不是尺规动作模拟器；有限运行不证明一般定理。

```python
from geometry import construct,verify,area,sub,det
from fractions import Fraction as F
c=construct(3,2,6,1,1); P=c['points']
print(P)
```

先有面积为6的BEFG，再用HB与FE的延长交点K把底边改为AB=3。

```python
print('HB与HK共线:',det(sub(P['B'],P['H']),sub(P['K'],P['H']))==0)
print('K:',P['K'],'M:',P['M'],'L:',P['L'])
```

I.43给出余形等积，坐标面积只是现代核验。

```python
print('BEFG面积:',area([P[k] for k in 'BEFG']))
print('ABML面积:',area([P[k] for k in 'ABML']))
print('核验:',verify(c))
```

改变初始BE长度，最终给定底边上的结果不变；这不声称一般尺规证书完整性。

```python
d=construct(3,7,6,1,1)
print('最终四点相同:',all(P[k]==d['points'][k] for k in 'ABML'))
```

错摆A或输入退化角会被拒绝；等积数值不能替代全部前提。

```python
c['points']['A']=(3,0)
print('错摆核验:',verify(c))
try:
 construct(3,2,6,1,0)
except ValueError as e:
 print('退化角:',type(e).__name__)
```
