# 欧几里得 VII.31：合数有质因数

```python
from prime_divisor_descent import descend
for mode in ('largest','smallest','middle'):
    print(mode,descend(84,mode))
```

## 每一步都严格变小，终点仍量度原数

```python
path = descend(84)
if any(not 1 < b < a or a%b or path[0]%b for a,b in zip(path,path[1:])):
    raise RuntimeError('bad descent')
p = path[-1]
if any(p%d == 0 for d in range(2,p)):
    raise RuntimeError('terminal is not prime')
print('terminal divisor:',p,'; count:',path[0]//p)
```

## 单位与已经是质数的输入不在合数前提内

```python
for n in (1,13):
    try:
        descend(n)
    except ValueError as exc:
        print(n,'rejected:',exc)
    else:
        raise RuntimeError('unexpected acceptance')
```
