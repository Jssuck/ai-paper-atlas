## 4. 等角不能代替侧向

```python
bad=fixture(); bad['E']=['1','1']
try:
    make_certificate(bad)
except ValueError as e:
    print('wrong-side reflected choice rejected:', str(e))
```

## 5. 有限正负样例

```python
count=sum(verify_certificate(d,make_certificate(d)) for d in positives())
print('positive inputs checked:', count)
rejected=0
for name,d in negatives():
    try: make_certificate(d)
    except ValueError: rejected+=1
    else: raise RuntimeError(name)
print('negative inputs directly rejected:',rejected)
```
