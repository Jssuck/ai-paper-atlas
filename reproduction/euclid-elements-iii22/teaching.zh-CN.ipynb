## 1. 正方形只是特殊正例
四个cot均为0，不代表所有内接四边形均为矩形。

```python
from fractions import Fraction as F
from geometry import certificate,circle_points,grid
from independent_oracle import check
Q=((1,0),(0,1),(-1,0),(0,-1))
print(certificate(Q)["cots"])
```

## 2. 原证的两次III.21替换
triangle ABC的BAC=BDC、ACB=ADB，后两角拼成ADC。负cot不可取绝对值。

```python
z=certificate(tuple(circle_points()[:4]))
print("四内角cot:",z["cots"])
print("原证替换:",z["same_segment"],"拼角:",z["split_matches_D"],"三角形关系:",z["triangle_sum"])
```

## 3. 练习：把B、C交换，结论还可直接用吗？
答案：不能；这不是凸四边形边界次序，程序应拒绝。非反角的两对角和在这个交叉正方形例子中为90度而非180度。

```python
try:
    certificate((Q[0],Q[2],Q[1],Q[3]))
except ValueError as e:
    print("已拒绝:",e)
```

## 4. 很近不等于重合
有理运算可区分距离极近的合法点；若设t=0则重复点应拒绝。

```python
t=F(1,10**12)
p=((1-t*t)/(1+t*t),2*t/(1+t*t))
print("近退化精确和:",certificate((Q[0],p,Q[1],Q[2]))["opposite_sums"])
```
