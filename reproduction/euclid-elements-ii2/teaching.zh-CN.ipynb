# II.2：整线与两段的矩形
目标：区分区域分割、面积数值与原文几何证明。点序为 A-D-E-B；AF 表示 A-D-F-C 区域，CE 表示 C-F-E-B 区域。
五格代码采用标准库 compile/exec 真运行并保存输出，未启动 kernel。有限有理坐标检查不能证明一般命题；不能据此声称复现整部《原本》。

1. 按原文点名构造正方形和切点

```python
from fractions import Fraction as Q
from geometry import construct, verify, summary
a=((0,0),(6,0),(0,6),Q(1,3))
c=construct(*a)
print(c["points"])
```

2. 区域证书与鞋带面积

```python
print(verify(*a,c))
print(summary(*a))
```

3. 旋转和反射不改变结论

```python
print(summary((1,2),(4,6),(-3,5),Q(2,5)))
print(summary((1,2),(4,6),(5,-1),Q(2,5)))
```

4. 面积相等不能代替区域绑定

```python
b=((0,0),(6,0),(0,6),Q(1,2))
bad=construct(*b)
bad["AF"],bad["CE"]=bad["CE"],bad["AF"]
print(verify(*b,bad))
print("segment squares:",2**2+4**2,"whole square:",6**2)
```

5. 原文内点模型拒绝退化和外切

```python
for t in [0,1,Q(3,2),True,0.5]:
    try: construct((0,0),(6,0),(0,6),t)
    except ValueError as e: print(type(e).__name__+": "+str(e))
```
