# I.40：等底等积与平行

有理坐标教学模型，非原文合成证明。原文 ABC/DCE、辅助 F；本模型推广为 ABC/DEF、辅助 G，把原 C 拆为底端点 C 和 E。有限实验不证明一般定理。

## 1 规范输入与结论
六点模型两底可分离、交叠、共端或重合，必须等长且共线。

```python
from geometry import make_certificate, verify_certificate, construction_trace
from cases import EXAMPLE, build, positives, negatives
c=make_certificate(EXAMPLE)
print(c['trace']['input'])
print(c['conclusion'])
print(c['trace']['double_area'])
```

## 2 辅助交点
G=E+t(D−E)；内外两例故意不等积，仅演示反证所比较的整体与真部分。

```python
for k in [4,1,2]:
    t=construction_trace(build(k=k))
    print(t['branch'], 't='+t['t'], 'G=',t['G'], t['double_area'])
    print('whole/part:',t['whole'],t['part'])
```

## 3 合法边界与非法输入
A=D 时 AD 不定义；同一平行线存在，但不能冒称由两重合点定义了直线。

```python
print(make_certificate(build(d=1))['conclusion'])
for label,data in negatives():
    try:make_certificate(data)
    except ValueError as e:print(label,':',str(e))
    else:raise RuntimeError('invalid geometry accepted')
```

## 4 证书严格绑定
不是只核零行列式：所有原始坐标和重算字段都绑定。

```python
import copy
bad=copy.deepcopy(c);bad['trace']['cross']['BC_AD']=True
print('布尔篡改通过:',verify_certificate(EXAMPLE,bad))
print('同面积移绑通过:',verify_certificate(build(a=2),c))
print('原证书通过:',verify_certificate(EXAMPLE,c))
```

## 5 有限批量核验
实际执行方式为标准库 compile/exec、非 Jupyter kernel；不代表形式化证明。

```python
count=0
for data in positives():
    if not verify_certificate(data,make_certificate(data)):raise RuntimeError('verification failed')
    count+=1
print('有限合法例:',count)
print('有限实跑不证明一般定理；标准库 compile/exec 非 kernel。')
```
