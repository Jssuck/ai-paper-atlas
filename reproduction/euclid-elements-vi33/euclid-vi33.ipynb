# 《几何原本》VI.33：比较弧与两类角的任意倍

```python
from fractions import Fraction as Q
from arc_angles import Arc,compare_arcs,principal_turns
a,b=Arc(3,Q(1,6)),Arc(3,Q(1,4))
print("三种比",a.length_over_tau()/b.length_over_tau(),
      a.central_turns()/b.central_turns(),a.inscribed_turns()/b.inscribed_turns())
for m,n in ((2,2),(3,2),(4,2)):
    print("倍数",m,n,"符号",compare_arcs(a,b,m,n))
```

连续加角保留累计转数；折成普通较小夹角会改变比较。

```python
x,y=Q(2,3),Q(1,2)
print("累计",x,y,"第一项较大",x>y)
print("折叠",principal_turns(x),principal_turns(y),"第一项较小",principal_turns(x)<principal_turns(y))
```

相等的半径必须为两弧共同因子。

```python
print("不等半径，弧等长而角不等",compare_arcs(Arc(1,Q(1,2)),Arc(2,Q(1,4)),1,1))
print("不等半径，角相等而弧不等长",compare_arcs(Arc(1,Q(1,4)),Arc(2,Q(1,4)),1,1))
```
