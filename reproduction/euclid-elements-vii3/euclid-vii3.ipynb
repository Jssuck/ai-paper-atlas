# 《几何原本》VII.3：求三个数的最大共同量数

```python
from three_measures import three,direct_common
for values in ((84,30,42),(84,30,20)):
    out=three(*values,ancient_domain=True)
    print(values,out)
    print("三数共同因子",direct_common(*values),
          "化简为两数后的共同因子",direct_common(out.first_measure,values[2]))
```

先选哪一对会影响中间轨迹，不影响最终的共同量数。

```python
from itertools import permutations
for values in permutations((12,18,20)):
    out=three(*values)
    print(values,out.first_measure,out.branch,out.final_measure)
```

每一对都有共同非单位，并不保证三者共享一个。

```python
print("三对最大共同因子",[max(direct_common(*p)) for p in ((6,10),(6,15),(10,15))])
print("三数共同因子",direct_common(6,10,15),"现代结果",three(6,10,15).final_measure)
```
