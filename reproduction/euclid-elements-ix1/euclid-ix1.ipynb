# IX.1：相似平面数的积
构造两个连比链，从第二条链恢复积的平方根。

```python
from square_product import square_product
r = square_product((3,6),(2,4))
print('Plane and lifted chains:', r.plane_chain, r.lifted_chain)
print('Root and reversed:', r.root, square_product((2,4),(3,6)))
```

```python
from math import isqrt
for left,right in [((3,6),(1,8)),((2,3),(2,5))]:
    n = left[0]*left[1]*right[0]*right[1]
    try:
        square_product(left,right)
    except ValueError as exc:
        print(n, 'square?', isqrt(n)**2==n, 'rejected:',exc)
    else:
        raise RuntimeError('Expected mismatched factors to fail')
```

```python
print('Modern unit:',square_product((1,2),(3,6)))
print('Modern equality:',square_product((3,6),(3,6)))
```
