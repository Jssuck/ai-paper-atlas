# 《几何原本》V.3：检查两阶段倍数后再展平

目标：区分两阶段分别相同与总倍数偶然相同。

## 1. 两列分别复制，保留原量类

内层每块 2 份，外层各 3 块。长度列、面积列各自累加，不跨列加。

```python
from fractions import Fraction as F
from nested_multiples import Magnitude as M, compose, regression_grid
bases = M(F(3,2)), M(F(5,2), "area")
middle = M(3), M(5, "area")
large = M(9), M(15, "area")
cert = compose(bases, middle, large)
print("内层、外层、合计:", cert.inner_count, cert.outer_count, cert.total_count)
print("两列总量:", cert.left_total.value, cert.right_total.value)
```

## 2. 互换分组仍可能给同一个乘积

左列每块 2 份，共 3 块；右列每块 3 份，共 2 块。

```python
bases, middle, large = (M(1),M(1)), (M(2),M(3)), (M(6),M(6))
print("两列最终倍数:", *(z.value/x.value for x,z in zip(bases,large)))
try:
    compose(bases, middle, large)
except ValueError as error:
    print("拒绝:", error)
```

## 3. 整数总倍数也不能证明外层是整倍

每列的总倍数是 5，外层同比却是 5/2，并非整数复制数。

```python
bases, middle, large = (M(2),M(4)), (M(4),M(8)), (M(10),M(20))
print("外层两列:", *(z.value/y.value for y,z in zip(middle,large)))
try:
    compose(bases, middle, large)
except ValueError as error:
    print("拒绝:", error)
print("有效组合与换序拒绝:", regression_grid())
```
