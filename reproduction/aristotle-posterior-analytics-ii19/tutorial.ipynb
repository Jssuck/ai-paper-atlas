## 1. 三个对象与一次观察
覆盖一个对象，仍有两个对象未知；四种相容解释不是四种等概率事件。

```python
from model import analyse, compatible_explanations, group_memories
u = ['a', 'b', 'c']
print(analyse(u, [('a', True)]))
```

## 2. 重复不增加覆盖
重复同一正例100次，观察次数增加，唯一对象数与相容解释数不变。

```python
r = analyse(u, [('a', True)] * 100)
if (r['covered_count'], r['compatible_count']) != (1, 4):
    raise RuntimeError('repeat invariance failed')
print(r)
```

## 3. 同一记录的相反延伸
a为真，未观察b。两种解释都符合记录，但对全称命题的答案相反。

```python
print(compatible_explanations(['a', 'b'], [('a', True)]))
print(analyse(u, [('a', True), ('b', False)]))
print(analyse(u, [('a', True), ('b', True), ('c', True)]))
```

## 4. 给定分类与经验分组
分类X/Y是输入，不是程序发现的本质；重复记忆仍不等于不同对象。

```python
print(group_memories(u, [('a', True), ('a', True), ('b', True)], {'a': 'X', 'b': 'X', 'c': 'Y'}))
try:
    analyse(u, [('a', True), ('a', False)])
except ValueError as error:
    print(type(error).__name__, str(error))
else:
    raise RuntimeError('conflict must be rejected')
```
