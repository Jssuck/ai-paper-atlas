```python
from square_divisibility import inspect, from_squares
print(inspect(4,12))
print('recovered side count:',from_squares(4,12))
```

```python
from fractions import Fraction
for c,d in ((12,4),(6,9),(4,10),(2,3)):
    t=inspect(c,d)
    a,e,b=t.bridge
    if Fraction(b,a).denominator != Fraction(d,c).denominator**2:
        raise RuntimeError('denominator mismatch')
    print(c,d,t)
```
