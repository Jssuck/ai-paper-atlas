# X.34：平方坐标的根式证书

```python
from fractions import Fraction as F
from ratio_witness import construct
for k in (F(1,2),F(2,5)):
    w=construct(2,k)
    print(f'k={k}:',w)
    print('positive area squared:',w.center_squared-w.offset_squared)
```
