# VI.12：作第四比例线
目标是用精确坐标重建相交步骤，并检查何时不能替换已知长度。

## 1. 构造与比例证书
在第一条射线上先取 DG=4，再接上 GE=6；第二条射线上取 DH=8。

```python
from ratio_chains import construct, certificate, sweep
points = construct(4, 6, 8)
print(dict(zip("DGEHF", points)))
print("HF =", certificate(4, 6, 8, points))
```

## 2. 错误复制
若误把 GE 设成 4，平行相交仍有解，但未解决原来的问题。

```python
wrong = construct(4, 4, 8)
print("HF with GE=4:", certificate(4, 4, 8, wrong))
try:
    certificate(4, 6, 8, wrong)
except ValueError as exc:
    print("Rejected:", exc)
else:
    raise RuntimeError("the copied length changed")
```
