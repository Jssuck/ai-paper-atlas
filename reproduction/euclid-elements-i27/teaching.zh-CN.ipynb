# I.27：角相等与两条无限直线

这是现代有理坐标教学核验；非欧几里得原始证明内核。随附执行器使用标准库 compile/exec，**不是 Jupyter kernel**。从本目录运行。

```python
from geometry import make_certificate, verify_certificate
from cases import positives, negative_cases
data = {"E":["0","0"],"F":["2","0"],"A":["1","1"],"D":["1","-1"]}
certificate = make_certificate(data)
print(certificate["angle_keys"])
```

两角都是锐角；射线在横截线异侧。方向行列式为零，横截偏移非零，故直线不重合。

```python
print(certificate["direction_determinant"])
print(certificate["transversal_side_determinants"])
print(certificate["opposite_scale"])
print(verify_certificate(data, certificate))
```

平方余弦会混淆互补角；这里的有符号指标不会。六类失败均应由生成器直接拒绝。

```python
rejected=[]
for name,bad in negative_cases().items():
    try: make_certificate(bad)
    except ValueError: rejected.append(name)
    else: raise RuntimeError(name)
print(rejected)
```

```python
import copy
attack=copy.deepcopy(certificate)
attack["schema"]=True
print("bool schema accepted:",verify_certificate(data,attack))
other=next(positives())
print("other input accepted:",verify_certificate(other,certificate))
```

有理变换批量检查只是有限样例，不证明一般定理。

```python
count=0
for p in positives():
    if not verify_certificate(p,make_certificate(p)): raise RuntimeError("failed")
    count+=1
print("positive cases:",count)
```
