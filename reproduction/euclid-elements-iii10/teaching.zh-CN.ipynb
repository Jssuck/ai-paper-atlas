## 3. 两交点、相切、相离及无理坐标

```python
for d in [1,2,3]:
    c=classify((0,0),1,(d,0),1)
    print('center distance',d,'count',c['count'],'N',c['discriminant'])
print('irrational points are not rounded:',rational_intersections((0,0),1,(1,0),1))
```
