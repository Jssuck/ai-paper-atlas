# 欧几里得 VI.13：半圆中的中比例项

用现代坐标复核尺规构造。先算高度平方，再判断高度本身能否用有理数表示。

```python
from fractions import Fraction
from ratio_chains import construct

c = construct(9, 16)
if (c.altitude_squared, c.rational_altitude, c.right_angle_dot) != (144, 12, 0):
    raise ValueError("unexpected rational example")
print(f"center x={c.center_x}, radius={c.radius}, BD²={c.altitude_squared}, BD={c.rational_altitude}")
```

平方值可以精确而高度不可用有理数表达。

```python
c = construct(1, 2)
if c.altitude_squared != 2 or c.rational_altitude is not None or c.right_angle_dot != 0:
    raise ValueError("unexpected irrational example")
print(f"BD²={c.altitude_squared}, rational height={c.rational_altitude}, right-angle dot={c.right_angle_dot}")
```

圆的最高点不一定在垂线 BD 上。把半径当作高度会错。

```python
c = construct(9, 16)
top_BD_squared = c.center_x**2 + c.radius**2
if c.center_x == 0 or top_BD_squared != Fraction(337, 2) or top_BD_squared == c.altitude_squared:
    raise ValueError("unexpected counterexample")
print(f"top x={c.center_x}, top distance²={top_BD_squared}, correct BD²={c.altitude_squared}")
```
