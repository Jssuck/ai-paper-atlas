```python
from ratio_order import Magnitude, mags, order_transport, proportion, bounded_difference
for values in ((9, 6, 6, 4), (6, 4, 9, 6), (6, 4, 6, 4)):
    print(values, order_transport(*mags(*values)))
```

## 接近但并不同比
小窗口没有找到反例不代表同比成立。


```python
near = mags(51, 50, 52, 50)
print(proportion(*near), bounded_difference(*near, 20), bounded_difference(*near, 26))
```

## 同比并不意味着允许跨类比较对应项


```python
mixed = (*mags(9, 6), Magnitude(6, 'area'), Magnitude(4, 'area'))
print('same ratio:', proportion(*mixed))
try:
    order_transport(*mixed)
except ValueError as exc:
    print('order rejected:', exc)
```
