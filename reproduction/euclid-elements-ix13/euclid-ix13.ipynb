# IX.13：素数幂的真因子
构造指数证书，核验单位、末项与合数底数的边界。

```python
from prime_power_divisors import proper_nonunit_divisors, extraction_trace
print('Proper nonunit divisors of 81:',proper_nonunit_divisors(3,4))
print('Witness for 27:',extraction_trace(3,4,27))
```

```python
print('Empty interior for 7:',proper_nonunit_divisors(7,1))
for divisor in (1,81):
    try:
        extraction_trace(3,4,divisor)
    except ValueError as exc:
        print(divisor,'excluded endpoint:',exc)
    else:
        raise RuntimeError('Expected endpoint exclusion')
```

```python
print('Remainder 6² divided by 4:',6**2%4)
try:
    proper_nonunit_divisors(6,2)
except ValueError as exc:
    print('Composite base rejected:',exc)
else:
    raise RuntimeError('Expected composite base exclusion')
```
