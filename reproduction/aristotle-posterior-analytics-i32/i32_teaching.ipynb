## 真结论不能倒推前提全真

非空集合模型：两个前提是 B⊆A 与 C⊆B，结论是 C⊆A。取 A=C={0}, B={1}，两个前提均假而结论真。真前提保真是包含关系传递性的数学结果；有限枚举只是核验实现，不代替证明。

```python
A, B, C = {0}, {1}, {0}
print({"B_subset_A": B <= A, "C_subset_B": C <= B, "C_subset_A": C <= A})
```

## 同一推理引擎不等于同一事实支持

geom:p 和 arith:r 是纯命题标签，不是几何或算术的真实公理。两条条件规则由同一个引擎处理；标签不自行生成跨领域桥梁。最小支持相对于固定规则，按集合包含而非仅按大小定义。

```python
facts = {"geom:p", "arith:r", "unused:u"}
rules = [(frozenset({"geom:p"}), "geom:q"), (frozenset({"arith:r"}), "arith:s")]
print("rules alone:", sorted(closure([], rules)))
print("q supports:", [sorted(s) for s in minimal_supports(facts, rules, "geom:q")])
print("s supports:", [sorted(s) for s in minimal_supports(facts, rules, "arith:s")])
print("no bridge:", sorted(closure(["geom:p"], rules)))
bridge = rules + [(frozenset({"geom:q"}), "arith:s")]
print("explicit bridge:", sorted(closure(["geom:p"], bridge)))
```

## 有限规则描述与无穷可生成式

规则是从任意项 x 生成 S(x)，起始项为 0。程序只打印深度 10 的 11 个式子。一般无限性来自语法归纳：第 n 项有恰好 n 个 S，故不同 n 的字符串互不相同；任意 n 可继续应用规则。这里是现代无界项语法，不应偷偷归给原文的有限项论证。

```python
terms = successor_prefix(10)
print("finite prefix size:", len(terms))
print("S-counts:", [t.count("S") for t in terms])
print("last term:", terms[-1])
```
