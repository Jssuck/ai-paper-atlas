# 《几何原本》VII.8：匹配若干份后逐份扣除

```python
from subtract_same_parts import construct,ratios
w=construct(10,15,6,9)
print("GH 与端点",w.gh(),w.endpoints)
print("各切点、局部余段、合并",w.cuts,w.leftovers,sum(w.leftovers))
print("三种比例",tuple(map(str,ratios(10,15,6,9))))
```

局部余段先由 VII.7 型单份关系同剩余整体比较，再求和。

```python
from fractions import Fraction as Q
print("每段余比",tuple(str(Q(piece,15-9)) for piece in w.leftovers))
print("单位组件情形",construct(4,6,2,3))
```

原比与移去之比不同，或移去全部，都不能产生本命题的见证。

```python
print("失配通常不能保比",tuple(map(str,ratios(10,15,6,12))))
for pair in ((10,15,6,12),(10,15,10,15),(4,6,1,3)):
    try:
        construct(*pair)
    except ValueError as exc:
        print(pair,type(exc).__name__)
```
