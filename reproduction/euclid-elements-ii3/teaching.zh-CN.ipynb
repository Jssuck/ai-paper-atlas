# Euclid II.3：整体乘所选分段

先核查所选分段CB上的正方形，再检查其向A延出的矩形。原创精确有理数教学；不是一般证明或原尺规过程的完整复现。

```python
from fractions import Fraction as Q
from geometry import construct, verify, summary, area
A=(0,0); u=(6,0); v=(0,4); t=Q(1,3)
print(summary(A,u,v,t))
```

## 点名与区域
AE为整体矩形；AD为左矩形；CE（亦可称DB）为右侧正方形。

```python
certificate=construct(A,u,v,t)
print({k:tuple(map(str,p)) for k,p in certificate['points'].items()})
print('accepted:',verify(A,u,v,t,certificate))
```

## 旋转与反射不改变判据
下一例AB=5、AC=2、CB=3，面积15=6+9。

```python
print(summary((1,-2),(3,4),(-Q(12,5),Q(9,5)),Q(2,5)))
print(summary((0,0),(6,0),(0,-4),Q(1,3)))
```

## 总面积相等仍可能是假图
把左块平移后面积未变，但它已不与右块准确拼合。

```python
import copy
bad=copy.deepcopy(certificate)
bad['AD']=tuple((x+1,y) for x,y in bad['AD'])
print('area equality:',area(bad['AE'])==area(bad['AD'])+area(bad['CE']))
print('certificate accepted:',verify(A,u,v,t,bad))
```

## 不可混用II.2的整体正方形
把高设为AB=6不符合本篇CB=4的正方形。

```python
try:
    construct(A,u,(0,6),t)
except ValueError as error:
    print('rejected:',error)
print('Finite cases do not prove the general theorem.')
```
