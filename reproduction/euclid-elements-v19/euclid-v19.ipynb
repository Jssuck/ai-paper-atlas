# 欧几里得 V.19：同类整体的余量与传本附论

正有理数教学模型，不替代任何一般证明。

```python
from ratio_remainders import Magnitude as Q, main_trace, porism_trace
a, b, u, v = map(Q, (10, 15, 4, 6))
print('whole scale:', a.value / b.value, 'removed scale:', u.value / v.value)
```

主命题：先交换中项，再分比，再交换中项，最后传递同比。

```python
for label, left, right in main_trace(a, b, u, v):
    print(label, left, right)
```

附论（重构）：分比、反比、合比均在各组内部操作，可令右组表示面积。

```python
b_area, v_area = Q(15, 'area'), Q(6, 'area')
for label, left, right in porism_trace(a, u, b_area, v_area):
    print(label, left, right)
```
