```python
from ratio_chains import ABC, D, E, F, M, N, probe, exercise_grid
p = probe(ABC, (D, E, F))
print("G:", tuple(map(str, p["g"])))
print("三个缩放因子平方:", tuple(map(str, p["side_ratios_sq"])))
assert p["ss_equal"] and p["copied_angles"] and p["equal_angles"]
```

## 反射与缺失的假设

把 D 镜像到 EF 的另一侧，三边和无向内角仍分别相等。只配齐两条边则不够。

```python
r = probe(ABC, (M, E, F))
n = probe(ABC, (N, E, F))
print("反射图同侧、三边相等、对应角相等:", r["same_half_plane"], r["ss_equal"], r["equal_angles"])
print("少一边的反例，三个比例平方:", tuple(map(str, n["side_ratios_sq"])))
assert not r["same_half_plane"] and r["equal_angles"] and not n["equal_angles"]
```
