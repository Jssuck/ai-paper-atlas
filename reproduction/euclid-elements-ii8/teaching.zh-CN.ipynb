## 参数与点资格
a=AC,b=BC=BD；正交等长基的长度另乘系数。a=b也合法。

```python
from fractions import Fraction as Q
from geometry import construct,verify
c=construct()
r=verify(c)
print(c["points"])
```

## 四个小平方
先核对每一格，不从最终恒等式倒推。

```python
print({k:str(r[k]) for k in ("CK","KD","GR","RN")})
```

## 四块矩形与八格曲尺形
AG和CK合成AK，八块互不重叠格面积为四倍AK。

```python
print({k:str(r[k]) for k in ("AG","MQ","QL","RF")})
print("gnomon",r["gnomon"],"four AK",4*r["AK"])
print("plus OH",r["gnomon"]+r["OH"],"whole",r["whole"])
```

## 有理数及非单位基
基向量(3,4),(-4,3)长度5，面积乘25。

```python
a,b=Q(2,3),Q(5,7)
z=verify(construct(a,b),a,b)
print("rational whole",z["whole"])
s=verify(construct(u=(3,4),v=(-4,3)),u=(3,4),v=(-4,3))
print("scaled whole",s["whole"])
```

## 错误证书
篡改交点Q必须被拒绝；有限检查不替代一般证明。

```python
bad=construct()
bad["points"]["Q"]=(Q(3),Q(3))
try:
    verify(bad)
except ValueError as e:
    print("rejected:",e)
else:
    raise RuntimeError("tampering accepted")
```
