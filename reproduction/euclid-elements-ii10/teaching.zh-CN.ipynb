# II.10：向外延长与同一AG的两条路线
原创精确教学核验。以下五格通过标准库compile/exec真实执行，未启动Jupyter kernel；有限实例不证明一般定理。

```python
from geometry import construct,verify,side_diagonal
from fractions import Fraction as Q
c=construct(3,2)
r=verify(c,3,2)
print({k:tuple(map(str,v)) for k,v in c['points'].items()})
```

## 检查四次I.47
先验证直角、等长和点证书，再比较四个平方关系。

```python
print({k:tuple(map(str,v)) for k,v in r['routes'].items()})
print('AG squared =',r['squares']['AG'],'; two routes =',r['lhs'],r['rhs'])
```

## 变换保持什么
非单位正交等长基把所有平方统一乘以基长平方。

```python
kw={'origin':(7,-4),'u':(0,2),'v':(2,0)}
z=verify(construct(3,2,**kw),3,2,**kw)
print('basis norm squared =',z['basis_norm_squared'],'; lhs =',z['lhs'])
```

## 非法证书不是近似正确
故意移动G，验证器必须拒绝。

```python
bad=construct();bad['points']['G']=(Q(8),Q(-3))
try:
    verify(bad)
except ValueError as e:
    print('rejected:',str(e))
else:
    raise RuntimeError('tampering accepted')
```

## 现代整数旁支

```python
for row in side_diagonal(8):
    print(row['s'],row['d'],row['difference'],'->',row['next_s'],row['next_d'],row['next_difference'])
```
