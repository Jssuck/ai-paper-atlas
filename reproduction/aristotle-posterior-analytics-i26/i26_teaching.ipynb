## 1. 全称否定的直接证明
参数顺序为 predicate, subject。A(A,B) 表示 B 包含于 A。空主项允许，所以全称肯定与全称否定可以同真。

```python
from model import certificate, witness_trace, run, diagnose
import json
v=certificate({2},{0,1},{0})
print(json.dumps(v,ensure_ascii=False,sort_keys=True))
print(witness_trace({2},{0,1},{0},0))
```

## 2. 驳倒全称肯定，不等于证明全称否定
C={0} 给出 B 中不属于 A 的见证，但 B 中另一对象1仍可属于A。

```python
v=certificate({1},{0,1},{0})
print('background:',v['reductio_background'])
print('not all B are A:',v['not_all_b_are_a'])
print('no B is A:',v['tempting_universal_negative'])
print('empty C:',certificate({0},{0},set()))
```

## 3. 有限枚举
枚举 n=0..4。出现零反例只覆盖这些赋值；一般有效性仍须正文见证证明。

```python
result=run()
print(json.dumps(result['totals'],ensure_ascii=False,sort_keys=True))
print('truth table rows:',len(result['truth_rows']))
```

## 4. 固定其余前提才可归咎于假设
规则 p且q 蕴含r 固定。只知非r不足以断言非p；背景无模型时不把空集上全称成立误报为有意义证据。

```python
print('fixed q:',diagnose(q=True,r=False))
print('unknown q:',diagnose(r=False))
print('false q:',diagnose(q=False,r=False))
print('inconsistent:',diagnose(p=True,q=True,r=False))
```
