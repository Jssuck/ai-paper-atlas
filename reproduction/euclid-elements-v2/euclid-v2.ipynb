# 《几何原本》V.2：同一对基量，两批整数倍

目标：分别核对每批的倍数，再看为什么合计倍数是两批计数的和。

## 1. 两批倍数可不同

左列统一按长度计，右列统一按面积计，各列分别相加，不做长度与面积的加法。

```python
from fractions import Fraction as F
from equimultiple_sum import Magnitude as M, compose, regression_grid
bases = (M(F(3,2)), M(F(5,2), "area"))
first = (M(3), M(5, "area"))
second = (M(F(9,2)), M(F(15,2), "area"))
c = compose(first, second, bases)
print("两批倍数、合计倍数:", c.first_count, c.second_count, c.total_count)
print("两列总量:", c.left_total.value, c.right_total.value)
```

## 2. 合计一致也可能来自互相抵消的错配

两列都是 6，但每批的左右倍数都不相等。

```python
first, second, bases = (M(2),M(3)), (M(4),M(3)), (M(1),M(1))
print("两列总量:", *(x.value+y.value for x,y in zip(first,second)))
try:
    compose(first, second, bases)
except ValueError as error:
    print("拒绝:", error)
```

## 3. 数值比例相同，不等于整数倍

两批公共比 3/2 和 5/2，合计出现整数也无法恢复前提。

```python
first, second, bases = (M(3),M(6)), (M(5),M(10)), (M(2),M(4))
print("两列合计对各自基量:", *((x.value+y.value)/b.value for x,y,b in zip(first,second,bases)))
try:
    compose(first, second, bases)
except ValueError as error:
    print("拒绝:", error)
print("合法组合与错配拒绝:", regression_grid())
```
