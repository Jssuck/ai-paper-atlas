## 1. 两条路径、一个半径

先生成仅供测试的有理尺度与平移；求解函数只接收顶点。用两个独立方法算圆心，再查验五个顶点到它的距离平方。

```python
from fractions import Fraction as F
import circumcircle as g

_, regular, moved, _ = g.fixtures(1, 1)
center_by_rays = g.center_from_bisectors(regular)
center_by_three = g.circumcenter_three(*regular[1:4])
if center_by_rays != center_by_three:
    raise AssertionError('centers disagree')
regular_radii = g.vertex_squares(regular, center_by_rays)
if regular_radii != (g.q(1),)*5:
    raise AssertionError('unit fixture not certified')
print('two centers agree:', True, '| radius squares:', [str(x.a) for x in regular_radii])
```

## 2. 练习：圆心不变，圆还过五点吗？

把 A 沿原半径推远 10%。先预测两个算法的候选是否改变，再猜哪一个半径平方改变。下一格给出核对；不要把候选圆与证书混为一谈。

```python
moved_radii = g.vertex_squares(moved, center_by_rays)
if g.center_from_bisectors(moved) != center_by_rays or g.circumcenter_three(*moved[1:4]) != center_by_three:
    raise AssertionError('moving only A changed the candidate center')
if moved_radii != (g.q(F(121,100)),)+(g.q(1),)*4:
    raise AssertionError('unexpected moved-vertex certificate')
print('moved A, same candidate:', True, '| radius squares:', [str(x.a) for x in moved_radii])
```

## 3. 正五边形不是共圆的必要条件

这些点从圆的有理参数式取得，边不全等。过任意三个不同顶点的圆仍可检验另两个点。可自行更换构造输入的三个顶点，观察何时不变。

```python
cyclic = g.cyclic_nonregular()
other_center = g.circumcenter_three(*cyclic[1:4])
if len({g.norm(g.sub(b,a)) for a,b in g.edges(cyclic)}) <= 1:
    raise AssertionError('control must be nonregular')
if g.vertex_squares(cyclic, other_center) != (g.q(1),)*5:
    raise AssertionError('nonregular cyclic control is not certified')
print('nonregular and cyclic:', True, '| exact grid (regular, moved A, shifted center):', g.check_grid())
```
