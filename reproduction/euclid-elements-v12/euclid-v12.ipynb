# 《几何原本》V.12：三个同比如何合并

在复现目录下运行。第一格记录同一个探针下逐项和总量的比较符号。

```python
from ratio_sums import COMMON, MIXED, CLOSE, exact_sum, probe_sum, bounded_component_mismatch
print(exact_sum(COMMON))
for m, n in ((1, 1), (3, 2), (2, 1)):
    print(m, n, probe_sum(COMMON, m, n))
```

## 反例：合计相等掩盖逐项差异

合计比由后项加权，不是简单平均。

```python
ratios, aggregate, same = exact_sum(MIXED)
print(ratios, aggregate, same)
print("unweighted:", sum(ratios)/len(ratios))
print(probe_sum(MIXED, 1, 1))
```

## 有界探针漏判

扩大窗口可揭露这组假同比；一般量的证明仍需任意正整数倍。

```python
for limit in (20, 40):
    print(limit, bounded_component_mismatch(CLOSE, limit))
print("exact:", exact_sum(CLOSE))
```
