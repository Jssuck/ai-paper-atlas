# 欧几里得 VII.35：最小公倍数整除任一公倍数

```python
from common_multiple_remainder import close_difference, verify_least_divides
print('least step:',verify_least_divides(12,18,108))
```

## 刻意换成非最小步长

```python
t = close_difference(12,18,72,108)
if t.remainder != 36 or any(t.remainder%d for d in (12,18)):
    raise RuntimeError('remainder check failed')
print('nonleast step:',t)
```

## 任意输入不必有公倍数余数

```python
print('raw division:',divmod(50,36))
try:
    verify_least_divides(12,18,50)
except ValueError as exc:
    print('rejected:',exc)
else:
    raise RuntimeError('unexpected acceptance')
```
