# I.46：指定线段上的正方形
有理坐标教学重构，不是尺规模拟。原证依I.11/I.3/I.31/I.34/I.29与定义22，不能用I.47替代。此文件以标准库compile/exec真实执行，非Jupyter kernel。

```python
from geometry import construct,verify,area
from cases import valid_cases,INVALID
c=construct((0,0),(3,4))
print({k:list(map(str,v)) for k,v in c["points"].items()})
```

边界顺序A-D-E-B。C只标示垂直射线，选择在D以外；它不是正方形顶点。

```python
print("verified",verify((0,0),(3,4),1,c))
print("side squared",c["side_squared"])
print("area",area([c["points"][k] for k in "ADEB"]))
```

同一底边的两侧各能构造。方向参数只允许整数+1或-1。

```python
for s in (-1,1):
    z=construct((0,0),(3,4),s)
    print(s,"D=",z["points"]["D"],"verified",verify((0,0),(3,4),s,z))
```

矩形不足以成为正方形；证书不由构造器自行盖章。

```python
import copy
bad=copy.deepcopy(c)
bad["points"]["E"]=(0,0)
print("tampered accepted",verify((0,0),(3,4),1,bad))
```

有限案例只验证实现行为，不能代替一般证明。

```python
n=sum(verify(A,B,s,construct(A,B,s)) for A,B,s in valid_cases())
r=0
for name,A,B,s in INVALID:
    try: construct(A,B,s)
    except ValueError:r+=1
print("valid",n,"invalid rejected",r)
```
