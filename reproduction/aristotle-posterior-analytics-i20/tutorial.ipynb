## 1 同样的深度，不同的宽度

两端固定且每条链很短，不表示所有不可比较候选合起来也很少。m=1才是全序区间；m大于1是菱形区间。

```python
for m in (1, 2, 8, 32):
    info = diamond(m).interval(0,m+1)
    print(f"m={m}: middle_count={info['middle_count']}, longest_edges={info['longest_edges']}, is_total={info['is_total']}")
```

## 2 有限枚举的分母

固定编号顺序0<1<2<3的6个可能正向边，有64种子集；多个原始图具有相同传递闭包。以下区间次数保留这些重复。它不是全部任意标签偏序的计数。

```python
print(json.dumps(enumerate_forward_dags(4), ensure_ascii=False, indent=2))
```

## 3 相邻前缀不消除后面的细分

把允许的有序域设为{-1}与[0,1]内有理数的并集：-1到0在这个域中相邻；但0到1仍可不断取中点。程序只输出8次插入的有限前缀，无法运行完无限过程。无限可继续需要正文中的一般代数论证。

```python
values = buffered_prefix(8)
print(', '.join(map(str, values)))
print('gap to 1:', 1-values[-2])
print('next midpoint:', (values[-2]+1)/2)
```
