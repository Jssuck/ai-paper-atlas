# 相交弦：同一圆、同一交点

以精确有理数区分两条弦的乘积、弦中点见证和有向扩展。

## 圆内三例

先核验两直径、单直径且垂直，以及两弦都不经过圆心。

```python
from chord_products import EXAMPLES, audit, sweep, require
for key, expected in (
    ("center two diameters",25),
    ("one diameter perpendicular",16),
    ("neither diameter",7),
):
    z = audit((0,0),5,*EXAMPLES[key])
    require(z["iii35_premises"] and z["unsigned_products"] == (expected,expected),key)
    for m in z["midpoints"]:
        require(m["half_sq"]-m["offset_sq"]==expected,"midpoint witness")
    print(key, z["point"], z["unsigned_products"])
```

## 回归对照

圆外的有向积依然相等，但不满足 III.35 的内部条件；失去共圆条件还可使乘积不等。

```python
out = audit((0,0),5,*EXAMPLES["exterior supports"])
require(not out["iii35_premises"] and out["directed_products"] == (24,24),"outside")
print("outside signed products", out["directed_products"])
bad = audit((0,0),5,*EXAMPLES["off-circle endpoint"])
require(not bad["same_circle"] and bad["unsigned_products"] == (16,12),"off circle")
print("off-circle unsigned products", bad["unsigned_products"])
```
