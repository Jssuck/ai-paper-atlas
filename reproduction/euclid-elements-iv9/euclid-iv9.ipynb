# 《几何原本》IV.9：给定正方形的外接圆

以有理坐标重建作图。目标是分别核验输入为正方形、对角线交点为圆心，以及四个顶点都在同一个圆上。从复现目录打开本文件。

## 1. 旋转正方形
使用有理起点。半径平方应为 10，不需开方。

```python
from circumcircle import (circumscribe, diagonal_intersection, examples, rectangle_circumcircle,
                          run_grid, show_point, verify_square_circle, vertex_distances2)
square, rectangle, rhombus = examples()
center, radius2 = circumscribe(square)
print("vertices:", tuple(map(show_point, square)))
print("center, radius^2:", show_point(center), str(radius2))
print("certified:", verify_square_circle(square, center, radius2))
```

## 2. 区分直角和等边
矩形虽然被 IV.9 严格规格拒绝，却仍有外接圆；菱形对角线交点到四个顶点不等距。

```python
o, r2 = rectangle_circumcircle(rectangle)
print("rectangle:", show_point(o), str(r2), tuple(map(str, vertex_distances2(o, rectangle))))
o = diagonal_intersection(rhombus)
print("rhombus:", show_point(o), tuple(map(str, vertex_distances2(o, rhombus))))
```
