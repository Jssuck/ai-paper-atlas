# 欧几里得 VII.16：交换乘数次序

```python
from commute_products import commute, exact_measure, modern
from itertools import product
t = commute(3,4)
print(t)
```

## 用网格转置另行核对两组和

```python
grid = set(product(range(t.a),range(t.b)))
if {(j,i) for i,j in grid} != set(product(range(t.b),range(t.a))) or len(grid) != t.ab or len(grid) != t.ba:
    raise RuntimeError('grid mismatch')
print('cells:', len(grid))
```
