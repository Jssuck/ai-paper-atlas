# III.15：弦长次序与圆心距离
有限精确有理实验，不是一般证明。标准库compile/exec，非Jupyter kernel。

```python
from geometry import *
c=Circle((0,0),5)
q=chord(c,(-4,3),(4,3))
print('length squared:',q['length2'],'distance squared:',q['distance2'])
```

# 独立投影
先投影得到垂足，再检验中点和平方关系，不用待证关系筛输入。

```python
print('foot:',q['foot'],'parameter:',q['projection_parameter'])
print('half squared + distance squared:',q['half_length2']+q['distance2'])
```

# 次序反转的两个方向
正长度平方严格单调。输出1表示第一条更长/更远，-1表示更短/更近。

```python
x=compare(c,(-4,3),(4,3),(-3,4),(3,4))
print(x['length_order'],x['distance_order'],x['order_reversal_holds'])
y=compare(c,(-3,4),(3,4),(-4,3),(4,3))
print(y['length_order'],y['distance_order'],y['order_reversal_holds'])
```

# 直径最大与等号条件
所有直径都并列最大，不能说每条直径严格大于其他直径。零距不当作非退化三角形。

```python
q=chord(c,(-5,0),(5,0))
print(q['length2'],4*q['radius2'],q['distance2'],q['diameter'])
print(compare(c,(-5,0),(5,0),(0,-5),(0,5))['length_order'])
```
