## 输入与几何资格
a、b为AC、CB的基系数；实际长度还乘基向量长度，均严格正，等分合法。

```python
from fractions import Fraction as Q
from geometry import construct, verify
c=construct()
r=verify(c)
print(c["points"])
```

## 余形同加共同正方形
AG=GE；各加CF得到AF=CE。

```python
print("AG=",r["AG"],"GE=",r["GE"],"CF=",r["CF"])
print("AF=",r["AF"],"CE=",r["CE"])
```

## 重叠必须计两次
两个矩形的面积和并非并集：并集为曲尺形，交集为CF。

```python
print("sum",r["AF"]+r["CE"],"union",r["gnomon"],"overlap",r["CF"])
print("whole+CF",r["whole"]+r["CF"],"2AF+DG",2*r["AF"]+r["DG"])
```

## 有理输入及缩放

## 拒绝不合构造的证书
满足最终面积恒等式不等于所有点资格成立；篡改G必须拒绝。

```python
bad=construct()
bad["points"]["G"]=(Q(3),Q(3))
try:
    verify(bad)
except ValueError as e:
    print("rejected:",e)
else:
    raise RuntimeError("tampering accepted")
```
