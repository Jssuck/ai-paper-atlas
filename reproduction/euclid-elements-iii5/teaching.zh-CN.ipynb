# III.5：同心与公共点
标准库精确教学重构；下面保存的是compile/exec的stdout，不是Jupyter内核输出，execution_count保持null。有限案例不代替一般证明。

```python
from geometry import certificate, circle_relation, verify, EXAMPLES, grid
c = certificate(*EXAMPLES['crossing_upper'])
print(c['classification'], c['center_distance_squared'])
```

一个公共点不区分相切与相交。

```python
for name in ('crossing_lower','external_tangent','internal_tangent'):
    c = certificate(*EXAMPLES[name])
    print(name, c['classification'], c['linear_lhs'], c['linear_rhs'])
```

重合圆有公共点且同心，说明必须保留两圆互异条件。

```python
c = certificate(*EXAMPLES['coincident'])
print(c['classification'], c['same_center'], c['distinct_circles'])
print(circle_relation((0,0),25,(0,0),9))
print(circle_relation((0,0),1,(5,0),1))
```

验证器区分布尔值与数值；改动未获证据支持的分类会被拒绝。

```python
c = certificate(*EXAMPLES['crossing_upper'])
print('original', verify(c))
c['same_center'] = 0
print('bool replaced by integer', verify(c))
```

有限变换网格只检查实现。负尺度是中心缩放，单坐标变号才是这里的镜像。

```python
count = sum(verify(certificate(*a)) for a in grid())
print('verified finite configurations', count)
```
