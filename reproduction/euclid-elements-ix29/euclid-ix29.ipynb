# IX.29：奇数乘奇数
每份隔出单位，再配对并放回余下的一个单位。

```python
from odd_product import odd_times_odd
print('3 copies of 5:',odd_times_odd(3,5))
print('5 copies of 3:',odd_times_odd(5,3))
```

```python
for a,b in ((4,5),(3,6),(4,6)):
    try:
        odd_times_odd(a,b)
    except ValueError as exc:
        print((a,b),a*b,'even substitution excluded:',exc)
    else:
        raise RuntimeError('Expected rejection')
```

```python
for a,b in ((1,5),(3,1),(0,5)):
    try:
        odd_times_odd(a,b)
    except ValueError as exc:
        print((a,b),a*b,'modern boundary:',exc)
    else:
        raise RuntimeError('Expected rejection')
```
