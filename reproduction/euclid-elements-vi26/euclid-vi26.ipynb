# 《几何原本》VI.26：相似、同向的角格为何沿同一对角线

```python
from fractions import Fraction as Q
from same_diagonal import U,V,area,construct,cross,sub,side_ratios2

a,b,c,d,e,f,g,h,k=construct(U,V,Q(1,2),Q(1,3))
print("F,H,K", f,h,k)
print("两方向对应边平方比", side_ratios2(a,b,d,e,g))
print("对角线与 AF 的行列式", cross(sub(c,a),sub(f,a)))
```

分别检查位于对角线两侧和恰在对角线上的情形。

```python
for s in (Q(1,4),Q(1,3),Q(1,2)):
    a,b,c,d,e,f,g,h,k=construct(U,V,s,Q(1,3))
    print("s",s,"F=H?",f==h,"行列式",cross(sub(c,a),sub(f,a)))
```

只相等的面积或公共角并不足以推出同一对角线。

```python
for s,t in ((Q(1,4),Q(1,4)),(Q(1,2),Q(1,8))):
    a,b,c,d,e,f,g,h,k=construct(U,V,s,t)
    print("面积",area((a,e,f,g)),"F 在 AC 上?",cross(sub(c,a),sub(f,a))==0)
```
