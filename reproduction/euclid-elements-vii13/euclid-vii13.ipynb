# 欧几里得 VII.13：换比

比较初始方向与换比方向的四种分支搭配。

```python
from alternate_ratio import alternate, demo
from fractions import Fraction
demo()
```

## 用原输入独立核对

```python
a,b,c,d = 8,12,20,30
r = Fraction(a,c)
if r != Fraction(b,d) or r != Fraction(*alternate(a,b,c,d).crossed):
    raise RuntimeError('ratio mismatch')
print('crossed:', r, 'reversed:', Fraction(c,a), Fraction(d,b))
```

## 反例：原比较已不成比例

```python
a,b,c,d = 8,12,20,32
try:
    alternate(a,b,c,d)
except ValueError as exc:
    print('rejected:', exc, '; crossed:', Fraction(a,c), Fraction(b,d))
else:
    raise RuntimeError('unexpected acceptance')
```
