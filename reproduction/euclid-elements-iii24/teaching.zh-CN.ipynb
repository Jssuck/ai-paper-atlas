## 1. 从点积而不是结论测角
a=2,k=3/2，圆心(0,3/2)，半径5/2，取射线斜率1。练习：为什么cot=3/4不是3/2？答案：半底长a也进入分母。

```python
from geometry import *
s=Segment(2,F(3,2)); p=s.arc_point(1)
print("P:",p,"cot:",s.measured_cot(p),"radius²:",s.radius2)
```

## 2. 刚体对齐不预设相似
目标基底采用有理旋转与反射。映射以两个正交框架定义，保持距离和所选侧；等底保证两端对应。映射构造未检查k相等，最后才核支撑圆是否一致。

```python
t=Segment(2,F(3,2),(5,-1),(F(3,5),F(4,5)),(F(4,5),F(-3,5)))
print("ends:",t.endpoints)
print("mapped P:",s.align_point_to(t,p))
print("same location before:",s.coincides(t),"circle matches after:",s.aligned_circle_matches(t))
```

## 3. 镜像不是不等的反例
反射改变选侧与圆心位置，但保持大小与所容角。原圆另侧则通常角互补，两种操作不能混淆；半圆是例外。

```python
m=Segment(2,F(3,2),n=(0,-1))
o=Segment(2,F(-3,2),n=(0,-1))
print("mirror:",s.similar(m),s.coincides(m))
print("same circle other side:",s.center==o.center,s.angle_cot,o.angle_cot)
```

## 4. 条件和边界
放大两倍保角但底不等，不能用刚体映射。练习：先构造映射就说明结论吗？答案：否，等底不同k也可以对齐底，却不对齐圆心。

```python
print("unequal k, equal bases:",s.aligned_circle_matches(Segment(2,2)))
print("scaled similar:",s.similar(Segment(4,3)))
try:
    s.align_point_to(Segment(4,3),p)
except ValueError as e:
    print("rejected:",e)
```
