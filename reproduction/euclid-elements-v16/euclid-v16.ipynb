```python
from ratio_alternation import Magnitude, demo, ratio, replay
demo()
```

第一步同时观察两列比较符号。共同倍数是左原始比内的 m 和右原始比内的 n，两者并不要求相等。练习：若四个数为 3、5、6、10，取 m=4、n=2，输出符号是什么？

```python
answer = replay(*map(Magnitude, (3, 5, 6, 10)), 4, 2)
print('alternate ratios and signs:', answer[2:])
```

答案为两个等号：12 对 12，20 对 20。常见陷阱：数值之商相同不代表跨类别的新比合法。最后尝试长度比和面积比；拓展练习可给不同同类输入寻找等号探针。

```python
a, b, c, d = Magnitude(2), Magnitude(3), Magnitude(4, 'area'), Magnitude(6, 'area')
print('legal internal ratios:', ratio(a, b), ratio(c, d))
try:
    replay(a, b, c, d, 2, 1)
except ValueError as error:
    print('refused:', error)
else:
    raise RuntimeError('ill-typed alternation was not rejected')
```
