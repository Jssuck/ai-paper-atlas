# 卷 V 定义 1–7：以整数倍比较两个比

适合学过分数与平方的读者。目标：区分整数度量、有限证据和全称同比。
1. 检查一个量能否整数量取另一个量。
2. 用精确平方判据寻找符号差异。
3. 判断哪个差异也能证明一个比更大。


## 1. 「能超过」不等于「能度量」

虽然 √2 不是 2 的整数个等分，但任一正长度的倍数仍可超过另一个正长度。


```python
from ratio import Q2, Magnitude, examples, measure, exceeding_multipliers, witness
root = Magnitude(Q2(0, 1), 'length')
two = Magnitude(Q2(2), 'length')
four_roots = Magnitude(Q2(0, 4), 'length')
print('√2 measure of 2:', measure(root, two))
print('√2 measure of 4√2:', measure(root, four_roots))
print('strict multipliers:', exceeding_multipliers(root, two))
```

## 2. 有限一致并非同比

六格上限内没找到反例，扩到七便出现严格大于与等号的差异。


```python
surd, rational, high, low = examples()
print('N=6:', witness(surd, rational, 6))
print('N=7:', witness(surd, rational, 7))
print('directional witness:', witness(surd, rational, 7, greater=True))
```

## 3. 练习：两个比不同，给出的见证便能证明谁大吗？

试比较 101:100 与 102:101：为什么上限 101 找到了差异，却没有得到左侧更大的定向见证？答案提示：符号对 (0,−) 不满足第一项严格大于；扩大一格后可用 (+,0)。可继续扫描更狭窄的有理区间，但不要把扫描失败当成同比证明。


```python
for limit in (100, 101, 102):
    print(limit, witness(high, low, limit), witness(high, low, limit, greater=True))
```
