# I.37 从两倍图形回到三角形

现代有理坐标教学，六代码格标准库compile/exec非kernel。

## 1 共同底与共同上支持线

```python
from geometry import make_certificate, verify_certificate
from cases import EXAMPLE, build, positives, negatives
c = make_certificate(EXAMPLE)
print(c['area'])
```

## 2 补成两平行四边形
E=A−(C−B)，F=D+(C−B)，对角线AB、DC。

```python
print(c['vertices'])
print(c['parallel_checks'])
print(c['halves'])
```

## 3 顶点沿平行线移动
同顶点重合为扩展边界，原图AD名称需另选支持线标记。

```python
for t in [-3, '-1/2', 0, '1/2', 3]:
    q = make_certificate(build(t))
    print(t, q['top_case'], q['area'])
```

## 4 异侧等高反例
面积可以仍相等，但不满足同一上支持线条件。

```python
bad = dict(EXAMPLE)
bad['D'] = [9,-3]
try:
    make_certificate(bad)
except ValueError as e:
    print('直接拒绝:', str(e))
else:
    raise RuntimeError('negative accepted')
```

## 5 有限精确输入族

```python
n = 0
for d in positives():
    if not verify_certificate(d, make_certificate(d)):
        raise RuntimeError('failed')
    n += 1
m = 0
for d in negatives():
    try:
        make_certificate(d)
    except ValueError:
        m += 1
    else:
        raise RuntimeError('negative accepted')
print('正例/直接拒绝负例:', n, m)
```

## 6 证书安全与边界

```python
import copy
bad = copy.deepcopy(c)
bad['area']['ABC'] = True
print('布尔攻击通过:', verify_certificate(EXAMPLE,bad))
shift = {k:[x+1 for x in v] for k,v in EXAMPLE.items()}
print('移绑通过:', verify_certificate(shift,c))
print('有限实跑不证明一般定理；compile/exec非kernel。')
```
