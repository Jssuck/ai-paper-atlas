## 6. 有限运行与边界
120个严格内部有理构型与12个非法输入；不是一般定理证明。D落在BC上时夹角可成为平角，但严格内部前提已经失效。

```python
ps=positive_cases();ns=invalid_cases()
passed=sum(accepted(*d,construct(*d)) for d in ps)
rejected=0
for d in ns:
    try:construct(*d)
    except ValueError:rejected+=1
print({'positive':passed,'invalid_rejected':rejected})
if passed!=len(ps) or rejected!=len(ns):raise RuntimeError('case failure')
```
