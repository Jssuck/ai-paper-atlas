# 欧几里得 VII.29：质数若不整除一数，就与它互素

```python
from prime_nondivisor import common_witness, inspect, prove
print('7, 15:',prove(7,15))
```

## 有公度时，其值只能是质数本身

```python
print('7 measures 14:', common_witness(7,14,7))
try:
    prove(7,14)
except ValueError as exc:
    print('rejected:',exc)
else:
    raise RuntimeError('unexpected acceptance')
```

## 合数不整除时，不能据此决定是否互素

```python
for p,n in ((6,9),(8,15)):
    print(p,n,inspect(p,n))
```
