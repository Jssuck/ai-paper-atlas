## 方向不可混淆

```python
xs = (24,12,6,3)
if not all(xs[i]%xs[j] == 0 and not flag for i,j,flag in inspect(xs).forward):
    raise RuntimeError('direction mismatch')
print('backward measures in all six pairs')
```

## 非等比输入不适用

```python
try:
    inspect((8,12,19))
except ValueError as exc:
    print('rejected:',exc)
else:
    raise RuntimeError('unexpected acceptance')
```
