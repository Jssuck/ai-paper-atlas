# 《几何原本》VI.19：相似三角形的面积平方比

在本文件所在目录运行。用现代有理坐标分开核对第三比例点、面积比和相似条件。

```python
from fractions import Fraction as Q
from triangle_areas import worked, area, third_point, side_squares, transform, signed_double_area

abc, deff, g = worked()
a, b, c = abc
print("G =", tuple(map(str, g)))
print("ABC, DEF, ABG 面积 =", [str(x) for x in (area(*abc), area(*deff), area(a,b,g))])
print("ABC : DEF =", area(*abc) / area(*deff))
```

辅助点始终在从 B 经过 C 的射线上，可能在 C 前面、恰好等于 C 或在 C 后面。

```python
for k in (Q(1,2), Q(1), Q(3,2)):
    point = third_point(b,c,k)
    print("EF/BC =", k, "G =", tuple(map(str,point)))
mirrored = tuple(transform(p, Q(3,2), (1, -3), (Q(3,5), Q(4,5)), reflected=True)
                 for p in abc)
print("镜像有向二倍面积 =", signed_double_area(*mirrored), "普通面积 =", area(*mirrored))
```

仅有底边比而没有相似条件，不能保证面积按平方比改变。

```python
_, e, f = deff
wrong = ((2,3), e, f)
print("面积比 DEF′/ABC =", area(*wrong)/area(*abc))
print("平方底比 =", Q(side_squares(*wrong)[1], side_squares(*abc)[1]))

wrong_inverse = ((0,Q(9,2)), e, f)
print("面积比恰好等于平方底比 =", area(*wrong_inverse)/area(*abc))
print("但对应边 DE²/AB² =", Q(side_squares(*wrong_inverse)[0], side_squares(*abc)[0]))
```
