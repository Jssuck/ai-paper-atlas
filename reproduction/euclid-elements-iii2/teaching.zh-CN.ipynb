# Euclid III.2：弦的内点为何严格在圆内

在本目录运行。原文通过等腰底角、外角与大角对大边归谬（I.5、I.16、I.19）；本 Notebook 是另列的现代有理坐标教学重构，不冒充原证或自动证明器。圆心与正半径平方已经给定。

```python
from fractions import Fraction as Q
from geometry import certificate, verify_certificate
r=certificate((0,0),25,(-3,4),(3,4),Q(1,2))
print("E =",r["E"],"distance squared =",r["distance_squared"],"gap =",r["gap"])
```

## 1. 恒等式到严格性
半径平方减去内点距离平方等于 t(1-t) 乘弦长平方。端点互异及 0<t<1 缺一不可。这里的圆内是距离平方小于半径平方。

```python
print("identity:",r["distance_squared"]+r["gap"]==r["radius_squared"])
print("strict gap:",r["gap"]>0,"verified:",verify_certificate(r))
```

## 2. 直径的中点
当 E 恰为圆心，原来的角图会退化。距离平方方法仍有定义，得到零距离和正间隙。应在文字证明单列此分支，而非给零射线命名角。

```python
d=certificate((0,0),25,(-5,0),(5,0),Q(1,2))
print("branch =",d["branch"],"distance squared =",d["distance_squared"],"gap =",d["gap"])
```

## 3. 端点不是严格内点
闭弦包含圆周端点，只有开弦严格在内。把端点参数误说为 0<t<1 会使“全部点严格在内”成为假命题。

```python
for t in (0,1):
    e=certificate((0,0),25,(-3,4),(3,4),t)
    print("t =",t,"branch =",e["branch"],"gap =",e["gap"])
```

## 4. 输入与伪证书
接口拒绝外插参数、重合端点和非精确类型。不能靠把浮点负误差截为零来认证严格性。证书所有字段都重新计算。

```python
bad=dict(r);bad["gap"]=Q(0)
print("forged gap accepted:",verify_certificate(bad))
for t in (2,True):
    try: certificate((0,0),25,(-3,4),(3,4),t)
    except (ValueError,TypeError) as e: print(type(e).__name__,str(e))
```
