# X.29：原线段与差平方的边

```python
from fractions import Fraction as F
from ratio_witness import construct
for h,n,m in [(1,3,2),(2,3,2),(F(7,5),2,1)]:
    w=construct(h,n,m)
    print(w, 'reconstructed h:',w.l_squared+w.g_squared)
```

对照：平方差是整数平方。

```python
try:
    construct(1,5,3)
except ValueError as exc:
    print(exc)
else:
    raise RuntimeError('square control not rejected')
```
