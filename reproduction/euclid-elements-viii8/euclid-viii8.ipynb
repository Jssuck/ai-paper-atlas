# 欧几里得 VIII.8：把已知等比链迁到同比的端点

```python
from transfer_geometric_chain import transfer
from fractions import Fraction
x = (16,24,36,54)
t = transfer(x,24,81)
if t.chain != tuple(Fraction(y*24,x[0]) for y in x):
    raise RuntimeError('rescaling mismatch')
print(t)
```

## 只有端点不能确定项数

```python
for row in ((4,16,64),(4,8,16,32,64)):
    print(transfer(row,4,64).chain)
```

## 相同差值不是相同比值

```python
try:
    transfer(x,24,62)
except ValueError as exc:
    print('rejected:',exc)
else:
    raise RuntimeError('unexpected acceptance')
```
