# I.43：余形等积
原创有理坐标教学。以下五格由标准库compile/exec真实顺序执行，非Jupyter kernel；有限运行不证明一般定理。

```python
from fractions import Fraction as Q
from geometry import example,make_certificate,verify,split,area
d=example(); c=make_certificate(d)
print(c['points'])
```

两余形分别为E-B-G-K与H-K-F-D；不是只有一条线段BK、KD。

```python
print('余形面积:',c['areas']['EBGK'],c['areas']['HKFD'])
print('关于对角线两块:',c['areas']['AEKH'],c['areas']['KGCF'])
```

原证接口：大三角形等积，分别减去相等的两小三角形。

```python
a=c['areas']
print(Q(a['ABC'])-Q(a['AEK'])-Q(a['KGC']))
print(Q(a['ACD'])-Q(a['AHK'])-Q(a['KFC']))
```

离开对角线后，两补形可不等积。这只是另一网格，不是原命题的合法输入。

```python
p=split((0,0),(6,0),(0,3),Q(1,3),Q(2,3))
print(area([p[x] for x in 'EBGK']),area([p[x] for x in 'HKFD']))
try:
 make_certificate({k:p[k] for k in 'ABCDK'})
except ValueError as e:
 print(type(e).__name__,str(e))
```

证书必须绑定输入；核验器共享构造实现，独立技术审查另提供检查。

```python
print('真实证书:',verify(d,c))
c['areas']['EBGK']='5'
print('篡改证书:',verify(d,c))
try:
 make_certificate(example(t=Q(0)))
except ValueError as e:
 print('退化端点:',str(e))
```
