## 1. 输入契约
E位于AB和CD之间；两条支撑直线不同。坐标只收整数或有理数字符串，拒绝布尔与浮点。

```python
from geometry import certificate, verify, angle_ledger
from cases import positives, run
from copy import deepcopy
data = next(positives())
print(data)
```

## 2. 检查反向比例
生成证书先检查两组反向射线和非零行列式，不把结论当作输入。

```python
claim = certificate(data)
print('opposite scales:', claim['opposite_scales'])
print('cross:', claim['cross_AE_CE'])
print('verified:', verify(data, claim))
```

## 3. 不使用acos的角描述
点积与行列式绝对值经共同L1归一化给出普通角方向描述；这不是角度数值。

```python
print('CEA, AED, DEB, BEC:')
print(claim['angles_CEA_AED_DEB_BEC'])
print('vertical pairs:', claim['vertical_pairs_equal'])
```

## 4. 原证加法结构的单独重构
以一直角为1，x+y=2和y+z=2同减y。此模型不声称从坐标求出分数角度。

```python
print(angle_ledger('2/3', '4/3', '2/3'))
```

## 5. 证书不能接受真假值与整数混淆
Python一般相等会把True当作1；本核验器使用严格JSON结构。

```python
bad = deepcopy(claim)
bad['schema_version'] = True
print('bool substituted for integer accepted:', verify(data, bad))
bad = deepcopy(claim)
bad['vertical_pairs_equal'][0] = 1
print('integer substituted for bool accepted:', verify(data, bad))
```
