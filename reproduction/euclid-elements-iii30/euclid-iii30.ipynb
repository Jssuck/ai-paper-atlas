# 《几何原本》III.30：选对平分指定弧的射线

把本文件和 `bisect_arc.py` 放在同一目录。用相同端点分别指定短弧和长弧，检查正确交点和错误交点。

```python
from bisect_arc import SelectedArc, square, sub

for direction in (1, -1):
    arc = SelectedArc((0, 0), (3, -4), (3, 4), direction)
    d, opposite = arc.endpoints_on_bisector()
    print("方向", direction, "弦中点", tuple(map(float, arc.chord_midpoint())))
    print("所选交点", d, "反向交点", opposite)
    print("全扫角", round(arc.full_sweep(), 9),
          "所选两段", [round(x, 9) for x in arc.parts(d)])
```

仅检查到两端点距离相等不能排除反向交点；按指定行进方向经过反向交点会多绕一整圈。

```python
arc = SelectedArc((0, 0), (3, -4), (3, 4), 1)
d, other = arc.endpoints_on_bisector()
for candidate in (d, other):
    print(candidate, "两条弦长的平方",
          [square(sub(candidate, endpoint)) for endpoint in (arc.a, arc.b)],
          "经过该点的方向扫角总和", round(sum(arc.parts(candidate)), 9))
```

两个半圆扫角一样大，但它们是不同的指定弧。只提供直径两端点并不足够。

```python
for direction in (1, -1):
    arc = SelectedArc((1, 1), (6, 1), (-4, 1), direction)
    d, _ = arc.endpoints_on_bisector()
    print(direction, "平分点", d, "两段扫角", [round(x, 9) for x in arc.parts(d)])
```
