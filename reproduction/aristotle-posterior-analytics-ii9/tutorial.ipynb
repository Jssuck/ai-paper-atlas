## 证书与前提承担
每一步只能使用已知前提。将a加入基底使推导可启动，不是证明a本身。


```python
chain=((frozenset('a'),'b'),(frozenset('b'),'c'))
closure,cert=derive(atoms,{'a'},chain)
print('闭包:',sorted(closure),'证书:',cert)
print('证书有效:',check_certificate(atoms,{'a'},chain,cert,'c'))
print('移除基底:',check_certificate(atoms,set(),chain,cert,'c'))
```

## 定义与存在
用两个谓词同外延近似一个定义约束，这不是亚里士多德本质定义的充分条件。空外延反例保留。


```python
print('空解释:',definition_case({0},set(),set()))
print('有见证:',definition_case({0},{0},{0}))
print('均存在但不同外延:',definition_case({0,1},{0},{1}))
```
