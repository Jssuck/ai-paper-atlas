精确有理数教学核验。代码在相同目录运行。保存输出来自标准库compile/exec，非Jupyter内核。有限样本不是一般证明。

```python
from geometry import examples, verify, grid, scalar, point, sub, norm
cs=examples()
print('named examples:', len(cs))
print('all valid:', all(verify(c) for c in cs.values()))
```

## 两条均非直径，但一条弦确实被平分
先预测midpoint_AC、midpoint_BD，再执行。

```python
c=cs['only_AC_midpoint']
print('E:', c['E'])
print('parameters:', c['t'], c['u'])
print('midpoints:', c['midpoint_AC'], c['midpoint_BD'])
print('original domain:', c['theorem_applies'])
```

## 证明接口：两方向残差不能同时消失
这只是现代坐标恒等式检查，不是原文的直角反证。

```python
print('direction determinant:', c['det'])
print('center residuals:', c['dot_v'], c['dot_w'])
z=c['input'];E=point(c['E'])
print('EB squared:', norm(sub(point(z['B']),E)))
print('ED squared:', norm(sub(point(z['D']),E)))
```

## 有限变换检查与练习
练习：为什么not(A and B)不等于not A and not B？答案：第二格的单平分案例满足前者，不满足后者。

```python
rows=grid()
print('finite cases:', len(rows))
print('valid:', all(verify(c) for c in rows))
print('both-midpoint cases:', sum(c['both_midpoints'] for c in rows))
```
