# 《几何原本》VI.27：固定缺块形状时的最大面积

```python
from maximal_defect import construct,area,loss_pieces,shape

for t in (2,5,8):
    applied,defect=construct(10,t,2)
    print("缺块底",t,"应用面积",area(applied),"缺块面积",area(defect))
```

比较中点构造与两侧缺块；把斜形统一剪切后，净损失表现为角上的小块。

```python
for t in (2,5,8):
    upper,side,corner=loss_pieces(10,t,2)
    print("缺块底",t,"上条、边条、小角",upper,side,corner,"净损失",upper-side)
```

让高度自由，则问题已经改变，中点上界不必成立。

```python
print("固定形状的中点面积",area(construct(10,5,2,0)[0]))
print("应用底长 8、高度独立选 7 时的面积",area(shape(8,7,0,0)))
```
