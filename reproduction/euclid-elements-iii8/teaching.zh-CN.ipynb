```python
from fractions import Fraction as F
from geometry import *
c=certificate(5,13,(3,4))
print('distance squared',c['squared_distance'])
print('bounds',c['minimum_squared'],c['maximum_squared'])
print('branch',c['branch'])
```

暴露近弧不是朝向外点的整个半圆；阈值为dx=r²。

```python
for r,d,p in [(5,6,(3,4)),(3,5,(F(9,5),F(12,5)))]:
    c=certificate(r,d,p)
    print('point',p,'branch',c['branch'],'horizon',c['horizon'])
```

同一视线可以命中两个不同长度的点，因此不能把视角作为全半圆的一一排序。

```python
s=secant(5,13,(3,4))
print('parameters',s['parameters'])
print('points',s['points'])
print('squared lengths',s['squared_lengths'])
print('squared product',s['squared_lengths'][0]*s['squared_lengths'][1])
```

等长点是轴镜像；最短与最长端点只有一个，不得声称所有长度都恰有两个点。

```python
for value in (63,64,116,324,325):
    print(value,equal_shell(5,13,value))
```
