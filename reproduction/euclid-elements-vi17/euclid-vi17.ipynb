# VI.17：中项的正方形

比较同一中项的两组比例与两块坐标图形。

```python
from fractions import Fraction as Q
from ratio_chains import compare, rational_middle
r = compare(9, 12, 16)
print('ratios:', *(str(q) for q in r.ratios))
print('areas:', *(str(q) for q in r.areas))
```

## 只更换中项

外项不变，但算术平均不是几何平均。

```python
r = compare(9, Q(25, 2), 16)
print('arithmetic candidate areas:', *(str(q) for q in r.areas))
print('proportional:', r.proportional, 'equal area:', r.equal_area)
```

## 有理根与无理根

None 只代表无法用 Fraction 表示，不表示几何平均不存在。

```python
print('rational middle of 2/3 and 3/8:', rational_middle(Q(2, 3), Q(3, 8)))
print('rational middle of 1 and 2:', rational_middle(1, 2))
print('the positive geometric mean still exists; its square is 2')
```
