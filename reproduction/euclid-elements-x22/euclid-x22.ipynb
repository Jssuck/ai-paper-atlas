# X.22：沿底边配置中项面积

```python
from fractions import Fraction as F
from ratio_witness import apply_medial
for q,b in [(6,2),(6,4),(F(16,15),F(2,3))]:
    w=apply_medial(q,b)
    print(f'q={q}, b={b}:',w,'equal area:',w.width_squared*b==q)
```

平方数对照不在本模型的有效输入域。

```python
try:
    apply_medial(F(9,16),3)
except ValueError as exc:
    print(exc)
else:
    raise RuntimeError('unexpected square input accepted')
```
