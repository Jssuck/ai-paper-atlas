模型与原图点名

```python
from fractions import Fraction as Q
from geometry import construct, verify, summary
a=((0,0),(6,0),(0,2),[Q(1,6),Q(1,2)])
c=construct(*a)
print(c["lower"])
print(c["upper"])
```

先核验几何证书，再读取面积

```python
print(verify(*a,c))
print(summary(*a))
```

斜置矩形与不同切点

```python
print(summary((1,-2),(4,2),(-3,1),[Q(1,5),Q(3,5)]))
```

面积相等不够：换序破坏绑定

```python
bad=dict(c)
bad["tiles"]=c["tiles"][::-1]
print(verify(*a,bad))
```
