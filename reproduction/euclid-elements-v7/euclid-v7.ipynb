```python
from fractions import Fraction as F
from equal_ratio_probes import Quantity as Q, probes, verify, equality_witness, finite_grid

a,b,c = Q(2),Q(2),Q(3)
for p,q in ((1,1),(3,2),(2,1),(2,3)):
    print((p,q), verify(a,b,c,p,q))
```

```python
t=Q(F(17,16)), Q(F(18,17)), Q(1)
hits=[(p,q) for p in range(1,9) for q in range(1,9) if any(x!=y for x,y in probes(*t,p,q))]
print("8 × 8 网格内的区别见证:", hits)
assert hits == []
print("构造见证:", equality_witness(*t))
print("交换下标的反向见证:", probes(*t,17,16)[1])
```
