## 外侧：开圆盘不交与一个共有边界点
数值分离判据的等价性由正文证明，未假设唯一性。

```python
b=Circle((8,0),3)
print(contact_certificate(a,b,(5,0),'external'))
```

## 内侧：真包含与共有边界点
排除重合圆；验证输入点与唯一候选公式相同。

```python
inner=Circle((2,0),3)
print(contact_certificate(a,inner,(5,0),'internal'))
print(same_contact(a,inner,(5,0),(5,0),'internal'))
```

## 两交点不是相切
弦中点严格在两个开圆盘内，直接否定外侧分离。这个事实本身不排除内含，内含情形需另一条论证。

```python
crossing=Circle((6,0),5)
print(chord_midpoint_certificate(a,crossing,(3,4),(3,-4)))
print('proper containment:',contains(a,crossing))
```
