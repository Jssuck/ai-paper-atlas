# Euclid I.9：选对等边顶点，还要核查内射线

原创教学坐标模型，不是完整尺规引擎。一般证明见正文；有限枚举不证明定理。所有几何判定在 Q(√3) 中精确计算。请从本目录顺序执行。

```python
from geometry import *
a,d,e=canonical(1,1)
f=construct(a,d,e)
print("F =", tuple(str(x) for x in f))
print(certificate(a,d,e,f))
```

## 为什么“夹角相等”还不够？
两等边顶点中，同侧顶点可能仍在内射线上，也可能等于 A，或位于反向射线上。下列三例分别展示三种情况。不能把“同侧”一概写成“外角平分线”。

```python
for u in (Q3(2),S,Q3(1)):
    a,d,e=canonical(u,1)
    wrong=point(u-S,0)
    checks=certificate(a,d,e,wrong)
    print("u =", str(u), "F =", tuple(str(x) for x in wrong))
    print(checks)
```

## 不等距截取破坏证明前提
