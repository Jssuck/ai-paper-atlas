# 《几何原本》III.33：把角约束变成弓形

在非退化线段上构造一段圆弧。角由非零射线 (u,v) 与水平方向给出，要求 v>0。先试三分支，再选错弧做对照。不求浮点角度，不需要第三方依赖。

```python
from segment import construct, check, sweep
A, B = (0, 0), (4, 0)
for u in (3, 0, -3):
    r = construct(A, B, u, 4)
    print(r['branch'], 'O=', r['center'], 'P=', r['alternate'])
    if not all(check(A, B, u, 4, r).values()):
        raise RuntimeError('certificate failed')
```

## 为什么看角值还不够？

把正确弧与同侧弧对换。一般得到补角；直角时角值仍然对，但侧别不对。

```python
for u in (3, 0, -3):
    r = construct(A, B, u, 4)
    r['alternate'], r['control'] = r['control'], r['alternate']
    c = check(A, B, u, 4, r)
    print(u, 'alternate_side=', c['alternate_side'], 'alternate_angle=', c['alternate_angle'])
```
