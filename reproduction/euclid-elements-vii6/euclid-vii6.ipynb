# 《几何原本》VII.6：同样的若干份合并

```python
from same_parts import construct,paired_totals,ratios
w=construct(8,12,10,15)
print("份数与整体份数",w.m,w.n)
print("逐份配对",w.partial_pairs,w.whole_pairs)
print("部分与整体",paired_totals(w.partial_pairs),paired_totals(w.whole_pairs))
```

分块时用到单位，不等于把单位列为题设中的数。

```python
w=construct(2,3,10,15)
print("两组组件大小",w.first_unit,w.second_unit)
print("相同比",tuple(map(str,ratios(2,3,10,15))))
```

最简表示的分子和分母皆须匹配，而且不是单独一份。

```python
for case in ((6,15,9,15),(8,12,6,15),(3,12,5,20)):
    try:
        construct(*case)
    except ValueError as exc:
        print(case,tuple(map(str,ratios(*case))),type(exc).__name__)
```
