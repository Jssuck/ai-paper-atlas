# X.31：改变两条边而保留差平方类别

```python
from ratio_witness import construct
for a,b in [(25,21),(2,1)]:
    w=construct(a,b)
    print(w)
    print('output product fourth equals b^4:',w.c_fourth*w.d_fourth==b**4)
```

输入不满足仅平方可公度的前提。

```python
try:
    construct(4,1)
except ValueError as exc:
    print(exc)
else:
    raise RuntimeError('square control not rejected')
```
