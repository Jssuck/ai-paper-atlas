# 欧几里得 VII.12：同率对应项分别求和

检查单位份和若干份如何逐对汇总。

```python
from paired_sum import demo, pool
from fractions import Fraction
demo()
```

## 由输入独立计算合计与分数

```python
pairs = ((2,3),(6,9),(8,12))
p = pool(pairs)
baseline = tuple(map(sum, zip(*pairs)))
if p.total != baseline or Fraction(*baseline) != Fraction(*pairs[0]):
    raise RuntimeError('aggregate mismatch')
print('aggregate:', baseline, 'ratio:', Fraction(*baseline))
```

## 错配时，总和不变也不能补回前提

```python
pairs = ((4,12),(6,8))
print('crossed, total:', tuple(map(sum, zip(*pairs))), 'local ratios:', [str(Fraction(*x)) for x in pairs])
try:
    pool(pairs)
except ValueError as exc:
    print('rejected:', exc)
else:
    raise RuntimeError('unexpected acceptance')
```
