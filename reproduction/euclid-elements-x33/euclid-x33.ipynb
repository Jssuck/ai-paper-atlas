# X.33：直角边平方坐标的共轭表示

```python
from fractions import Fraction as F
from ratio_witness import construct
for h,l in [(9,2),(F(3,2),F(1,3)),(2,1)]:
    w=construct(h,l)
    print(w)
    print('product:',w.center**2-w.coefficient**2*w.radicand)
```

分别触发两条非平方约束。

```python
for args in [(4,1),(4,3),(25,9)]:
    try:
        construct(*args)
    except ValueError as exc:
        print(args,exc)
    else:
        raise RuntimeError('square control not rejected')
```
