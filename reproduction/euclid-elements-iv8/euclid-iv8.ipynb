## 一、转过且平移过的正方形

以对角点中点为圆心；半径平方等于边长平方的四分之一。再用点到直线的距离公式核对，并验证触点在线段内部。

```python
from inscribed_circle import EXAMPLE, RECTANGLE, inscribe, formatted
from inscribed_circle import line_distance_squares, run_grid, verify

center, r2, contacts = inscribe(EXAMPLE)
print("vertices:", formatted(EXAMPLE))
print("center, radius^2:", formatted((center,))[0], r2)
print("contacts:", formatted(contacts))
print("side-line distance^2:", tuple(map(str, line_distance_squares(EXAMPLE, center))))
print("verified:", verify(EXAMPLE, (center, r2, contacts)))
```
