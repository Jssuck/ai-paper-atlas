# 欧几里得 VII.19：比例与交叉积

```python
from cross_products import check
from fractions import Fraction
t = check(4,6,10,15)
print(t)
```

## 独立核对两个桥接比及结论

```python
if not Fraction(4,6) == Fraction(t.g,t.f) or not Fraction(10,15) == Fraction(t.g,t.e) or t.e != t.f:
    raise RuntimeError('bridge failure')
print('bridge:', Fraction(t.g,t.f), Fraction(t.g,t.e))
```

## 非比例数据与零分母陷阱

```python
print(check(4,6,10,16))
print('cross products at zero denominator:', 4*0, 0*6)
try:
    Fraction(4,0)
except ZeroDivisionError:
    print('ratio 4:0 is undefined')
else:
    raise RuntimeError('unexpected acceptance')
```
