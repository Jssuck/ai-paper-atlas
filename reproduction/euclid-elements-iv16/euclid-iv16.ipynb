# IV.16：从弧差到正十五边形
目标：用精确分数重建弧差，再用多项式余数核对弦和角。请从本目录依次运行三个单元；没有第三方依赖。

## 1. 同向、同圆、公共起点
五边形一边走五分之一圈，三角形一边走三分之一圈。弧差减去公共弧以后再平分。

```python
from fractions import Fraction
from polygon15 import certify, chord_sq, grid, half_arc, step_exponent, walk
turn = half_arc()
path = walk(3, step_exponent(turn))
print("三角形弧 − 五边形弧:", Fraction(1,3) - Fraction(1,5))
print("半弧:", turn, "| 从 B 出发:", path)
print("边长平方（按 1,z,…,z^7 的系数）:", chord_sq(3, 4))
print("正十五边形:", certify(path).regular)
```

## 2. 反例：等弦、闭合也可能走错
把五边形的边放到另一侧，沿原分支方向求弧差，得到不同步幅。检查它是否沿圆的边界前进。

```python
wrong = half_arc(pentagon_direction=-1)
c = certify(walk(3, step_exponent(wrong)))
print("错侧半弧:", wrong, "| 证书:", c)
print("同侧大弧的一半:", half_arc(major=True), "| 不是原命题步幅")
```
