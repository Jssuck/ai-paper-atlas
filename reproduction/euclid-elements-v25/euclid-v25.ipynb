# 欧几里得 V.25：比例与极端项之和

先追踪原证，再用乘积模型比较两个和，最后观察移除可比较性之后的失败。请从本目录运行。

## 1. 原证轨迹
先减去成比例的两个部分，再将相等的量加回。

```python
from ratio_chains import joyce_signs, parameterized, v25

t = v25(12, 6, 8, 4)
for name in ("copied_parts", "remainders", "equal_addends", "sums"):
    print(name, tuple(map(str, getattr(t, name))))
print("common ratio", t.common_ratio, "gap", t.gap)
```

## 2. 数值核算与等号边界
两个缩放因子都必须大于一；中间项相等不妨碍严格结论。

```python
q, g = parameterized(4, 2, 3)
print("parameters", tuple(map(str, q)), "factored gap", g, "trace gap", v25(*q).gap)
print("AM–GM sample", v25(9, 6, 6, 4).gap / 2)
for degenerate in ((4, 4, 2, 2), (4, 2, 4, 2)):
    a, b, c, d = degenerate
    print("degenerate", degenerate, "sum gap", a + d - b - c)
```

## 3. 字典序边界反例
前两个输出是对应倍数的两项前提比较，第三个为极端项之和与中间项之和的比较；结论看倍数 (1,1)。

```python
for m, n in ((1, 2), (1, 1), (3, 2)):
    print(m, n, joyce_signs(m, n))
```
