## 3 相对于有限词表的叶节点
{0,1}在旧词表中不可再分；加入{0}立即可分。空集合不算原子。

```python
print('old vocabulary:', relative_atom(F({0,1}), [F({0,1,2,3}),F({0,1})]))
print('new vocabulary:', relative_atom(F({0,1}), [F({0,1,2,3}),F({0,1}),F({0})]))
print('empty cell:', relative_atom(F(), []))
```
