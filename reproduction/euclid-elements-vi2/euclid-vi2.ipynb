## 1. 平行截线

检验线段比与面积比，核对 DE 和 BC 的方向行列式。

```python
from fractions import Fraction
from ratio_chains import CASES, examine

good = examine(*CASES["parallel"])
print("线段比 / 面积比:", tuple(map(str, good.segment_ratios)),
      tuple(map(str, good.area_ratios)))
print("两倍面积 ADE、BDE、CDE:", tuple(map(str, good.double_areas)))
print("方向行列式:", good.parallel_det)
if good.segment_ratios != (2, 2) or good.area_ratios != (2, 2) or good.parallel_det:
    raise AssertionError("平行正例核对失败")
```

## 2. 非平行截线

保持分子、分母的顺序。问：如果只倒置第二个比，读数还可靠么？

```python
bad = examine(*CASES["nonparallel"])
left, right = bad.segment_ratios
print("原比值:", left, right, "；倒置第二比值:", 1 / right)
print("方向行列式:", bad.parallel_det)
if (left, right, bad.parallel_det) != (2, Fraction(1, 2), 3):
    raise AssertionError("非平行例核对失败")
```

## 3. 越界反例

关闭内点限制只用于负对照；比较 B 和 C 相对 DE 的有向侧值。

```python
outside = examine(*CASES["outside"], interior=False)
print("比值:", tuple(map(str, outside.segment_ratios)))
print("两倍面积 ADE、BDE、CDE:", tuple(map(str, outside.double_areas)))
print("侧值 B、C:", tuple(map(str, outside.side_values)),
      "；方向行列式:", outside.parallel_det)
if outside.same_side or outside.converse or outside.parallel_det != 12:
    raise AssertionError("越界反例核对失败")
```

## 练习与答案

把第一例的 E 换成 (0,2)，预测哪一条面积比变化。

**答：** BDE 和 ADE 的两倍面积同乘 2，比仍为 2；CDE 的两倍面积从 2 变成 1，同时 ADE 从 1 变成 2，所以 CDE:ADE 变成 1:2。另可尝试交换基底两个坐标轴：方向行列式符号会反转，无向比不会变。
