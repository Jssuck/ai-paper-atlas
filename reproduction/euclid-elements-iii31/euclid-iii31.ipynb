# 《几何原本》III.31：弓形内接角与端点割线极限

将本文件与 `segment_angles.py` 放在同一目录。先按弦的哪一侧确定弓形，再从独立的点积判断内接角。接着换一个顶点与一种定义，探究端点的一侧极限；它不是欧几里得曲直角的重新命名。

```python
from segment_angles import examples, classify, dot, sub
from math import degrees

cases = tuple(examples())
for s in cases:
    value = s.inscribed_sign_value()
    print(s.kind(), "顶点 P", s.p, "点积", value, classify(value),
          f"内接角 {degrees(s.inscribed_angle()):.6f}°")
```

端点 A 的现代极限用沿指定弧出发的切向射线与弦射线，不用顶点 P。交换切向射线，非直角分类就会翻转。

```python
for s in cases:
    u = sub(s.b, s.a)
    v = s.toward_arc_tangent()
    print(s.kind(), "切向射线向量", v, "点积", dot(u, v),
          f"现代角 {degrees(s.modern_limit_angle()):.6f}°")
```

沿指定大弓形的边界，从 A 出发取逐渐靠近 A 的有理圆周点。表中误差用弧度计。

```python
s = cases[0]
for n in (8, 64, 512, 4096):
    q = s.sample_toward_endpoint(n)
    print(n, f"误差 {s.modern_limit_angle()-s.secant_angle_at_a(q):.12f} rad")
```
