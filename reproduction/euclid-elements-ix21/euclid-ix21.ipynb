# IX.21：偶数相加
构造两个相等的半部，检查反向命题和现代边界。

```python
from even_sum import even_sum
print('Two equal halves:',even_sum((2,4,6)))
print('Single even term:',even_sum((2,)))
```

```python
for terms in ((2,3),(3,5),(2,3,5)):
    try:
        even_sum(terms)
    except ValueError as exc:
        print('Mixed/non-even:',terms,'total',sum(terms),'rejected:',exc)
    else:
        raise RuntimeError('Expected rejection')
```

```python
for terms in ((),(0,2),(-2,6)):
    try:
        even_sum(terms)
    except ValueError as exc:
        print('Modern boundary:',terms,'rejected:',exc)
    else:
        raise RuntimeError('Expected rejection')
```
