# 《几何原本》IV.13：内切圆核验

给定正五边形，从 C、D 的内部角平分射线求交，再检查五条边。

```python
from incircle import fixtures, center_from_bisectors, inspect_center, check_grid
expected, p, irregular, shifted = fixtures(1,1)
center = center_from_bisectors(p)
print("两条角平分线找回中心:", center == expected)
```

## 切线条件

不仅检验垂直，也检验垂足落在边段、五个距离相同。

```python
ok = inspect_center(p, center)
print({k:v for k,v in ok.items() if k != "distance_squares"})
r2 = ok["distance_squares"][0]
print("内切半径平方的两个系数:", r2.a, r2.b)
```

## 两类错误证书

扰动 A（交点不变）与平移中心（正五边形不变）。

```python
for poly, f in ((irregular, center_from_bisectors(irregular)), (p, shifted)):
    facts = inspect_center(poly, f)
    print("垂足在边段上、五边等距:", facts["feet_strictly_inside_sides"], facts["equal_positive_distance_squares"])
print("正例、扰动、偏心的精确例数:", check_grid())
```
