## 错误迁移
把外部投影的校正改成减号，或漏掉二倍，均不能得到 BC²。直角和锐角输入不满足本命题假设，程序拒绝而不悄悄更改定理。

```python
print("true =",r["BC2"],"; wrong minus =",r["AB2"]+r["AC2"]-r["correction"])
for B in [(0,3),(2,3)]:
    try: construct((0,0),B,(5,0))
    except ValueError as e: print(str(e))
```

## 有限回归
5 × 5 × 5 × 2 = 250 个有理参数组合，包含关于基线的镜像；这只是实现检验，不是全称证明。

```python
values=[F(1,7),F(1,2),F(1),F(3),F(11,2)]
n=0
for c in values:
    for d in values:
        for h in values:
            for sign in [-1,1]:
                q=normalized(c,d,sign*h)
                if not all(q["checks"].values()): raise RuntimeError("check failed")
                n+=1
print("exact cases checked:",n)
```
