## 1. 中项的外延候选

在声明的有限域U中，C⊆M⊆A足以推出C⊆A。搜索全部子集不是搜索全部自然原因。先预测：C={0}、A={0,1}会有多少个中项外延？

```python
from model import candidates, candidate_count, completions, ratio_in_palette, exhaustive_summary
U = {0, 1, 2}
C, A = {0}, {0, 1}
ms = candidates(U, C, A)
print("中项候选外延:", [sorted(m) for m in ms])
print("存在:", bool(ms), "唯一外延:", len(ms) == 1)
print("计数公式:", candidate_count(U, C, A))
```

答案：两种。固定C中的点必须进M，A外点不能进M，A\C中每个点可独立选择，所以在C⊆A时为2的|A\C|次方，否则为0。即便只有一种外延，也不能区分不同描述的因果角色。

## 2. 一个观察点没有排除未见反例

假设只完整观察0，且0具有M与A；其余两点未知。先预测是否所有兼容世界都满足M⊆A。

```python
worlds = completions(U, {0}, {0}, {0})
good = sum(m <= a for m, a in worlds)
print("兼容扩展:", len(worlds), "普遍式真:", good, "普遍式假:", len(worlds)-good)
m, a = next((m, a) for m, a in worlds if not m <= a)
print("一个反例扩展:", "M=", sorted(m), "A=", sorted(a))
```

答案：16种中9种真、7种假。每个未见点有(M,A)四种标签，其中三种满足蕴涵；因而有4²种扩展和3²种真扩展。这不是说原文否认感知作用，而是指出从单次标签到普遍解释需要额外桥梁。
