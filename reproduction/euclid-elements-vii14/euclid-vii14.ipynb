```python
from endpoint_ratio import ex_aequali, demo
from fractions import Fraction
demo()
```

## 四项例子：独立从原数计算首末比

```python
x,y = (4,6,10,14), (10,15,25,35)
r = Fraction(*ex_aequali(x,y).endpoints)
if not r == Fraction(x[0],x[-1]) == Fraction(y[0],y[-1]):
    raise RuntimeError('wrong endpoint ratio')
print('endpoint:', r)
```

## 反例：仅首末相同，不保证相邻前提

```python
x,y = (4,6,10), (10,16,25)
print('endpoint ratios:', Fraction(x[0],x[-1]), Fraction(y[0],y[-1]))
try:
    ex_aequali(x,y)
except ValueError as exc:
    print('rejected:', exc)
else:
    raise RuntimeError('unexpected acceptance')
```
