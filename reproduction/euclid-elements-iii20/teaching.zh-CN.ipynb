## 3. 反角的现代扩展与错误 minor 规则

```python
r=check(*EXAMPLES["reflex_extension"])
wrong=(r["central"][0], abs(r["central"][1]))
print("selected", r["central"], "wrong minor", wrong)
if wrong == r["doubled"]: raise RuntimeError("counterexample lost")
```
