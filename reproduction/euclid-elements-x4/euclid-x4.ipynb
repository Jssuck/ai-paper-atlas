# X.4：三个量的最大公度量
先处理前两量，再核验第三量。

```python
from fractions import Fraction as F
from greatest_common_measure import greatest_common_measure
print('Direct:',greatest_common_measure(F(21,10),F(9,10),F(6,10)))
print('Delayed:',greatest_common_measure(F(21,10),F(9,10),F(4,10)))
```

```python
from itertools import permutations
for terms in permutations((6,10,15)):
    print(terms,greatest_common_measure(*terms))
```

```python
from math import lcm
a,b,c=F(21,10),F(9,10),F(4,10)
scale=lcm(a.denominator,b.denominator,c.denominator)
sizes=[int(s*scale) for s in (a,b,c)]
print('Independent enumeration:',[F(d,scale) for d in range(1,min(sizes)+1) if all(s%d==0 for s in sizes)])
```
