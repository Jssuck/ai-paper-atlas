重合圆不是相切；公共点也不能单独证明相切。

```python
print(relation((0,0),2,(0,0),2))
print(relation((0,0),2,(0,0),3))
print(relation((0,0),5,(6,0),5))
try:
    certificate((0,0),5,(6,0),5)
except ValueError as e:
    print(str(e))
```
