# III.16：切线与不可插入直射线
现代精确有理模型。有限实跑不是一般证明，不给角隙赋正实数角度。标准库compile/exec，非Jupyter kernel。

```python
from geometry import *
c=Circle((0,5),25)
p=(0,0)
a=analyze(c,p,(1,0))
print(a['kind'], a['k'], a['distance2'])
print('tangent noncontact power:',ray_power(c,p,(1,0),2))
```

# 向内射线
从接点出发的方向与指向圆心向量点积为正。计算第二交点和严格内部见证，不预设接点唯一。

```python
a=analyze(c,p,(3,4))
print('second:',a['second_parameter'],a['second_point'])
print('interior:',a['interior_parameter'],a['interior_point'],a['interior_power'])
```

# 射线与整条直线不同
向外射线没有新的交点，但反向射线有，不能将它叫切线。

```python
a=analyze(c,p,(3,-4))
print(a['kind'], 'other line parameter:',a['second_parameter'])
print('positive parameter power:',ray_power(c,p,(3,-4),1))
```
