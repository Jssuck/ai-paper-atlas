# 欧几里得 VII.27：平方与立方保持互素

```python
from coprime_powers import binary_gcd, prove
t = prove(8,15)
print('ordered square witnesses:', t.square_steps)
print('cross pairs:', t.cross_pairs)
```

## 交叉条件与两个立方

```python
if t.cubes != (8**3,15**3) or binary_gcd(*t.cubes) != 1:
    raise RuntimeError('cube verification failed')
print('cubes:',t.cubes)
```

## 缺失互素前提

```python
try:
    prove(6,15)
except ValueError as exc:
    print('rejected:',exc,'; cube gcd:',binary_gcd(6**3,15**3))
else:
    raise RuntimeError('unexpected acceptance')
```
