# Euclid I.7：同侧与对应距离
原创现代有理坐标教学。有限检查不是一般证明；标准库顺序执行，不是Jupyter内核。

```python
from geometry import certificate,reflect,verify_claim
A=(0,0); B=(4,0); C=(1,2)
r=certificate(A,B,C,C)
print({k:r[k] for k in ('premises','conclusion','distinct')})
```

同一个顶点满足距离和同侧条件；原命题排除的是另一个不同顶点。

```python
D=reflect(A,B,C)
r=certificate(A,B,C,D)
print(D, {k:r[k] for k in ('equal_a','equal_b','same_side','premises','distinct')})
```

异侧镜像保留两距离，但不满足同侧条件。

```python
r=certificate(A,B,C,(-1,2))
print({k:r[k] for k in ('equal_a','equal_b','same_side','premises')})
```

只固定一条对应距离还不够。

```python
r=certificate((1,2),(4,6),(-2,3),(7,-1))
print({k:r[k] for k in ('distance_difference_identity','norm_cross_identity_c','reconstruction_c')})
```
