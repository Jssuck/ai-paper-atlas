## 独立检查两个端点为精确幂

```python
from test_unit_endpoint_chains import root_search
roots = tuple(root_search(k,len(x)-1) for k in (x[0],x[-1]))
if roots != (2,3):
    raise RuntimeError('wrong roots')
print('integer roots:',roots)
```
