输入核验先检查圆盘包含及给定公共点，再检查共线；候选构造与核验分离。

1. 精确模型与输入

```python
from fractions import Fraction as Q
from geometry import Circle, relation, verify_contact, contact_candidate, grid
outer=Circle((0,0),5)
inner=Circle((3,0),2)
print(relation(outer,inner))
```

2. 独立提供切点，再核对结论

```python
certificate=verify_contact(outer,inner,(5,0))
print(certificate)
```

3. 平移旋转，不只检验水平图

```python
a=Circle((2,-3),5)
b=Circle((Q(19,5),Q(-3,5)),2)
print(verify_contact(a,b,(5,1)))
print("modern candidate:",contact_candidate(a,b))
```

4. 两交点反例不满足内含

```python
other=Circle((6,0),5)
print(relation(outer,other))
try:
    verify_contact(outer,other,(3,4))
except ValueError as exc:
    print("expected rejection:",str(exc))
else:
    raise RuntimeError("invalid certificate accepted")
```
