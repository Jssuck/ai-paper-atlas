# VIII.19：相似立体数的两个中项
教学重构。用两组指定的有序三因子计算辅助链与主链。

```python
from similar_solid import solid_chain
r = solid_chain((2,4,6),(3,6,9))
print('K,M,L:', r.kml, 'A,N,O,B:', r.anob, 'step:', r.step)
```

```python
from fractions import Fraction
from math import prod
print('Endpoint ratio:', Fraction(prod((3,6,9)),prod((2,4,6))))
print('Scale cubed:', r.step**3)
print('Reversed:', solid_chain((3,6,9),(2,4,6)).anob)
```

```python
try:
    solid_chain((2,4,6),(2,9,9))
except ValueError as exc:
    print('Same endpoints, mismatched factors:', exc)
else:
    raise RuntimeError('Expected mismatched factors to be rejected')
print('Modern unit boundary:', solid_chain((1,2,3),(2,4,6)).anob)
```
