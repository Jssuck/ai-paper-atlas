# 《几何原本》VII.5：逐份配对后再相加

```python
from same_part import witness,ratios
w=witness(3,12,5,20)
print("两列与配对",w.left,w.right,w.pooled)
print("还原两整体与合并整体",w.totals())
```

分母必须相同，且要求每一对是一份，不能把更宽的现代结论当前提。

```python
for case in ((3,12,5,20),(3,12,5,15),(4,6,6,9)):
    print(case,tuple(map(str,ratios(*case))))
```

以 1 为输入是可选的现代扩展。

```python
try:
    witness(1,4,5,20)
except ValueError as exc:
    print("古域拒绝",exc)
print("现代扩展",witness(1,4,5,20,True).totals())
```
