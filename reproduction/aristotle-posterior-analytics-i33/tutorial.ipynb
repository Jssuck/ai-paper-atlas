第 0 个情形的真与跨所有已选情形的真分开。有限情形并不穷尽本质上可能的情形；把有限模型的必要性直接叫作本质知识是额外错误。

## 2. 相同中项链不自动给出相同把握

每个情形用三个集合表示 A、B、C。2 个对象给每个集合 4 种取法，每个情形 64 种，两个有序情形 4096 种。一般集合传递由正文证明，这里仅穷举检实现。

意见在原文也可以有所以然；`reason_trace` 只是论证结构记录，不会凭相同字符串证明因果、本质或知识。

```python
trace = ('C subset B', 'B subset A', 'therefore C subset A')
j = Judgment('reader-1', 't0', 'essential', 'C', 'C subset A', 'necessary', trace)
k = replace(j, agent='reader-2', mode='can_be_otherwise')
print('Same reason trace:', j.reason_trace == k.reason_trace)
print('Different agents:', conflicts([j, k]))
print('Same agent:', conflicts([j, replace(k, agent='reader-1')]))
print('Uncommitted:', conflicts([j, replace(j, mode='uncommitted')]))
```

## 3. 把握标签不是事实生成器

`can_be_otherwise` 是主体断言该谓述可不成立，不是单纯不知道。`uncommitted` 不作这两种模态断言。只有同人、同时、同方面、同对象、同命题的两种相反断言才触发此一致性约定。没有建模现代广义 belief，也没有心理实验或心理不可能性证明。

```python
d = demonstration()['necessary_truth_can_be_misapprehended']
print(json.dumps(d, indent=2, sort_keys=True))
print('False necessity claim:', Judgment('reader', 't0', 'r', 'o', 'p', 'necessary').mode, modal_profile([True, False]))
```
