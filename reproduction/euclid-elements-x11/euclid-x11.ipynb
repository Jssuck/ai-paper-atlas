# X.11：等比与各自的公度见证

```python
from fractions import Fraction as F
from ratio_witness import transfer
print(transfer(F(9,4),F(18,8),4,10))
print(transfer(2,2,4,10))
```

各自可公度不推出等比。

```python
try:
    transfer(F(9,4),F(25,4))
except ValueError as err:
    print(err)
else:
    raise RuntimeError('unequal ratios accepted')
```
