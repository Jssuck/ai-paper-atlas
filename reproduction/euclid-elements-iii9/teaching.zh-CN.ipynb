## 4. 两点等距不足

```python
p=[(3,4),(3,-4),(-5,0)]
print('two distances squared =',distance2(p[0],(3,0)),distance2(p[1],(3,0)))
try:
    make_certificate(p,(3,0))
except ValueError as e:
    print('three-point check rejects:',str(e))
```
