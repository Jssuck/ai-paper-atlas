面向会读比号、了解 Python 基础的读者。用精确有理数查看四步推导，识别共用后项和可比较条件分别起什么作用。请从本目录运行全部单元；数学论证见 [中文精读](../../papers/ancient/euclid-elements-v24.zh-CN.md)。

1. 看原证中的四步数值轨迹。
2. 改掉一条前提，再用独立整数参照批量检查。
3. 用字典序量演示为什么有理数上的结果不能无条件外推。

```python
from ratio_chains import addition_trace, grid_evidence, lex_probe

case = (6, 4, 15, 10, 2, 5)
for name, pairs in addition_trace(*case).items():
    print(name, tuple(map(str, pairs)))
```

第一行是翻转第二前提，第二行暂时消去原来的两个后项，第三行作合比，最后一行恢复后项。预期两边的值先后都是 2、3、4、2。

**练习**：如果只把最后一项 5 改成 6，第一项前提还在吗？哪一项先失效？运行下段核对，并查看独立整数参照覆盖了多少组。

```python
try:
    addition_trace(*case[:-1], 6)
except ValueError as exc:
    print(exc)
else:
    raise AssertionError("the second premise should fail")
print("扫描组数、满足双前提组数：", grid_evidence(5))
```

第一对比例未变，但第二对已不同，不能继续推理。枚举也不能证明一般情形。再看非阿基米德模型：若把两个整数坐标按字典序比较，当倍数比为 1:2 时，两个前提的各自比较符号相同，但结论的对应符号不同。此模型允许某些正量的所有整数倍仍小于较大量，不能不加限制地套用卷 V 的规则。

```python
for m, n in [(1, 1), (1, 2), (1, 3)]:
    print((m, n), lex_probe(m, n))
```

输出三对符号依次对应第一前提、第二前提和结论。注意：输出无误并不意味着有限倍数已经穷尽定义。可继续练习用首坐标分别压住非临界情形，再处理临界倍数；这是与 Python 测试独立的论证。
