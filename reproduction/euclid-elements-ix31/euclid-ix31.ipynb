检验共同因子从倍数回到原数的过程。

```python
from doubling_coprime import common_divisors, doubled_coprime, transfer_double_measure
print('Doubling 8:',doubled_coprime(15,8))
print('Common divisors before and after:',common_divisors(15,8),common_divisors(15,16))
```

```python
print('Modern extension:',common_divisors(45,30),common_divisors(45,60))
print('15 measures 30 this many times:',transfer_double_measure(45,30,15))
```
