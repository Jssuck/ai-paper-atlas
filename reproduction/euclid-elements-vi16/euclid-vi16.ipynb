# VI.16：把比例改成等积问题

输入是四条正长度 a, b, c, d。比较 a:b 与 c:d 时，外项矩形是 a × d，内项矩形是 b × c。

## 1. 读一组等比例输入

两个矩形在不同坐标位置；用实际顶点计算面积。

```python
from fractions import Fraction as Q
from ratio_chains import compare

r = compare(6, 4, 9, 6)
print("ratios:", *r.ratios)
print("rectangle areas:", *r.areas)
print("proportional and equal area:", r.proportional, r.equal_area)
```

## 2. 改变一个长度，再试错误配对

错误配对 a × c 与 b × d 不是原命题中的两个矩形。

```python
r = compare(6, 4, 8, 6)
print("ratios:", *r.ratios)
print("rectangle areas:", *r.areas)
print("wrong pair products for proportional example:", 6 * 9, 4 * 6)
```

## 3. 用正分数并检查零长度

有限的有理数运算不替代欧几里得对任意正几何量的论证。

```python
r = compare(Q(1, 2), Q(3, 4), 2, 3)
print("rational areas:", *r.areas)
try:
    compare(1, 2, 3, 0)
except ValueError as exc:
    print("zero rejected:", exc)
```
