## 4 类型攻击
Python 中 True==1，证书必须拒绝这种类型替换。

```python
bad=certify(case());bad['schema_version']=True
print('bool accepted',verify(case(),bad))
if verify(case(),bad):raise RuntimeError('type attack accepted')
```

## 5 删除一个边相等条件
第一对边都长1；第二对分别长10与1。角60°的底平方91，而角90°的底平方2。较大底不再保证较大角。

```python
bad_input=case();bad_input['ABC']={'a2':[1,1],'b2':[100,1],'dot':[5,1]};bad_input['DEF']={'a2':[1,1],'b2':[1,1],'dot':[0,1]}
try: certify(bad_input)
except ValueError as e: print(str(e))
else: raise RuntimeError('missing hypothesis accepted')
```
