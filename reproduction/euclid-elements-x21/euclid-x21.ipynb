# X.21：从两边的平方坐标得到中项

```python
from fractions import Fraction as F
from ratio_witness import classify
for a,b in [(2,3),(2,8),(F(2,3),F(8,5))]:
    w=classify(a,b)
    print(f'{a}, {b}:',w)
```

在同一参考尺度里，比较矩形面积的平方与正方形边的四次幂。

```python
for a,b in [(2,3),(F(2,3),F(8,5))]:
    w=classify(a,b)
    c,d=w.area_radical
    print(c*c*d,w.equal_square_side_fourth,c*c*d==w.equal_square_side_fourth)
```
