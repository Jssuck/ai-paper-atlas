```python
from fractions import Fraction as F
from geometry import build_certificate, inspect_configuration, verify_certificate, angle_model
from cases import run_cases
print("Exact Fraction model; no trigonometric floating point")
```

## 1. 同时满足两个前提
公共射线BA沿正横轴，BC向右上，BD向左下。证书在检查异侧和角和后生成相反射线比例见证。

```python
args=((0,0),(1,0),(2,2),(-1,-1))
cert=build_certificate(*args)
print("side crosses:",cert["side_crosses"])
print("angle product:",cert["angle_product"])
print("positive scale:",cert["positive_scale"])
print("replay:",verify_certificate(cert,*args))
```

## 2. 删除异侧前提
45°和135°的两条射线在同侧，角和仍为半周，却不共线。度数只是此特殊图形的现代说明。

```python
bad=inspect_configuration((0,0),(1,0),(1,1),(-1,1))
print({k:bad[k] for k in ("opposite_sides","supplementary","opposite_rays")})
try:
    build_certificate((0,0),(1,0),(1,1),(-1,1))
except ValueError as e:
    print(type(e).__name__+": "+str(e))
```

## 3. 两个直角也不够
同侧两条射线重合，仍各与BA成直角；结论需要相反射线而不只是共线。

```python
overlap=inspect_configuration((0,0),(1,0),(0,1),(0,2))
print({k:overlap[k] for k in ("opposite_sides","supplementary","conclusion_cross","conclusion_dot","opposite_rays")})
print("60+120 same side:",angle_model(F(1,6),F(1,3),1,1))
```

## 4. 输入绑定与篡改
证书必须针对独立给出的原输入重算；不能仅信任证书内自报的输入。

```python
altered=dict(cert); altered["positive_scale"]="2"
print("tampered:",verify_certificate(altered,*args))
print("changed common ray:",verify_certificate(cert,(0,0),(2,0),(2,2),(-1,-1)))
```

## 5. 有限回归
这些输入覆盖多个平移、旋转、反射、比例和符号，不穷尽有理输入，更不能替代一般角序/旋转论证。

```python
result=run_cases()
print({k:result[k] for k in ("positive_certificates","negative_certificates","rational_turn_inputs","positive_replay","negative_rejection")})
```
