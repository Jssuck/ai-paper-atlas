# 《几何原本》V.5：减去同比的真部分

核查两个不同的辅助构造，不把有限实跑当成一般证明。

```python
from fractions import Fraction as F
from remainder_multiples import make_case, grid, solve
result = solve(*make_case(15, 5, 6, 2))
print('共同倍数:', result.factor)
print('原证：小余量候选及重建小全量:', result.auxiliary_small_remainder, result.rebuilt_small_whole)
print('加法路线：大余量候选及重建大全量:', result.added_large_remainder, result.rebuilt_large_whole)
print('两个余量:', result.large_remainder, result.small_remainder)
```

```python
for case in [(15, 5, 4, 2), (15, 5, 15, 5), (6, 4, 3, 2)]:
    try:
        solve(*make_case(*case))
    except ValueError as e:
        print(case, '拒绝:', e)
    else:
        raise RuntimeError('expected rejection')
```
