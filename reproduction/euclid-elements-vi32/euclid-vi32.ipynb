# 《几何原本》VI.32：正确拼接后的共线与顺序

```python
from straight_on import construct,data,sub
a,b,c=(0,0),(0,2),(3,0)
d,e=construct(a,b,c,2)
print("D,E",d,e)
print("向量 BC, CE",sub(c,b),sub(e,c))
print(data(a,b,c,d,e))
```

符号翻转不改长度，却会改拼接。

```python
for signs in ((-1,1),(-1,-1)):
    d1,e1=construct(a,b,c,2,*signs)
    print("方向",signs,"D,E",d1,e1,data(a,b,c,d1,e1))
```

只保留长度比，不足以确定位置。

```python
d2,e2=(7,0),(7,6)
print("只保留比例",data(a,b,c,d2,e2))
```
