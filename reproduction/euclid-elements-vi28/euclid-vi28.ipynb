# 《几何原本》VI.28：给定面积，作同形缺块

```python
from fractions import Fraction as Q
from apply_deficit import solve,area,applied,diagonal_cells,radical_area_terms

for target in (32,50):
    sol=solve(10,2,target)
    print("目标",target,"平方间距",sol.gap2,"缺块底",sol.rational_cuts(),
          "实算面积", [area(applied(10,2,t)) for t in sol.rational_cuts()])
```

半底上的构图中扣除小相似块，剩下的三个区域构成曲尺形。

```python
parts=diagonal_cells(Q(5),Q(3),Q(2))
print("小块、两补形、另一角块的面积", list(map(str,map(area,parts))))
print("曲尺形面积",sum(map(area,parts[1:])))
```

不将无理根冒充有理坐标；在乘积中精确核对根项抵消。

```python
sol=solve(10,2,49)
print("平方间距",sol.gap2,"是否有有理解",sol.rational_cuts() is not None)
print("两个分支的常数项和根项系数",
      [radical_area_terms(sol.half,sol.gap2,2,s) for s in (-1,1)])
```
