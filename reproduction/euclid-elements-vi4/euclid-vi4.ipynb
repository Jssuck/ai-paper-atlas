# 《几何原本》VI.4：两次平行截线
用精确坐标核对原证中的构造、比例换项和连比。该有限练习不是一般证明。先在目录内运行，要求 Python 标准库。

```python
from fractions import Fraction as Q
from ratio_chains import A,B,C,D,E,audit,equal_angles,point,length,require_audit,exercise_grid
r = audit(A,B,C,D,E)
require_audit(r)
f = r["f"]
print("F =", f, "两组平行 =", r["parallel"], "两组等长 =", r["replacements"])
if f != point(0,Q(15,2)):
    raise AssertionError(f)
```

两次 VI.2 共用 BC:CE。这里显示原始长度比；在库中用正的比平方作完全精确的比较。

```python
cuts = (length(B,A)/length(A,f), length(B,C)/length(C,E), length(f,D)/length(D,E))
chains = tuple(length(a,b)/length(c,d) for a,b,c,d in ((A,B,B,C),(B,C,C,A),(A,B,A,C)))
print("BA/AF, BC/CE, FD/DE:", *cuts)
print("第一次与第二次换项及 V.22 得到的同侧比:", *chains)
if cuts != (Q(2,3),)*3 or chains != (Q(3,4),Q(4,5),Q(3,5)):
    raise AssertionError((cuts,chains))
```

等角是三角形的性质；位于同半平面是特定辅助图的要求。把第二个三角形单独对底线镜像后，原命题不失效，但要重新放置才能复用构造。

```python
dr = point(4,Q(-9,2))
bad = point(4,5)
print("单独镜像仍等角?", equal_angles(A,B,C,dr,E))
print("改变 D 后仍等角?", equal_angles(A,B,C,bad,E))
try:
    audit(A,B,C,dr,E)
except ValueError as exc:
    print("构造前提:", exc)
else:
    raise AssertionError("wrong-side configuration accepted")
print("精确网格配置数:", exercise_grid())
```
