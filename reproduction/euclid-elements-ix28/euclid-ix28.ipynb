# IX.28：奇数乘偶数
以奇数决定份数，累加偶数及其每份的半部。

```python
from even_product import odd_times_even
print('Three copies of eight:',odd_times_even(3,8))
```

```python
print('Equal halves at each step:',[(i*4,i*8) for i in range(1,4)])
```

```python
for a,b in ((4,8),(8,3),(1,8),(0,8),(3,0)):
    try:
        odd_times_even(a,b)
    except ValueError as exc:
        print((a,b),'modern product:',a*b,'excluded:',exc)
    else:
        raise RuntimeError('Expected rejection')
```
