## 1. 给定三角形
A=(0,0)、B=(3,0)、C=(0,4)，把BA越过A延长，AD=AC=4，得到D=(-4,0)。

```python
from geometry import *
from cases import positive_cases,invalid_cases
from copy import deepcopy
tri=(point(0,0),point(3,0),point(0,4))
cert=construct(*tri)
print(cert['D'])
print(cert['t'])
```

## 2. 点序不是图像猜测
CA是CB、CD的正线性组合，所以射线CA位于角BCD内部。单二次域计算保留精确性。

```python
print(cert['CA_cone_coefficients'])
print({k:v for k,v in cert['checks'].items() if 'cone' in k or 'between' in k})
```

## 3. 原证和现代检验分层
原证使用I.5、整体大于部分、I.19；这里点积比较只是独立现代核验，不能用True字段代替这些证明。

```python
print(cert['checks'])
print('replay',accepted(*tri,cert))
```

## 4. 严格JSON与输入绑定
布尔与整数不可混同；更换输入后原证书不能沿用。

```python
bad=deepcopy(cert)
bad['checks']['t_positive']=1
print('integer replacing true accepted',accepted(*tri,bad))
print('changed input accepted',accepted(point(0,0),point(3,0),point(0,5),cert))
```

## 5. 有限真实运行
遍历128个有理坐标输入。数量不是一般证明，优化模式另由命令日志记录。

```python
cases=positive_cases()
passed=sum(accepted(*c,construct(*c)) for c in cases)
print({'cases':len(cases),'passed':passed})
if passed!=len(cases):raise RuntimeError('positive check failed')
```

## 6. 删除非退化前提
共线1、2、3长度给等号而不是严格不等；这种输入被几何构造拒绝，但独立长度比较器仍能展示边界。

```python
rejected=0
for c in invalid_cases():
    try:construct(*c)
    except ValueError:rejected+=1
print({'invalid':len(invalid_cases()),'rejected':rejected})
print('1+2>3',sum_greater(1,4,9))
if rejected!=len(invalid_cases()):raise RuntimeError('negative check failed')
```
