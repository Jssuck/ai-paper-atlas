# I.38：等底而不必同底

现代有理坐标教学；I.36整体等积和I.34二分是原证明依赖，不由这些有限例证明。

## 1 六点输入与同一对平行支持线

```python
from geometry import make_certificate, verify_certificate
from cases import EXAMPLE, build, positives, negatives
c=make_certificate(EXAMPLE)
print(c['input'])
print(c['area'])
```

## 2 补成GBCA与DEFH
G=A−(C−B)，H=D+(F−E)，分别由对角线AB和DF平分。

```python
print('G/H:',c['vertices'])
print('平行叉积:',c['parallel_checks'])
print('AB/DF二分:',c['halves'])
```

## 3 改变底的位置和方向
等长不要求同端点、有向向量相同或互不相交。

```python
for s in [-2, '1/2', 0, 1, 3]:
    r=make_certificate(build(s=s))
    print('底平移',s,r['area'])
print('反向端点:',make_certificate(build(direction=-1))['signed_double_areas'])
```

## 4 标签重合不等于退化
支持线本身为几何对象；不能用相同两标签确定一条新线。异侧等高虽可等积，却不满足本命题共同上支持线前提。

```python
for label,data in [('A=D',build(a=0,d=0)),('G=H',build(a=0,d=-2)),('B=F',build(s=-1))]:
    print(label,make_certificate(data)['area'])
for data in negatives():
    try: make_certificate(data)
    except ValueError as e: print('拒绝:',str(e))
    else: raise RuntimeError('非法几何被接受')
```

## 5 有限核验和完整输入绑定

```python
count=0
for data in positives():
    if not verify_certificate(data,make_certificate(data)): raise RuntimeError('certificate failed')
    count+=1
print('有限正例:',count)
import copy
bad=copy.deepcopy(c);bad['area']['ABC']=True
print('布尔篡改通过:',verify_certificate(EXAMPLE,bad))
print('同面积移绑通过:',verify_certificate(build(a=1),c))
print('有限实跑不证明一般定理；标准库compile/exec非kernel。')
```
