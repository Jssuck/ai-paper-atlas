## 相反射线不是同侧角复制
即使某种面积公式出现偶合，非法输入仍须在构造前拒绝。

```python
n=0
for label,p in invalid_cases():
    try: make_certificate(p)
    except ValueError: n+=1
print('invalid cases directly rejected:',n)
```

## 有限回归与防止证书换绑
精确有理数只覆盖此数据域，不认证所有实数、尺规可作性或任意输入角的原始复制过程。

```python
ps=list(valid_cases())
print('valid cases:',len(ps))
print('all accepted:',all(verify_certificate(p,make_certificate(p)) for p in ps))
print('same-area rebind rejected:',not verify_certificate(example(-1),c))
```
