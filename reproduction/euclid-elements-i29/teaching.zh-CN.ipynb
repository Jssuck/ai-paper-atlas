## 3. 有限正例
旋转、平移、改变斜率和正尺度，不声称穷尽。

```python
samples = list(positives())
print("positive inputs:", len(samples))
print("all verified:", all(verify_certificate(x, make_certificate(x)) for x in samples))
```

## 4. 负输入必须直接拒绝
不是只让空证书验证失败。

```python
rejected = 0
for x in negatives():
    try:
        make_certificate(x)
    except ValueError:
        rejected += 1
    else:
        raise RuntimeError("bad input accepted")
print("negative inputs rejected:", rejected)
```
