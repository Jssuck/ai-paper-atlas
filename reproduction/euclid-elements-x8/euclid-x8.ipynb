```python
from fractions import Fraction as F
from ratio_witness import classify
for q in [F(3,2), F(25,36), F(50,72), F(4,9)]:
    c = classify(q, 6)
    print(f'{q}: floors={c.floors} ratio={c.ratio} A={c.a}')
```

```python
c = classify(F(3,2))
print([(r*r, t, (r+1)**2) for r,t in zip(c.floors,(3,2))])
```
