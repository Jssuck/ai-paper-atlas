# 欧几里得 VII.21：互素数是给定比中的最小对

```python
from coprime_least import analyze, divisor_count
print('8:15:', analyze(8,15))
print('16:30:', analyze(16,30))
```

## 改看倍数自身为公度

```python
k = divisor_count(16,30,8,15)
if k != 2 or 16%k or 30%k:
    raise RuntimeError('witness mismatch')
print('common divisor:', k)
```

## 同比且更小，不一定是最小

```python
from fractions import Fraction
k = divisor_count(12,18,6,9)
r = analyze(12,18)
if Fraction(6,9) != Fraction(*r.least) or r.least == (6,9):
    raise RuntimeError('candidate control failed')
print('candidate count:', k, '; diagnostic:', r)
```
