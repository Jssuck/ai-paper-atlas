# I.22：三边如何构造两个镜像顶点
这是原创坐标教学重构，不是古代原证的机器认证。以下六格用标准库 compile/exec 顺序执行，并非 Jupyter kernel。

```python
from geometry import construct, verify, classify, Quad
from fractions import Fraction as Q
print('exact rational inputs, one quadratic field')
```

## 1. 直角例
三边分别为 FK=3、GK=4、FG=5；投影不是任意选择。

```python
c = construct(3,4,5)
print(c['K'])
print(c['squared_distances'])
```

## 2. 非有理高度
等边三角形的高度平方为3/4，不以浮点近似替代。

```python
e = construct(1,1,1)
print(e['height_squared'])
print((Quad(0,1,Q(3,4))*Quad(0,1,Q(3,4))).rational_value())
```

## 3. 镜像而非唯一坐标
固定基边，正负高度给两个顶点；两者边长相同。

```python
print(construct(2,3,4,1)['K'])
print(construct(2,3,4,-1)['K'])
```

## 4. 严格条件不能换成非严格
相切只给退化图形；分离与包含均无交点。

```python
for lengths in [(1,1,2),(3,1,2),(1,1,3),(3,1,1),(2,3,4)]:
    print(lengths, classify(*lengths))
```

## 5. 证书不是任意JSON
输入、方向和类型必须完全绑定。有限测试不证明所有输入，也不证明古代公设已有圆交点存在性。

```python
import copy
print(verify(c,3,4,5))
t = copy.deepcopy(c); t['schema_version'] = True
print(verify(t,3,4,5))
print(verify(c,4,3,5))
```
