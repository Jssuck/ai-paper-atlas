## 接近直角不等于满足精确前提
微小扰动在入口拒绝；有限测试不替代综合几何证明。

```python
try:
    construct((0,0),(1,0),(Q(1,10**20),1))
except ValueError as e:
    print("rejected:",e)
print("Finite tests do not prove the theorem.")
```
