# 《几何原本》VI.31：直角三角形三边上的相似图形

```python
from similar_side_areas import TEMPLATE,area,area_three,place,projection,triangle
a,b,c=triangle(2,1)
print("A 与垂足",a,projection(a,b,c))
print("正方形面积",tuple(map(str,area_three(a,b,c,((0,0),(1,0),(1,1),(0,1))))))
```

保持同一模板以及对应底边，换成四边形。

```python
p,q,r=area_three(a,b,c)
print("单位模板面积",area(TEMPLATE),"边上三块",tuple(map(str,(p,q,r))))
print("两腿图形对斜边图形的面积比",p/r,q/r,"相加等于斜边图形",p+q==r)
```

分别改变一个模板，或改变直角三角形的前提。

```python
half_triangle=((0,0),(1,0),(0,1))
print("斜边改作三角形，面积",area(place(half_triangle,b,c)))
s,t,u=area_three((1,1),(0,0),(4,0))
print("非直角且同模板，腿上面积和与对边面积",s+t,u)
```
