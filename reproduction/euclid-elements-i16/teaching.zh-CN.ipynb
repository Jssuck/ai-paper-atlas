```python
from geometry import construct, verify
from cases import run_cases
p={'A':[1,2],'B':[0,0],'C':[3,0],'D':[5,0]}
c=construct(p)
print(c['points'])
```

## 首个内对角
E 是 AC 中点，F=A+C−B。CF=CA+3CD/2，因此 CF 严格在外角 ACD 内。原证 SAS 与坐标检查属于不同论证层。

```python
print(c['coefficients']['CF_in_CA_CD'])
print({k:v for k,v in c['checks'].items() if k in ('AE_equals_EC','BE_equals_EF','vertical_at_E','BAC_equals_ECF','CF_inside_ACD')})
```

## 第二个内对角
M 是 BC 中点，N=B+C−A，G=2C−A。CN=CB+CG，先证 BCG 大于 ABC，再用对顶角转回 ACD。

```python
print({k:v for k,v in c['checks'].items() if k in ('BM_equals_MC','AM_equals_MN','vertical_at_M','ABC_equals_MCN','CN_inside_BCG','BCG_equals_ACD')})
```

## 输入绑定和严格类型
Python 的 True==1 不能让布尔模式号冒充整数模式号。

```python
import copy
bad=copy.deepcopy(c);bad['schema']=True
print('original:',verify(p,c),'bool schema:',verify(p,bad),'changed input:',verify(dict(p,D=[6,0]),c))
```

## 方向不是图像印象
D 必须在 BC 越过 C 的延长线上；重合、回退、离线全部拒绝。

```python
for d in ([3,0],[2,0],[5,1]):
    try: construct(dict(p,D=d))
    except ValueError as e: print(d,str(e))
```
