# V.6｜计数相减

量类各列内部一致，跨列可不同。输入必须各自为至少两份的严格整倍数，并留有正余量。

```python
from fractions import Fraction as F
from difference_counts import reconstruct, make_inputs, finite_grid, Quantity

for m, n in ((4, 3), (7, 3)):
    c = reconstruct(*make_inputs(F(2, 3), F(5, 4), m, n))
    print("整体、被扣、余量的次数:", m, n, c.remainder_count, c.branch)
    print("两列余量:", *(q.value for q in c.remainders))
    print("构造重建的全量:", *(q.value for q in c.rebuilt_wholes))
```

```python
base = (Quantity(2, "length"), Quantity(3, "area"))
whole = (Quantity(10, "length"), Quantity(18, "area"))
cut = (Quantity(6, "length"), Quantity(12, "area"))
print("错配案例的余量:", *(w.value-c.value for w, c in zip(whole, cut)))
try:
    reconstruct(base, whole, cut)
except ValueError as exc:
    print("拒绝:", exc)
else:
    raise AssertionError("错配输入未被拒绝")
```
