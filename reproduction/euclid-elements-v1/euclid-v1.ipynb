# 《几何原本》V.1：从按量分行到按份分列

目标：验证每一对给定量具有同一个整数倍数后，按份重分组。

## 1. 同一种量：三列分别等于一套基量

用两个长度 3/2 与 5/2，各复制三次。列内可相加；每列长度 4，总长度 12。

```python
from fractions import Fraction
from equimultiples import Magnitude as M, proof_columns, totals, aggregate_certificate, regression_grid
bases = (M(Fraction(3, 2)), M(Fraction(5, 2)))
large = (M(Fraction(9, 2)), M(Fraction(15, 2)))
k, rows, columns = proof_columns(large, bases)
print("行:", [[str(m.value) for m in row] for row in rows])
print("列:", [[str(m.value) for m in col] for col in columns])
print("份数、总长度、基量总长度:", k, totals(large).value, totals(bases).value)
```

## 2. 倍数相同不代表度量单位可混加

可作成对倍数检查，但长度和面积不能在列内相加。

```python
bases = (M(1), M(2, "area"))
large = (M(3), M(6, "area"))
print("成对层面的倍数:", proof_columns(large, bases)[0])
try:
    aggregate_certificate(large, bases)
except ValueError as error:
    print("不能混加:", error)
```

## 3. 反向推断会漏掉什么？

总量之比为 4:1，但逐组倍数为 3、5。

```python
large, bases = (M(3), M(5)), (M(1), M(1))
print("只看总量之比:", totals(large).value / totals(bases).value)
try:
    proof_columns(large, bases)
except ValueError as error:
    print("倍数证书被拒绝:", error)
print("合法分组与错倍数拒绝:", regression_grid())
```
