# 欧几里得 VII.24：互素性在乘积下保持

```python
from coprime_product import certificate, force_divisor
from fractions import Fraction
print('D:',certificate(8,9,35))
```

## 放开第二个互素前提，观察反证链

```python
t = force_divisor(8,10,35,5)
if Fraction(t.e,8) != Fraction(10,t.f) or 10 % t.e:
    raise RuntimeError('certificate mismatch')
print(t)
```

## 两个缺前提的反例

```python
from math import gcd
for a,b,c in ((8,10,35),(8,9,14)):
    print(a,b,c,'product gcd:',gcd(a*b,c))
    try:
        certificate(a,b,c)
    except ValueError as exc:
        print('rejected:', exc)
    else:
        raise RuntimeError('unexpected acceptance')
```
