# 《几何原本》VI.29：给定面积，作同形超出块

```python
from fractions import Fraction as Q
from excess import solve, area, area_cells, applied
sol=solve(10,2,48)
x=sol.overshoot()
print("半底、扩大后的底、超长",sol.half,sol.half+x,x)
print("实算应用面积",area(applied(10,x,2,Q(1,3))))
```

从扩大图形中扣除原角块，再用两块等大的补形与超出角块恢复应用面积。

```python
sizes=tuple(map(area,area_cells(sol.half,x,2,Q(1,3))))
print("原角、两块补形、超出角", list(map(str,sizes)))
print("大图面积",sum(sizes),"扣除原角",sum(sizes[1:]))
```

正超长只使用正平方根分支；另一代数根落在几何定义域外。

```python
print("有理例的两代数根", x, -2*sol.half-x)
irr=solve(10,2,49)
print("另一例的平方证书",irr.diagonal_base2,"有理超长",irr.overshoot())
print("常数项与根项系数",irr.radical_area(2))
```
