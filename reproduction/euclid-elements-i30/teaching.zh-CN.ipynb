```python
from geometry import make_certificate, verify_certificate
from cases import example, positives, negatives
d = example()
c = make_certificate(d)
print(c["scales"], c["distinctness_determinants"])
```

两个比例可以异号。直线没有固定朝向，不应把负比例当作不平行。

```python
print(c["transversal"])
print(c["conclusion"])
```

检查有限族，覆盖斜率、侧向、相反方向、旋转和平移。

```python
count = 0
for item in positives():
    if not verify_certificate(item, make_certificate(item)): raise RuntimeError("failed")
    count += 1
print("positive inputs:", count)
```

严格平行不能包含重合。每个负例直接调用生成器，防止空证书掩盖错误。

```python
rejected = 0
for item in negatives():
    try: make_certificate(item)
    except ValueError: rejected += 1
    else: raise RuntimeError("invalid input accepted")
print("direct rejections:", rejected)
```

证书必须绑定调用者给出的完整输入。哈希不是身份认证。

```python
from copy import deepcopy
other = deepcopy(d)
other["B"] = ["3", "1"]
print("rebound accepted:", verify_certificate(other, c))
bad = deepcopy(c)
bad["schema"] = True
print("boolean substitution accepted:", verify_certificate(d, bad))
```
