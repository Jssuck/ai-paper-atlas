## 1. 固定一组错位等比的三元组

```python
from ratio_chains import probe, first_mismatch

left, right = (6, 4, 3), (16, 12, 8)
print(left, right)
```

## 2. 保留每个错位配对

按 G、H、K、L、M、N 的次序输出倍量。两条路线的最后一对比值都应该相同。符号 1、0、−1 分别代表大于、等于、小于。

```python
for q in (1, 2, 3):
    row = probe(left, right, 1, q)
    print(q, tuple(map(str, row["scaled"])), row["signs"],
          tuple(map(str, row["original"][-1])),
          tuple(map(str, row["alternative"][-1])))
```
