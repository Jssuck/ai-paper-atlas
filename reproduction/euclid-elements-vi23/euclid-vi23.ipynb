# 《几何原本》VI.23：面积的复合比

```python
from fractions import Fraction as Q
from compound_areas import SLOPE, area, parallelogram, ratio_chain, three_areas

p,t,q = three_areas(6,2,4,5)
print("三个面积", list(map(str,(p,t,q))))
print("面积中介比", Q(p,t), Q(t,q), "首尾", Q(p,q))
```

给任意正首项，选择后续两条长度以代表两次比值。

```python
for seed in (6,Q(2,7),3):
    k,l,m = ratio_chain(6,2,4,5,seed)
    print("K,L,M", list(map(str,(k,l,m))), "首尾", Q(k,m))
```

改动夹角：比较不等正弦的反例，以及现代意义下互补角的延伸。

```python
for direction in ((Q(5,13),Q(12,13)),(Q(-3,5),Q(4,5))):
    print("目标方向", direction, "面积比",
          Q(p,area(parallelogram(4,5,direction))))
```
