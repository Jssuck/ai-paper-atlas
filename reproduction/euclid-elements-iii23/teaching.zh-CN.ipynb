## 1. 外角不是差一个符号
A-C-D沿同一射线，ACB是三角形CBD在C的外角，ADB=CDB。cot在(0,π)严格递减，cot(C)<cot(D)意味着角C>角D。负cot表示钝角，不能取绝对值。

```python
print(ordered_witness(-1,1))
```

## 2. 去掉同侧后的非重合，不是大小不等
上侧k=1与下侧k=-1互为关于AB的反射，相似且全等，但不是同一位置的集合。它只能反驳“不限侧也位置唯一”，不是原句similar and unequal的反例。

```python
p=point(1,F(1,4)); q=point(-1,F(-1,4))
print("reflection:",p,q,"cots:",cot(p),cot(q),"congruent, noncoincident, NOT unequal")
```

## 3. 近相等与退化
练习：k差10^-30能否判作相等？答案：不能；这里使用精确Fraction。t=0使射线落在底线上，必须拒绝。

```python
z=ordered_witness(1,1+F(1,10**30))
print("exact cot difference:",z["cot_D"]-z["cot_C"])
try:
    point(1,0)
except ValueError as e:
    print("rejected:",e)
```
