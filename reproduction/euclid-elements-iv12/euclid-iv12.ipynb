# 《几何原本》IV.12：从切点到外接正五边形

先生成五个等间隔切点，再求相邻切线交点；需要分别验证切点真在线段上，且所连边界不是星形。程序不模拟尺规的底层作图。

```python
from tangent_pentagon import construct, envelope, envelope_checks, dot
points = construct()
vs = envelope(points)
print(envelope_checks(points))
print("单位圆的顶点距离平方:", dot(vs[0], vs[0]))
```

## 同点集，但配错相邻关系

等弦、等普通角、有切点并不足以建立正确的外接多边形。另对照不等距切点生成的凸外接多边形。

```python
from tangent_pentagon import wrong_order, irregular_contacts
for kind, sample in [("星形", wrong_order(points)), ("不等距切点", irregular_contacts(1, 1))]:
    out = envelope_checks(sample)
    print(kind, {k: out[k] for k in ("tangent_lines", "strict_crossings", "convex_envelope", "equal_sides", "equal_ordinary_angles")})
```
