# 欧几里得 VII.11：交叉扣除后比较余数

比较原数、扣除量和剩余量三个比。

```python
from alternate_remainders import demo, subtract
from fractions import Fraction
demo()
```

## 独立检查两个分支及单位余数

```python
for a,b,c,d in ((12,36,4,12),(18,30,6,10),(3,6,2,4)):
    p = subtract(a,b,c,d)
    if Fraction(a,b) != Fraction(c,d) or Fraction(a,b) != Fraction(*p.remainders):
        raise RuntimeError('ratio mismatch')
    print(p.subtraction, p.remainders, Fraction(*p.remainders))
```

## 破坏前提：不配对地扣除

```python
try:
    subtract(12,18,4,8)
except ValueError as exc:
    print('rejected:', exc, '; original/remainders:', Fraction(12,18), Fraction(8,10))
else:
    raise RuntimeError('unexpected acceptance')
```
