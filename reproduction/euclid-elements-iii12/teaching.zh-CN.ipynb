1. 半径和而非半径差

```python
from fractions import Fraction as Q
from geometry import Circle,relation,verify_contact,contact_candidate,grid
a=Circle((0,0),3)
b=Circle((5,0),2)
print(relation(a,b))
```

2. 独立输入公共点，检查结论

```python
print(verify_contact(a,b,(3,0)))
print("separate construction:",contact_candidate(a,b))
```

3. 非轴向平移及交换圆心

```python
a=Circle((2,-3),3)
b=Circle((5,1),2)
t=(Q(19,5),Q(-3,5))
print(verify_contact(a,b,t))
print(verify_contact(b,a,t))
```

4. 共线内切不等于外切

```python
a=Circle((0,0),3)
b=Circle((1,0),2)
print(relation(a,b))
try:
    verify_contact(a,b,(3,0))
except ValueError as exc:
    print("expected rejection:",str(exc))
else:
    raise RuntimeError("invalid input accepted")
```
