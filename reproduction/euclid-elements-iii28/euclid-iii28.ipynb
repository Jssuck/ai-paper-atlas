# 《几何原本》III.28：等圆上的等弦与两条弧

把 Notebook 与 `chords.py` 放在同一目录。先看精确弦长和近似弧，再看何时换配或删条件会错。这里使用现代坐标重构，不是欧几里得原证。

```python
from chords import pair_family, unequal_radius_pair, point, Chord, polyline_arc_estimate

a, b = next(pair_family())
for chord in (a, b):
    minor, major = chord.endpoint_arcs()
    print("r² =", chord.r2, "c² =", chord.c2, "d² =", chord.midpoint_d2())
    print(f"端点极角折算（弧度）：{minor:.9f} / {major:.9f}")
print("错配差", round(b.endpoint_arcs()[1] - a.endpoint_arcs()[0], 9))
```

按带符号的路径从第一个端点走到第二个端点；折线长是有限边数下的近似。

```python
for n in (8, 32, 128):
    print(n, [round(polyline_arc_estimate(a, sweep, n), 9)
              for sweep in a.endpoint_sweeps()])
```

接下来取消等半径。保留等弦并不足以保留两种弧。直径则把长、短分类变成并列。

```python
from math import sqrt

for chord in unequal_radius_pair():
    r, c = sqrt(chord.r2), sqrt(chord.c2)
    print("半径", r, "弦长", c, "弧长", [round(r*x, 9) for x in chord.endpoint_arcs()])
diameter = Chord(point(0, 0), point(5, 0), point(-5, 0))
print("直径两半圆扫角", diameter.endpoint_arcs())
```
