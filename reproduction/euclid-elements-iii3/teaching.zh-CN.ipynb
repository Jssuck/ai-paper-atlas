# Euclid III.3：平分与垂直，条件不能省

本目录中运行。原证正向用I.8及直角定义，逆向用I.5/I.26。本Notebook独立采用现代有理坐标，不冒充古代证明或一般尺规系统。E是圆心，A、B为不同圆周点且AB非直径；F严格在弦内。

```python
from fractions import Fraction as Q
from geometry import certificate,verify_certificate,perpendicular,grid
c=certificate((0,0),25,(-3,4),(3,4),(0,4))
print(c["branch"],"AF2",c["AF_squared"],"BF2",c["BF_squared"],"dot",c["dot_EF_AB"])
```

## 1 等价的数值证书
由等半径展开，AF²−BF²=2 EF·AB。非零方向下，点积为零认证垂直；距离非负使平方相等可推出距离相等。原文证明仍应单独阅读。

```python
c=certificate((0,0),25,(-3,4),(3,4),(1,4))
print("bisects",c["bisects"],"perpendicular",c["perpendicular"])
print("identity",c["AF_squared"]-c["BF_squared"],2*c["dot_EF_AB"],"verified",verify_certificate(c))
```

## 2 直径反例
A=(-5,0)、B=(5,0)的中点就是圆心。方向(1,1)的中心线平分这条弦，却不垂直。不能拿EF=0与任何向量点积都为零去制造直角。

```python
print("diagonal versus diameter perpendicular",perpendicular((1,1),(10,0)))
try: perpendicular((0,0),(10,0))
except ValueError as e: print(type(e).__name__,str(e))
```

## 3 不过圆心也不够
圆心(0,0)、半径5，弦从(-3,4)到(3,4)。垂线x=1在F=(1,4)相交，但AF=4、FB=2；它不经过圆心，不能套用III.3。

```python
print("off-center vertical perpendicular",perpendicular((0,1),(6,0)))
print("AF2",Q(16),"BF2",Q(4),"equal",Q(16)==Q(4))
```

## 4 有限测试与伪证书
网格含中点和非中点，不仅验证前提为真的案例。严格字段类型拒绝True与1互换。

```python
rows=list(grid())
print("grid",len(rows),"midpoints",sum(x["bisects"] for x in rows),"verified",all(verify_certificate(x) for x in rows))
bad=dict(rows[0]);bad["equivalent"]=1
print("integer-as-bool accepted",verify_certificate(bad))
```
