# 《几何原本》VI.25：既保留形状，又达到指定面积

```python
from fractions import Fraction as Q
from equal_area import TRIANGLE,area,construct,rectangle,scale

source=area(TRIANGLE)
model=construct(6,source,27)
print("原面积", source, "共高", model.height, "转移底", model.transfer_base, "输出底", model.output_base())
print("转移面积", area(rectangle(model.transfer_base,model.height)))
print("实算相似三角形面积", area(scale(TRIANGLE,Q(model.output_base(),6))))
```

若输出底不是有理数，改为保留正根的精确平方证书。

```python
for target in (6,12,27):
    c=construct(6,source,target)
    print("目标", target, "输出底平方", c.output_base2,
          "输出底（None 表示不是有理数）", c.output_base(), "平方尺度", c.scale2)
```

面积比不是长度缩放的系数。

```python
print("误用面积比 27/12 直接缩放边长，得到面积", area(scale(TRIANGLE,Q(27,12))))
```
