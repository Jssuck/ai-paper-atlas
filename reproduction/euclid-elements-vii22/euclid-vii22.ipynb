# 欧几里得 VII.22：最小数对必互素

```python
from least_coprime import shrink, diagnostic
from fractions import Fraction
for c in (2,6):
    t = shrink(18,30,c)
    if Fraction(*t.smaller) != Fraction(18,30):
        raise RuntimeError('shrinking changed the ratio')
    print(t)
```

## 单位商：另作现代数值解释

```python
print('modern 3:6:', diagnostic(3,6))
try:
    shrink(3,6,3)
except ValueError as exc:
    print('bounded model:', exc)
else:
    raise RuntimeError('unexpected acceptance')
```
