# 《几何原本》III.32：弦与切线的两个角
安排：先验证两个配对，再引入反例，最后跑有限枚举。这里只比较角，不是在重演作图或证明定理。


```python
from alternate_segments import A, B, C, D, E, F, O, audit, confirm, finite_sweep

for p, ray in [(A, F), (C, E)]:
    result = audit(O, B, D, p, ray)
    print(result)
    confirm(result["opposite"] and result["equal"], "expected alternate pair")
```

点落在弦的哪一侧，取决于叉积的符号；角的比较却用叉积的绝对值，因为此处比较的是无向角。点积的正负不能丢，否则会把锐角和钝角混为一谈。


```python
for label, p, ray in [("同侧", C, F), ("非切线", A, C)]:
    result = audit(O, B, D, p, ray)
    print(label, result)
    confirm(not result["equal"], "bad control must not compare equal")
for p in [(-5, 5), (5, 5)]:
    result = audit(O, B, A, p, F)
    print("直径", result)
    confirm(result["equal"], "both semicircle angles are right")
```

```python
n, good, same, bad, right = finite_sweep()
print(f"{n} points per radius; {good+same} triples; {good} alternate equals")
print(f"{same} same-side controls; {bad} unequal; {right} diameter equal")
```

有限样本只检验实现。直径时两个角都是直角，所以把顶点错配到同侧也可能数值相等；不能用角值相等来反推侧别。
