# 欧几里得 VII.32：质数自身或合数的质因数

```python
from prime_or_divisor import classify
print(classify(13))
print(classify(84))
```

## 合数可以有不同的有效见证

```python
for largest in (True,False):
    c = classify(84,largest=largest)
    if c.branch != 'VII.31' or 84%c.prime or any(c.prime%d == 0 for d in range(2,c.prime)):
        raise RuntimeError('bad witness')
    print(c)
```

## 单位不是这条命题的给定数

```python
try:
    classify(1)
except ValueError as exc:
    print('rejected:',exc)
else:
    raise RuntimeError('unexpected acceptance')
```
