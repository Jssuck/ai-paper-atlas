## 3. 名字是另一份输入
绑定与非空检查分离。改字典不会改变集合成员事实。

```python
forms={"equidistant":{0}, "empty":set()}
lex={"circle":"equidistant", "goat_stag":"empty"}
for claim in [NameClaim("circle","equidistant"),NameClaim("mountain_copper","equidistant"),NameClaim("goat_stag","empty")]:
    print(claim.name, "binding=",binding_valid(lex,claim),"named witness=",named_witness(u,forms,lex,claim,0))
```
