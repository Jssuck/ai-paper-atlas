适合已知弦、圆周角和 Python 基本语法的读者。依次核查原命题的五个圆周角、边界的绕行、星形反例。本现代模型用代数生成五个点，不模拟古代尺规动作。

## 1. 先检查三角形与角平分
用精确算术生成一组圆上点。验证 ACD 的 36°–72°–72° 角形以及 CE、DB 的内部平分。

```python
from fractions import Fraction
from pentagon import check_grid, construct, polygon_checks, source_checks, star

p = construct(Fraction(3, 7))
print(source_checks(p))
```

## 2. 沿圆周边界走一圈
看圆上点、弦长、普通内角与绕数，不靠坐标近似。

```python
c = polygon_checks(p)
print('边界：绕数、交叉数、108°判定', c['winding'], c['strict_crossings'], c['ordinary_angles_108'])
```

## 3. 练习：跳过一个顶点
每次跨两个圆周间隔连接，等边和普通顶角相等仍然成立吗？预测绕数再运行。

```python
s = polygon_checks(star(p))
print('星形：等弦、等普通角', s['equal_chords'], s['equal_ordinary_angles'])
print('星形：绕数、交叉数、正五边形判定', s['winding'], s['strict_crossings'], s['regular_convex_boundary'])
print('精确网格的正例数、星形对照数', check_grid())
```

答案：两项局部相等都成立，但绕数为 2、严格交叉数为 5，所以它并不是沿边界的正五边形。试反向取点，检查绕数符号变了而边界判定不变。
