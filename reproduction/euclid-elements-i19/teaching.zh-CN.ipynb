## 1. 从独立角比较进入
普通角的余切严格递减，dot/abs(cross) 可精确比较；不调用浮点反三角函数。

```python
from geometry import inspect_triangle, build_certificate, verify_certificate, infer_side
from cases import positive_cases, invalid_cases
from copy import deepcopy
r=inspect_triangle([0,0],[3,0],[0,4])
print(r)
```

## 2. 两个排除分支
证书记录 I.5 与交换标签后的 I.18；坐标计算是另一路现代一致性核验，不证明这些历史依赖。

```python
certificate=build_certificate([0,0],[3,0],[0,4])
print(certificate['excluded'])
print(certificate['remaining'])
print(verify_certificate([0,0],[3,0],[0,4],certificate))
```

## 3. 删除前提会怎样
不可比较不是线段的通常关系，而是人为反模型，展示三分性角色。

```python
print('完整假设',infer_side(1))
print('删除全序',infer_side(1,total=False))
print('删除等长保等角',infer_side(1,equal_implies_equal=False))
print('删除反序保反角',infer_side(1,less_implies_less=False))
```

## 4. 证书必须绑定输入和类型
Python 中 True == 1；本检查器不把这种相等误认为JSON证书相同。

```python
bad=deepcopy(certificate)
bad['schema']=True
print('布尔替换通过？',verify_certificate([0,0],[3,0],[0,4],bad))
print('换输入通过？',verify_certificate([0,0],[3,0],[0,5],certificate))
```

## 5. 真正跑有限样例
有限输入数量不是普遍定理证明。

```python
cases=positive_cases()
passed=sum(verify_certificate(*case,build_certificate(*case)) for case in cases)
print({'positive_cases':len(cases),'passed':passed})
if passed!=len(cases): raise RuntimeError('certificate failure')
```

## 6. 拒绝退化和不满足题设的输入
错误输入不被计为数学反例；现代抽象模型的反例另在第3步。

```python
rejected=0
for case in invalid_cases():
    try: build_certificate(*case)
    except ValueError: rejected+=1
print({'invalid_inputs':len(invalid_cases()),'rejected':rejected})
if rejected!=len(invalid_cases()): raise RuntimeError('negative failure')
```
