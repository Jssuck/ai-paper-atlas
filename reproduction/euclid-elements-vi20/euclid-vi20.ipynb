# 《几何原本》VI.20：相似多边形的面积平方比

```python
from fractions import Fraction as Q
from polygon_areas import CONCAVE, double_polygon, double_triangle, fan, intersection, worked

p,q = worked()
for polygon in (p,q):
    print("扇形各块二倍面积", [str(double_triangle(*t)) for t in fan(polygon)],
          "整体二倍面积", double_polygon(polygon))
print("整体面积比（目标/原图）", Q(double_polygon(q),double_polygon(p)))
```

原图的交点是两块面积传递的辅助点，不是扇形分割的顶点。

```python
fmt = lambda point: tuple(map(str,point))
(a,b,c,d,e),(f,g,h,k,l)=worked()
print("M,N", fmt(intersection(a,c,b,e)), fmt(intersection(f,h,g,l)))
print("第二组交点", fmt(intersection(b,d,e,c)), fmt(intersection(g,k,l,h)))
```

任取锚点的代数扇形求和不是一般的多边形实际分割。

```python
terms=[double_triangle(*t) for t in fan(CONCAVE,0)]
print("带符号二倍面积", terms)
print("符号和", sum(terms), "绝对值和", sum(map(abs,terms)))
```
