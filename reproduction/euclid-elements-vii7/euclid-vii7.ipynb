# 《几何原本》VII.7：扣除同一份后的份比

```python
from subtract_same_part import reconstruct,ratios
ab,cd,ae,cf=8,24,3,9
w=reconstruct(ab,cd,ae,cf)
print("G,C,F,D 的坐标",w.coordinates)
print("EB,CG,GF,FD",w.eb,w.cg,w.gf,w.fd)
print("GF-CD",w.gf-cd,"扣除同一个 CF 后",w.gf-cf,cd-cf)
```

比较原比、移走项的比、余项的比。

```python
for case in ((8,24,3,9),(8,24,3,12),(3,9,2,6)):
    print(case,tuple(map(str,ratios(*case))))
```

区分题设里给出的数和运算中产生的单位余项。

```python
print("EB 恰为单位",reconstruct(3,9,2,6).eb)
for case in ((8,24,0,0),(8,24,8,24),(3,9,1,3)):
    try:
        reconstruct(*case)
    except ValueError as exc:
        print(case,type(exc).__name__)
```
