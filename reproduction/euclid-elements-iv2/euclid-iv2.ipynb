# IV.2：从模板三角形作圆内同角三角形

固定切点和圆。按 D、E、F 对应 A、B、C，目标是精确核验三角均相等且输出三点都在圆周，而不是只比较角集合。

```python
from inscribed_triangle import OBTUSE, angle_keys, inscribe, sweep, verifies
abc = inscribe(OBTUSE, 5)
print("A,B,C:", abc)
print("template:", angle_keys(OBTUSE))
print("output:", angle_keys(abc))
print("verified:", verifies(OBTUSE, abc, 5))
```

## 负例：交换对应关系

两个点仍在圆上，但一个锐角与一个钝角不能配对。

```python
a,b,c = abc
print("swap B,C verified:", verifies(OBTUSE, (a,c,b), 5))
```
