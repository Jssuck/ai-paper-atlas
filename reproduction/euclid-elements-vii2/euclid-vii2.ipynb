# 《几何原本》VII.2：求最大共同量数

```python
from common_measure import trace,backward_witness,common_divisors,greatest_measure
rows=trace(84,30)
for row in rows:
    print(row.a,row.b,row.q,row.r)
print("反向倍数证书",backward_witness(rows))
```

最后非零余数能量尽原两数；任何共同量数也能量尽它。

```python
g=greatest_measure(84,30,True)
print("最大",g,"共同正因子",common_divisors(84,30))
print("每个共同正因子都整除最大项",
      all(g%d==0 for d in common_divisors(84,30)))
```

初始整除，以及现代整数定义允许的其他输入。

```python
for a,b in ((21,7),(9,9),(35,22),(9,1)):
    print((a,b),"最终一段",trace(a,b)[-1],"现代 gcd",greatest_measure(a,b))
```
