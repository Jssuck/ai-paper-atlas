```python
from geometry import make_certificate, verify_certificate
from cases import positives, negative_cases, sample
certificate=make_certificate(sample())
print(certificate["angle_keys"])
```

E=(0,0), F=(2,0), B=(1,1), D=(3,1)。G=2E−F，故GEB和EFD都是135度；cot指标均为−1。

```python
interior=make_certificate(sample("interior_sum"))
print(interior["angle_keys"])
print(interior["direction_determinant"],interior["same_scale"])
print(verify_certificate(sample("interior_sum"),interior))
```

同侧内角BEF=45度、EFD=135度。指标1与−1的和为0，只有在角都在(0,π)中时才等价于和为π；不能把角大小与cot值相加混为一谈。

```python
rejected=[]
for name,bad in negative_cases().items():
    try: make_certificate(bad)
    except ValueError: rejected.append(name)
    else: raise RuntimeError(name)
print("directly rejected:",len(rejected))
```

```python
import copy
attack=copy.deepcopy(certificate)
attack["schema"]=True
print("bool schema accepted:",verify_certificate(sample(),attack))
print("mode rebound accepted:",verify_certificate(sample("interior_sum"),certificate))
```

有限有理样例含正负斜率、反射、旋转、平移、尺度，不能代替一般证明。

```python
count=0
for data in positives():
    if not verify_certificate(data,make_certificate(data)): raise RuntimeError("failed")
    count+=1
print("positive cases:",count)
```
