# I.18：截取一个等腰三角形

原创坐标教学重构。标准库精确二次域；下列记录由 compile/exec 生成，不是 Jupyter kernel。原证依赖 I.3、I.5、I.16 和角的严格包含；点积比较只是另一路现代核查，不能冒充原典证明。

```python
from geometry import *
from cases import positives, negatives
a,b,c=point(0,0),point(1,1),point(4,0)
k=construct(a,b,c)
print("AC squared:",dist2(a,c),"AB squared:",dist2(a,b))
print("certificate accepted:",accepted(a,b,c,k))
```

## 1. 截取
令 t 为 AB²/AC² 的正平方根。0<t<1 保证 D 严格在 AC 内部，AD=AB。此例 t=√(1/8)，不作浮点近似。

```python
t,d=parameters(a,b,c)
print("t =",t)
print("D =",tuple(map(str,d)))
print("AD squared =",dist2(a,d))
```

## 2. 角的严格内部
BD=(1−t)BA+tBC，两个系数严格正且 BA、BC 非共线，因此 BD 为角 ABC 的内部射线。

```python
print("cone coefficients:",str(1-t),str(t))
print("cone identity:",sub(d,b)==add(scale(1-t,sub(a,b)),scale(t,sub(c,b))))
```

## 3. 原证路线与现代核查分开
等腰底角、外角严格大于内对角和严格角部分，拼成 ABC > ABD = ADB > BCA。下面布尔量是精确坐标核查，不是形式化公理证明。

```python
for name,value in k["checks"].items():
    print(name,value)
```

## 4. 不可信证书
输入绑定及严格 JSON 类型防止另一三角形的证书、篡改和 Python 中 True==1 的陷阱。

```python
import copy
bad=copy.deepcopy(k)
bad["checks"]["AD_equals_AB"]=1
print("Boolean replaced by integer accepted:",accepted(a,b,c,bad))
print("different input accepted:",accepted(point(1,0),b,c,k))
```

## 5. 有限覆盖
任意有理非共线输入是接口范围；下面只枚举有限样本，不能证明一般定理。等边界、反向边序和退化输入应拒绝。

```python
count=sum(accepted(*args,construct(*args)) for args in positives())
rejected=0
for args in negatives():
    try:
        construct(*args)
    except (TypeError,ValueError):
        rejected+=1
print("positive certificates:",count,"negative inputs rejected:",rejected)
```

## 练习与答案
1. t=1 为什么不能支持严格结论？D=C，使角内部分拆不严格；两边相等只给底角相等。
2. 删除 AC>AB 会怎样？t可能超过1，D不在AC内部，原证链失去前提。
3. 代码接受原始浮点坐标吗？不接受，使用整数或Fraction。
4. 代码的一次通过证明全部三角形吗？不，有限运行只检查已列样本。
