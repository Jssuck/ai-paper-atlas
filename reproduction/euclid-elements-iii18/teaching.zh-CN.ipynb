```python
from geometry import *
print("Exact rational teaching model; not an ancient proof engine")
```

切线候选：方向没有预先筛选


```python
c=certify((0,0),(3,0),3,(0,1))
print(c["kind"], verify(c))
print([line_power(c,t) for t in [-2,-1,0,1,2]])
```

斜线：垂足是严格内部见证


```python
d=certify((0,0),(3,0),3,(1,1))
print("t =",d["t"], "H =",d["h"], "power =",d["power"])
```

射线不能替代整条直线


```python
e=certify((0,0),(3,0),3,(1,2))
print("forward powers", [line_power(e,t) for t in [0,1,2]])
print("backward interior witness", e["t"], e["power"])
```
