## 2. 练习与答案
把第一例切点换成 (0, 5)，它还在同一圆上，为什么程序应拒绝？先做预测。

```python
o, r, d, c, a, _ = DIAMETER
bad_b = point(0, 5)
try:
    verify(o, r, d, c, a, bad_b)
except ValueError as e:
    print('拒绝：', e)
else:
    raise RuntimeError('应拒绝非切线')
```

答案：半径向量为 (0, 5)，从新 B 指向 D 的向量为 (13, −5)，点积为 −25。点在圆上不等于切线垂直于半径。
