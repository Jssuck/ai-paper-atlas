# 《几何原本》VI.30：中外比的精确坐标演示

以有理数对表示 a+b√5。

```python
from extreme_mean import cut,coerce,conjugate_root,rectangle,signed_double_area

total=coerce(10)
x,y=cut(10)
fmt=lambda z: (str(z.a),str(z.b))
print("长段、短段",fmt(x),fmt(y),"长段平方",fmt(x*x))
```

从两个等面积图形扣除同一个公共矩形，比较剩余部分。

```python
double=lambda a,b: signed_double_area(rectangle(a,b))
common=double(total,x)
print("总正方形和应用图形的二倍面积",fmt(double(total,total)),fmt(double(total+x,x)))
print("扣除公共块后的二倍面积",fmt(double(total,total)-common),fmt(double(total+x,x)-common))
```

几何要求长段、短段均正，且指定 x 为长段。

```python
bad=conjugate_root(10)
print("另一根的符号",bad.sign(),"能满足面积方程吗",(total+bad)*bad==total*total)
print("调换长短段后中外比仍成立吗",y*y==total*x)
```
