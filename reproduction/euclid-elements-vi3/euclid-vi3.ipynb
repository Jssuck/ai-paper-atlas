# 《几何原本》VI.3：角平分线与底边比例

适合熟悉平面坐标、内积和 Python 基础的读者。目标：区分角相等、对应有序的线段比和内点条件。

1. 用独立的内积计算检查内角平分。
2. 练习：交换两段比会发生什么？
3. 删除内点条件，观察无向长度比为什么误判。

与模块放在同一目录，从头运行。下列是现代教学探针，不是欧几里得的证明。

```python
from ratio_chains import A, B, C, D, P, Q, construct_e, probe
from fractions import Fraction as F

inside = probe(A, B, C, D)
e = construct_e(A, B, C, D)
print("D:", D, "BD/DC:", inside["segment_ratio"], "AB/AC:", inside["side_ratio"],
      "内角平分:", inside["interior_bisector"], "E:", e)
if not inside["interior_bisector"] or e != (F(-28), F(0)):
    raise AssertionError(inside)
```

## 练习：倒置比例

上一步的两个角等大，但若把从 B 量起的线段比倒成 4:3，选出来的 Q 还会是角平分线上的点吗？先预测，再用同一个内积检查。

```python
wrong = probe(A, B, C, Q)
print("Q:", Q, "BQ/QC:", wrong["segment_ratio"],
      "余弦差乘 AQ:", wrong["cosine_delta_times_ad"])
if wrong["segment_ratio"] != F(4, 3) or wrong["interior_bisector"]:
    raise AssertionError(wrong)
```

## 边界：延长线上的 P

Q 在底边内部但比例错误；P 则比例正确却在底边之外。注意检查的是内角平分射线，而非任意一条角平分线。可尝试交换 B、C，再检查两边比是否同时倒置。

```python
outside = probe(A, B, C, P)
print("P:", P, "BP/PC:", outside["segment_ratio"], "内点:", outside["interior"],
      "外角平分线所在直线:", outside["external_bisector_line"])
if not outside["equal_ratio"] or outside["interior_bisector"] or not outside["external_bisector_line"]:
    raise AssertionError(outside)
```
