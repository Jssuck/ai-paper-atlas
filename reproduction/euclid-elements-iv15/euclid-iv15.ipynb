# 《几何原本》IV.15：构作内接正六边形
目标：先从两圆方程算交点，再取对径点，最后检验证书。本教程使用有理数对表示选定根式基底中的系数。

## 1. 构造
令 u=(3,4)，则原圆半径为 5。辅助圆半径也为 5。注意打印的是基底系数而非直角坐标。

```python
from fractions import Fraction as Q
from hexagon import certificate, construct, grid
u = (3, 4)
v = construct(u)
result = certificate(v, u)
print("A,B,C,D,E,F:", [tuple(map(str,p)) for p in v])
print("all edge squares:", list(map(str,result["edges2"])))
print("all interior angle dot products:", list(map(str,result["angle_dots"])))
print("regular:", result["regular"])
```

## 2. 保持共圆但破坏正则性
修改平方半径比 q：第一个反例边长不等；第二个反例点集完全相同，却不能按原命题顺序连接。

```python
bad = certificate(construct(u,Q(12,19)),u)
print("wrong radius, on circle:", bad["on_circle"], "| regular:", bad["regular"])
print("normalized edge squares:", [str(n/25) for n in bad["edges2"]])
wrong_order = construct(u,3)
order_check = certificate(wrong_order,u)
print("same point set:", set(v) == set(wrong_order), "| ordered:", order_check["ccw_steps"])
```
