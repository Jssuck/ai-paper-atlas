## 内部垂足与三个相似关系

```python
from fractions import Fraction as Q
from ratio_chains import dist2, projection, sweep, witnesses
A, B, C = (0,0), (15,0), (0,20)
D, t, similarities, identities = witnesses(A,B,C)
print('D, t =', D, t)
print('same angles and side scale for each pair:', similarities)
print('each squared identity:', identities)
```

## 没有顶角为直角的对照

```python
A, B, C = (0,0), (3,0), (2,4)
D, t = projection(A,B,C)
print('Oblique triangle, interior foot:', D)
print('AD² and BD·DC:', dist2(A,D), t*(1-t)*dist2(B,C))
```
