## V.20：相邻比例相同，端点次序也相同

用正有理数模型读证明的三个分支。运行前把工作目录设为这个 Notebook 所在目录。

```python
from ratio_chains import chain, verify_v20

left = chain((6, 4, 3), "长度")
right = chain((12, 8, 6), "面积")
trace = verify_v20(left, right)
print(f"左组 A:B, C:B = {trace.left_ratio}, {trace.left_reversed_ratio}")
print(f"右组 D:E, F:E = {trace.right_ratio}, {trace.right_reversed_ratio}")
print(f"端点的共同次序：{trace.sign:+d}")
```

### 等号与反方向

只比较各组三量中的端点；不比较长度和面积的绝对大小。

```python
for left, right in (((6, 4, 6), (12, 8, 12)), ((3, 4, 6), (6, 8, 12))):
    trace = verify_v20(chain(left, "长度"), chain(right, "面积"))
    print(f"{left} / {right}: {trace.sign:+d}")
```

### 缺失比例前提时

跳过第二组比例检查会放过第一个反例；跳过第一组检查会放过第二个。两个反例的左组三量均有 A > C，右组三量却有 D < F。

```python
for right in ((12, 8, 15), (12, 20, 15)):
    try:
        verify_v20(chain((6, 4, 3), "长度"), chain(right, "面积"))
    except ValueError as exc:
        print(f"{right}: {exc}")
    else:
        raise RuntimeError("expected mismatched premises to be rejected")
```
