## 不可删不等于最少
以下删除顺序实验属于现代教学；原文不主张最小基数优化。

```python
fs = [u-{1,2,3}, u-{4}, u-{1,4}, u-{2}, u-{3}]
for order in [range(5), reversed(range(5))]:
    ids = greedy_reduce(u,target,fs,order)
    print(ids, describe(u,target,[fs[i] for i in ids]))
```
