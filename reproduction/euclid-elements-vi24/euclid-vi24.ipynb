# 《几何原本》VI.24：对角线上的两个子平行四边形

```python
from fractions import Fraction as Q
from diagonal_cells import U,V,partition,area,axis_square_ratios,scale_match

whole,ll,ur,lr,ul = partition(U,V,Q(1,3))
print("整图及四块面积", list(map(lambda p: str(area(p)), (whole,ll,ur,lr,ul))))
print("对角交点", ll[2])
```

先明确对应方向，再比较各方向边长的比例。

```python
for label,cell in zip(("起角格","对角格","非对角格一","非对角格二"),(ll,ur,lr,ul)):
    print(label,"对应两轴平方比", list(map(str,axis_square_ratios(cell,whole))))
```

移走交点后，两方向的切分比例不再一致。

```python
whole,ll,ur,_,_=partition(U,V,Q(1,3),Q(1,2))
print("偏离对角线的交点", ll[2])
print("原起角格和对角格仍按标记轴相似？", scale_match(ll,whole), scale_match(ur,whole))
```
