# X.2：交互相减
精确比较相同商前缀后的终止与符号续推。

```python
from fractions import Fraction
from reciprocal_subtraction import rational_trace,sqrt2_trace
r=rational_trace(Fraction(99,70),1)
print('Rational quotients:',[row[2] for row in r],'last remainder:',r[-1][-1])
```

```python
s=sqrt2_trace(7)
print('Symbolic quotients:',[row[2] for row in s],'last remainder coefficients:',s[-1][-1])
```

```python
print('Common measure (rational):',r[-1][1])
print('Two-index strict inequalities:',[2*w<x for x,y,q,w in r if w])
```
