分别检验平方支、立方支及合数次数的现代反例。

```python
from missing_powers import positional_certificate, floor_root
print('Nonsquare base 8:', [(r.exponent,r.root) for r in positional_certificate(8,6,2)])
print('Noncube base 4:', [(r.exponent,r.root) for r in positional_certificate(4,6,3)])
```

```python
for degree in (2,3):
    try:
        positional_certificate(64,6,degree)
    except ValueError as exc:
        print('64, degree',degree,':',exc)
    else:
        raise RuntimeError('Expected branch precondition rejection')
```

```python
print('4 is not a fourth power:',floor_root(4,4)**4!=4)
print('4² is a fourth power:',floor_root(4**2,4)**4==4**2)
```
