# 《几何原本》III.29：给定等弧，连接端点

把本文件和 `arcs_to_chords.py` 放在同一目录。以指定弧的半径和正向扫角为输入，依次看普通角度的数值对照、特例的精确坐标及删去条件后的反例。

```python
from fractions import Fraction as Q
from arcs_to_chords import Arc, unequal_radius_equal_length

for sweep in (Q(3, 8), Q(1, 2), Q(5, 8)):
    arcs = (Arc(Q(5, 3), sweep, Q(2, 7), (3, -2)),
            Arc(Q(5, 3), sweep, Q(-1, 13), (-4, 2)))
    values = [round(a.coordinate_chord2(), 9) for a in arcs]
    print(f"弧扫过 {sweep} 圈：坐标 c² = {values}，半角公式 c² = {arcs[0].trigonometric_chord2():.9f}")
```

较小弧与补成整周的较大弧不是同一个输入；相等弦长不会保留这个区别。

```python
for s in (Q(1, 4), Q(1, 2), Q(3, 4)):
    arc = Arc(1, s)
    print("扫角（圈）", s, "精确弦长平方", arc.exact_quarter_grid_chord2())
```

跨圆时保持弧长不变，不保持扫角不变。用一份整周弧度对应的长度单位来精确比较。

```python
for arc in unequal_radius_equal_length():
    print("半径", arc.radius, "扫角（圈）", arc.sweep,
          "弧长 / 2π", arc.radius*arc.sweep, "精确弦长平方", arc.exact_quarter_grid_chord2())
```
