# 《几何原本》VI.11：第三比例项

从 AB=4、AC=6 作出 CE，先用平行交点决定 E，再独立检验比例与长度。

```python
from ratio_chains import construct, verify, ratio_squares, sub, norm2, sweep
c = construct(4, 6)
print("D, E:", c[3:])
print("CE²:", norm2(sub(c[4], c[2])))
print("ratio squares:", ratio_squares(c))
print("passed:", verify(c, 4, 6))
```

## 错抄长度

平行线依然可以构造，但选错复制长度会得到错误的第三比例项。

```python
bad = construct(4, 6, copy_length=4)
print("D, E:", bad[3:])
print("CE²:", norm2(sub(bad[4], bad[2])))
print("ratio squares:", ratio_squares(bad))
print("passed:", verify(bad, 4, 6))
```
