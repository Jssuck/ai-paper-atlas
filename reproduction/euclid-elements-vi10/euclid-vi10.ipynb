# VI.10：用平行线把一条线段的分段比例搬到另一条

在本目录运行；只需要 Python 标准库。

```python
from ratio_chains import EXAMPLE, affine, certificate, construct, sweep
f, g, h, k = construct(*EXAMPLE)
print('F, G, H, K:', f, g, h, k)
print('ratio groups:', certificate(*EXAMPLE, f, g, h, k))
```

保留 G 在底边内部，只改其位置：点序不足以保证比例。

```python
try:
    certificate(*EXAMPLE, f, (9, 0), h, k)
except ValueError as exc:
    print('Rejected:', exc)
else:
    raise RuntimeError('perturbation unexpectedly passed')
```

比较交点法和直接插值，扫描有理构形；有限核对不是一般证明。

```python
print('affine example:', affine(*EXAMPLE[:3], EXAMPLE[3:]))
print('configurations, opposite orientation =', sweep())
```
