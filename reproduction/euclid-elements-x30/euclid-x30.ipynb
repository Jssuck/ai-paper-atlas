# X.30：两项比例均没有有理见证

```python
from fractions import Fraction as F
from ratio_witness import construct
for h,m,n in [(1,1,2),(F(3,2),1,1),(2,2,1)]:
    w=construct(h,m,n)
    print(w, 'reconstructed h:',w.l_squared+w.g_squared)
```

```python
try:
    construct(1,3,4)
except ValueError as exc:
    print(exc)
else:
    raise RuntimeError('Pythagorean control not rejected')
```
