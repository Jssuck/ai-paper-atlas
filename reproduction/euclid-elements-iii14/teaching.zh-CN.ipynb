# III.14：等弦与等距
同圆内两弦长度相等，当且仅当圆心到弦所在直线的距离相等。有限有理实验不是一般证明。本Notebook以标准库compile/exec实际运行，非Jupyter kernel。

```python
from geometry import *
c=Circle((0,0),5)
q=chord(c,(-4,3),(4,3))
print(q)
```

## 垂足为何是中点
程序先按投影计算参数，不把1/2作为输入；等半径端点使投影参数为1/2。

```python
print('projection parameter:',q['projection_parameter'])
print('half chord squared + distance squared:',q['half_length2']+q['distance2'])
```

## 双向比较
比较长度平方与距离平方，非负性使平方相等与长度相等一致。

```python
for ends in [((-4,-3),(4,-3)),((-3,4),(3,4))]:
    x=compare(c,(-4,3),(4,3),*ends)
    print(x['equal_chords'],x['equal_distances'],x['equivalence_holds'])
```

## 直径的垂距为零
圆心自身是垂足；不把零长线段当成一条有方向的垂线。

```python
diameter=chord(c,(-5,0),(5,0))
print('foot:',diameter['foot'],'distance squared:',diameter['distance2'])
print(compare(c,(-5,0),(5,0),(0,-5),(0,5))['equivalence_holds'])
```
