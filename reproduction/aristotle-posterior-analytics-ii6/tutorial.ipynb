## 1 同一外延不决定本质标签
两份记录具有相同集合关系，只有外部提供的unity声明不同；程序不认证该声明。

```python
from model import conjunction_report, minimal_supports, independent_support, opposite_transfer
for label in (False,True):
    print(conjunction_report({'a'}, {'a','b'}, {'a','c'}, label))
```

## 2 哪些事实真的进入推导？
别名必须明确列入禁用事实；规则本身的隐含循环不在检查范围。

```python
rules = [(['alias'],'target'), (['observation'],'target')]
supports = minimal_supports({'alias','observation'},rules)
print(sorted(sorted(s) for s in supports['target']))
print(independent_support(supports,'target',{'target','alias'}))
print(independent_support(minimal_supports({'alias'},rules),'target',{'target','alias'}))
```

## 3 相反关系不是无偿转移规则
source-definition、两项contrary记录、bridge是四个不同的布尔条件，不把contrary当集合补集。

```python
print('with bridge:', opposite_transfer(True,True,True,True))
print('without bridge:', opposite_transfer(True,True,True,False))
print('no seed cycle:', {k: sorted(v) for k,v in sorted(minimal_supports([], [(['a'],'b'),(['b'],'a')]).items())})
```
