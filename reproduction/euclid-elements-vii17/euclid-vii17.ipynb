# 欧几里得 VII.17：同乘数保持比

```python
from scale_ratio import scale, compare_products
from fractions import Fraction
t = scale(3,4,6)
print(t)
```

## 从原数及两个积直接核对

```python
d,e = t.products
if not Fraction(4,d) == Fraction(6,e) == Fraction(*t.internal_ratio) or not Fraction(4,6) == Fraction(d,e) == Fraction(*t.crossed_ratio):
    raise RuntimeError('ratio mismatch')
print('internal:', Fraction(4,d), 'crossed:', Fraction(d,e))
```

## 改变其中一个乘数

```python
try:
    compare_products(3,4,6,12,24)
except ValueError as exc:
    print('rejected:', exc, '; ratios:', Fraction(4,6), Fraction(12,24))
else:
    raise RuntimeError('unexpected acceptance')
```
