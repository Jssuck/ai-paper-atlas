这是原创有限 Horn 教学模型，不是古代科学统一性的判定算法。先进入本目录再运行。标准库逐单元执行不等于 Jupyter 内核验收。

```python
from model import derive, check_certificate, minimal_supports, typed_rules, shares_root
rules = [(('p',),'u'),(('q',),'v'),(('u','v'),'w')]
known, certificate = derive({'p','q'}, rules)
print('closure:', sorted(known))
print('certificate valid:', check_certificate({'p','q'}, rules, certificate, 'w'))
```

## 1 支持与共享逻辑
通用规则自身不是事实。最小支持是相对于给定根和规则的包含最小集合，不是最短证明。

```python
print('no roots:', sorted(derive(set(), rules)[0]))
print('minimal supports:', [sorted(s) for s in minimal_supports('w', {'p','q'}, rules)])
print('same label, no rule:', 'b' in derive({'a'}, [])[0])
```

## 2 类型证书与桥
类型标签由建模者声明，不能当作已经发现自然之类。显式许可桥也不是桥的经验真实性证明。

```python
genus = {'p':'N','u':'N','q':'G','v':'G','w':'G'}
plain, rejected = typed_rules(rules, genus)
bridged, _ = typed_rules(rules, genus, [(('u','v'),'w')])
print('without bridge:', 'w' in derive({'p','q'}, plain)[0])
print('with declared bridge:', 'w' in derive({'p','q'}, bridged)[0])
print('rejected:', [(sorted(b), h) for b,h in rejected])
```

## 3 交叠不是学科等价关系
A与B共享p，B与C共享q，但A与C没有共同根。不能简单把“有一个共同根”当成同一学科的等价关系。

```python
A,B,C = {'p'}, {'p','q'}, {'q'}
print('AB, BC, AC:', shares_root(A,B), shares_root(B,C), shares_root(A,C))
from model import exhaustive
print(exhaustive())
```
