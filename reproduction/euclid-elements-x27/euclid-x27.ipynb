# X.27：从仅平方可公度的输入构造有理面积

```python
from fractions import Fraction as F
from ratio_witness import construct
for a,b in [(2,3),(3,2),(F(2,3),F(8,5))]:
    w=construct(a,b)
    print(f'a={a}, b={b}:',w)
    print('product fourth:',w.c_fourth*w.d_fourth,w.rectangle_area**4)
```
