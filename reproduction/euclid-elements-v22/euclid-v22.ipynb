# V.22：顺次同比与任取等倍

正有理数教学例。请在本目录按顺序运行。三次探测并非全称证明。

```python
from fractions import Fraction
from ratio_chains import compare_scaled_endpoints, magnitudes

x = magnitudes((8, 4, 2, 1), "length")
y = magnitudes((12, 6, 3, Fraction(3, 2)), "area")
print(x, y)
```

## 让末项分别落在首项的两侧及相等处
中间各项的倍数不需要等于首末倍数。

```python
expected = ((1, 1), (0, 0), (-1, -1))
results = tuple(compare_scaled_endpoints(x, y, (1, 5, 3, n)) for n in (7, 8, 9))
if results != expected:
    raise AssertionError((results, expected))
print(results)
```

## 反例：首尾可以碰巧同比，相邻条件仍可能失效
换掉中项，不要仅凭一次首尾比较接收前提。

```python
bad = magnitudes((16, 9, 2), "area")
try:
    compare_scaled_endpoints(magnitudes((8, 4, 1), "length"), bad, (1, 1, 1))
except ValueError as error:
    print(error)
else:
    raise AssertionError("expected rejection")
```
