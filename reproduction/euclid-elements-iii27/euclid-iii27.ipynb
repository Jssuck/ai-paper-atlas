## 1. 任意起点不改变选定弧扫角
预期：长度前提及两项同类角结论均为真。

```python
from fractions import Fraction as Q
from arcs import Arc, copy_smaller, inspect

a = Arc(Q(3), Q(0), Q(1, 5))
b = Arc(Q(3), Q(9, 10), Q(1, 10))
report = inspect(a, b)
if report != dict(premises=True, central=True, inscribed=True):
    raise AssertionError(report)
print(report)
```

## 2. 构造严格子弧
即使大弧穿过零方向，复制的小弧也是它的严格初始子弧。预期：复制弧等长于原小弧，但短于大弧。

```python
large = Arc(Q(2), Q(11, 12), Q(1, 6))
small = Arc(Q(2), Q(1, 2), Q(7, 12))
copied = copy_smaller(small, large)
inside = large.contains_interior(copied.end)
lengths = (small.length_over_tau, copied.length_over_tau, large.length_over_tau)
if not (inside and lengths[0] == lengths[1] < lengths[2]):
    raise AssertionError(lengths)
print("长度除以一整周的弧度数：", *(str(x) for x in lengths))
```

## 3. 练习与答案
半径为 2、扫角为 1/4 转的弧，和半径为 1 的哪条弧等长？能推出圆心角相等吗？先写出答案，再执行下一格。

```python
x = Arc(Q(2), Q(0), Q(1, 4))
y = Arc(Q(1), Q(0), Q(1, 2))
if not (x.length_over_tau == y.length_over_tau and x.sweep != y.sweep):
    raise AssertionError("counterexample not reproduced")
print("答：半圆；不能。半径不相等，两个扫角分别为 1/4 与 1/2 转。")
```
