## 第四比例量

长度组 2,3，面积组 4,6，两组各自同类。第四比例量恰好等于右组原底量 6。

```python
a,b = Magnitude(2),Magnitude(3)
c,d = Magnitude(4,'area'),Magnitude(6,'area')
trace = replay(a,b,c,d)
print('candidate and remaining area:', trace.candidate.value, trace.remainder.value)
if trace.candidate.value != Fraction(6):
    raise ArithmeticError('unexpected candidate')
```

## 整倍比较的边界

若后项倍率不超过前项倍率，就不调用余比的比较；若超过，则用正的倍率差。

```python
for k in range(2,7):
    r = direct_probe(a,b,c,d,3,k)
    print(k,r.joint_signs,r.separated_signs)
```
