# 欧几里得 VII.18：分别乘同一个数

```python
from common_multiplicand import check_products, transport
from fractions import Fraction
t = transport(4,6,3)
print(t)
```

## 直接从原输入核对乘积和比

```python
d,e = t.products
if (d,e) != (4*3,6*3) or Fraction(d,e) != Fraction(4,6) or Fraction(*t.ratio) != Fraction(4,6):
    raise RuntimeError('comparison mismatch')
print('product ratio:', Fraction(d,e))
```

## 错误的共同被乘数，保比也不能补回原前提

```python
print('same ratio but wrong scale:', Fraction(16,24))
try:
    check_products(4,6,3,16,24)
except ValueError as exc:
    print('rejected:', exc)
else:
    raise RuntimeError('unexpected acceptance')
```
