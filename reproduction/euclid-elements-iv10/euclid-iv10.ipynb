## 两圆与切线

切线垂直于小圆在 D 的半径。D 的上下两支应给出相同判定。

```python
for branch in (1, -1):
    checks = construct(F(7, 5), branch=branch).checks()
    print(branch, checks)
    if not all(checks.values()):
        raise ValueError(checks)
```
