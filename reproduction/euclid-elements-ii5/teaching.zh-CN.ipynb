## 输入与范围
本Notebook只检查右半内切的有理坐标模型。输入t=CD/CB严格在0与1之间。有限实跑不证明一般命题。

```python
from fractions import Fraction as Q
from geometry import construct, verify
c = construct()
a = verify(c)
print("A-C-D-B; t=2/5")
print(c["points"])
```

## 余形及中间矩形
先比较CH与HF，再分别加DM得到CM与DF。

```python
for name in ("CH", "HF", "DM", "CM", "DF", "AL"):
    print(name, a[name])
```

## 磬折形与补方
AH与磬折形等积，再加LG补满平方。

```python
print("AH", a["AH"], "gnomon", a["gnomon"])
print(a["AH"], "+", a["LG"], "=", a["square"])
```

## 分数实例
坐标单位改变不改变推理链。

```python
z = verify(construct(u=(1,0), v=(0,1), t=Q(2,7)), u=(1,0), v=(0,1), t=Q(2,7))
print(z["AH"], "+", z["LG"], "=", z["square"])
```

## 拒绝伪证书
等面积不够；点H必须处在原对角线上。

```python
bad = construct()
bad["points"]["H"] = (Q(2),Q(2))
try:
    verify(bad)
except ValueError as e:
    print("rejected:", str(e))
else:
    raise RuntimeError("tampering accepted")
```
