# 欧几里得 VII.15：整除次数的交换

```python
from unit_measures import measure
from fractions import Fraction
t = measure(3,4,12)
print('Y cuts:', t.y_to_z_cuts, 'X cuts:', t.x_to_z_cuts)
```

## 直接用精确分数核对

```python
before, after = Fraction(t.z,t.y), Fraction(t.z,t.x)
if (before,after) != (t.x,t.y):
    raise RuntimeError('measure mismatch')
print('two counts:', before,after)
```

## 少一条整除前提，以及单位边界

```python
try:
    measure(3,4,13)
except ValueError as exc:
    print('rejected:', exc)
else:
    raise RuntimeError('unexpected acceptance')
print('modern X=1:', measure(1,4,4,modern=True).x_to_z_cuts)
```
